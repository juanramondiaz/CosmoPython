### A.1. Flat FLRW Metric

---

The FLRW metric for a spatially flat universe (curvature $ k = 0 $) is:
$$ds^2 = -dt^2 + a(t)^2 (dx^2 + dy^2 + dz^2),$$
where $ a(t) $ is the scale factor, and the Hubble parameter is $ H = \dot{a}/a $, with the dot denoting derivative with respect to $ t $.


### A.2. Action and Energy-Momentum Tensor

---

The model considers a homogeneous scalar field $ \phi(t) $ (inflaton) with potential $ V(\phi) $. The action is:
$$S = \int d^4x \sqrt{-g} \left[ \frac{M_{Pl}^2}{2} R - \frac{1}{2} \partial_\mu \phi \partial^\mu \phi - V(\phi) \right],$$
where $ R $ is the scalar curvature.
Varying with respect to the metric, we obtain Einstein’s equations:
$$G_{\mu\nu} = \frac{1}{M_{Pl}^2} T_{\mu\nu},$$
with the scalar field energy-momentum tensor:
$$T_{\mu\nu} = \partial_\mu \phi \partial_\nu \phi - g_{\mu\nu} \left( \frac{1}{2} \partial^\lambda \phi \partial_\lambda \phi + V(\phi) \right).$$
For $ \phi = \phi(t) $ (homogeneous), the relevant components are:

* Energy density: $ \rho = T_{00} = \frac{1}{2} \dot{\phi}^2 + V(\phi) $,
* Pressure: $ p = \frac{1}{3} T_{ii}/a^2 = \frac{1}{2} \dot{\phi}^2 - V(\phi) $ (for $ i=1,2,3 $).

**Note**: The field behaves as a perfect fluid with equation of state $ w = p/\rho $, which varies from $ w \approx -1 $ (potential-dominated) to $ w \approx 1 $ (kinetic-dominated).


### A.3. Friedmann Equations

---

Einstein’s equations in the FLRW metric yield the Friedmann equations. The first one (00 component):
$$H^2 = \left( \frac{\dot{a}}{a} \right)^2 = \frac{1}{3 M_{Pl}^2} \rho = \frac{1}{3 M_{Pl}^2} \left( \frac{1}{2} \dot{\phi}^2 + V(\phi) \right).$$
The second equation (acceleration equation, combining components):
$$\frac{\ddot{a}}{a} = \dot{H} + H^2 = -\frac{1}{6 M_{Pl}^2} (\rho + 3p).$$
Substituting $ \rho + 3p = 2\dot{\phi}^2 - 2V(\phi) $:
$$\dot{H} + H^2 = -\frac{1}{6 M_{Pl}^2} (2\dot{\phi}^2 - 2V) = -\frac{1}{3 M_{Pl}^2} (\dot{\phi}^2 - V).$$
To obtain $ \dot{H} $ directly, we use the continuity equation (from conservation of $ T^{\mu\nu} $):
$$\dot{\rho} + 3H (\rho + p) = 0.$$
Substituting $ \rho + p = \dot{\phi}^2 $:
$$\dot{\rho} = -3H \dot{\phi}^2.$$
Now, differentiate the first Friedmann equation:
$$2 H \dot{H} = \frac{1}{3 M_{Pl}^2} \dot{\rho} = \frac{1}{3 M_{Pl}^2} (-3H \dot{\phi}^2) = -\frac{H \dot{\phi}^2}{M_{Pl}^2}.$$
Dividing by $ 2H $ (assuming $ H \neq 0 $):
$$\dot{H} = -\frac{\dot{\phi}^2}{2 M_{Pl}^2}.$$


### A.4. Klein-Gordon Equation for the Inflaton

---

Varying the action with respect to $ \phi $, we obtain the equation of motion:
$$\Box \phi + V'(\phi) = 0,$$
where $ \Box = \nabla_\mu \nabla^\mu $ and $ V' = dV/d\phi $.
In the FLRW metric, for homogeneous $ \phi(t) $:
$$\ddot{\phi} + 3 H \dot{\phi} + V'(\phi) = 0.$$
(The term $ 3H \dot{\phi} $ arises from the expansion factor, acting as Hubble friction.)

$$H^2 = \frac{1}{3 M_{Pl}^2} \left( \frac{1}{2} \dot{\phi}^2 + V(\phi) \right), \quad \dot{H} = -\frac{\dot{\phi}^2}{2 M_{Pl}^2}, \quad \ddot{\phi} + 3 H \dot{\phi} + V'(\phi) = 0.$$


### A.5. Slow-Roll Approximation

---

Inflation requires accelerated expansion ($ \ddot{a} > 0 $), i.e., $ \dot{H} + H^2 > 0 $ or $ \epsilon < 1 $, where we define the slow-roll parameters:
$$\epsilon = -\frac{\dot{H}}{H^2}, \quad \eta = -\frac{\ddot{\phi}}{H \dot{\phi}}.$$
(There are equivalent definitions in terms of the potential.)
In slow-roll, we assume the field "rolls slowly": $ \dot{\phi}^2 \ll V(\phi) $ (potential dominance) and $ |\ddot{\phi}| \ll |3 H \dot{\phi}|, |V'| $ (small second derivative).
From the second Friedmann equation:
$$\epsilon = -\frac{\dot{H}}{H^2} = \frac{\dot{\phi}^2}{2 M_{Pl}^2 H^2} = \frac{3 \dot{\phi}^2}{2 (\dot{\phi}^2 + 2V)} \approx \frac{3 \dot{\phi}^2}{4V} \ll 1.$$
Approximating the equations:

From Klein-Gordon: $ 3 H \dot{\phi} + V' \approx 0 $ ⇒ $ \dot{\phi} \approx - \frac{V'}{3H} $.
From the first Friedmann equation: $ H^2 \approx \frac{V}{3 M_{Pl}^2} $.

Substituting:
$$\dot{\phi}^2 \approx \frac{(V')^2}{9 H^2} \approx \frac{(V')^2}{3} \cdot \frac{3 M_{Pl}^2}{V} = M_{Pl}^2 \frac{(V')^2}{V}.$$
Then:
$$\epsilon \approx \frac{M_{Pl}^2}{2} \left( \frac{V'}{V} \right)^2, \quad \eta \approx M_{Pl}^2 \left( \frac{V''}{V} - \frac{1}{2} \left( \frac{V'}{V} \right)^2 \right).$$
The number of e-folds $ N = \int H dt \approx -\frac{1}{M_{Pl}^2} \int_{\phi_i}^{\phi_f} \frac{V}{V'} d\phi $, and inflation ends when $ \epsilon \approx 1 $.
These approximations allow computation of observables such as the spectral index $ n_s \approx 1 - 6\epsilon + 2\eta $ and the tensor-to-scalar ratio $ r \approx 16\epsilon $, which can be compared with CMB data.


| Magnitude    | Formula (slow-roll)                               | Interpretation                          |
| ------------ | ------------------------------------------------- | --------------------------------------- |
| $\epsilon_V$ | $\frac{1}{2}\bar M_{\rm Pl}^2 (V'/V)^2$           | measures the slope of the potential     |
| $\eta_V$     | $\bar M_{\rm Pl}^2 V''/V$                         | measures the curvature of the potential |
| $n_s$        | $1 - 6\epsilon_V + 2\eta_V$                       | scalar spectral index                   |
| $r$          | $16\epsilon_V$                                    | tensor-to-scalar ratio                  |
| Evaluate at  | $\phi_*$ when $N_{\text{end}} - N_* \simeq 50-60$ | horizon exit of the pivot mode          |

### B.1. Quadratic Potential (Quad) - Chaotic Inflation

---

**Quadratic Potential**$$V(\phi) = \frac{1}{2} m^2 \phi^2$$
Derivative:
$$V'(\phi) = m^2 \phi$$

Physical Description

* **Model**: Chaotic inflation, proposed by Linde (1983).
* **Characteristics**:
  * Simple "large-field" type potential, where $ \phi \gg M_{Pl} $ initially.
  * Inflation occurs when the potential term dominates ($ \frac{1}{2} \dot{\phi}^2 \ll V(\phi) $).
  * Parameter $ m $ (inflaton mass) is tuned to $ m \approx 6 \times 10^{-6} M_{Pl} $ to match the amplitude of CMB fluctuations ($ A_s \approx 2.1 \times 10^{-9} $).

**Slow-Roll Parameters:**

* $ \epsilon_v \approx \frac{M_{Pl}^2}{2} \left( \frac{V'}{V} \right)^2 = \frac{2 M_{Pl}^2}{\phi^2} $
* $ \eta_v \approx M_{Pl}^2 \frac{V''}{V} = \frac{2 M_{Pl}^2}{\phi^2} $
* Number of e-folds: $ N \approx \frac{\phi_0^2 - \phi_f^2}{4 M_{Pl}^2} $, with end of inflation at $ \phi_f \approx \sqrt{2} M_{Pl} $.
* Observables: $ n_s \approx 1 - \frac{4 M_{Pl}^2}{\phi_0^2} $, $ r \approx \frac{32 M_{Pl}^2}{\phi_0^2} $.
* Expected values for $ \phi_0 \approx 15 M_{Pl} $, $ n_s \approx 0.964 $, $ r \approx 0.142 $.

**Comments**

* Advantages: Simple to implement and analyze. Well studied.
* Limitations: The value of $ r $ is high compared with Planck 2018 limits ($ r < 0.036 $), making it marginally disfavored.
* Use in Code: $ m = \exp(\log m) $ ensures $ m $ is positive, useful for numerical optimization.



### B.2. Quartic Potential (Quartic)

---

**Quartic Potential**
$$V(\phi) = \frac{1}{4} \lambda \phi^4$$
**Derivative:**
$$V'(\phi) = \lambda \phi^3$$
**Physical Description**

* **Model**: Another example of "large-field" chaotic inflation.
* **Characteristics:**
  * Higher-power potential ($ \phi^4 $), steeper than the quadratic case.
  * $ \lambda $ is a dimensionless coupling, typically $ \lambda \approx 10^{-13} $ to match $ A_s $ with CMB data.
  * Produces a larger contribution of tensor modes (gravitational waves).

**Slow-Roll Parameters:**

* $ \epsilon_v \approx \frac{M_{Pl}^2}{2} \left( \frac{V'}{V} \right)^2 = \frac{8 M_{Pl}^2}{\phi^2} $
* $ \eta_v \approx M_{Pl}^2 \frac{V''}{V} = \frac{6 M_{Pl}^2}{\phi^2} $
* End of inflation: $ \phi_f \approx 2 \sqrt{2} M_{Pl} $.
* $ N \approx \frac{\phi_0^2 - \phi_f^2}{8 M_{Pl}^2} $.
* Observables: $ n_s \approx 1 - \frac{6 M_{Pl}^2}{\phi_0^2} $, $ r \approx \frac{64 M_{Pl}^2}{\phi_0^2} $.
* For $ \phi_0 \approx 15 M_{Pl} $, $ n_s \approx 0.973 $, $ r \approx 0.284 $.

**Comments**

* Advantages: Similar to the quadratic case, but produces a larger $ r $, useful for studying tensions with gravitational-wave constraints.
* Limitations: More strongly disfavored by Planck due to its high $ r $.



### 1.2.3. Hilltop Potential

---

**Hilltop Potential**
$$V(\phi) = V_0 \left( 1 - \left( \frac{\phi}{\mu} \right)^p \right)^2$$
**Derivative:**
$$V'(\phi) = -2 V_0 \left( 1 - \left( \frac{\phi}{\mu} \right)^p \right) \frac{p \phi^{p-1}}{\mu^p}$$

* Maximum at $\phi \approx 0$ (for even $p$). Inflation occurs near the top.
* Typical parameters: $p=2, ,4;,\mu\lesssim \mathcal{O}(M_{Pl})$

**Physical Description**

* **Model**: "Small-field" inflation, where $ \phi \ll M_{Pl} $, and the field rolls from near the maximum of the potential ($ \phi \approx 0 $).
* **Characteristics:**
  * $ V_0 $: Energy scale of the potential.
  * $ \mu $: Field scale that determines the width of the potential.
  * $ p $: Exponent controlling the shape (typically $ p = 2, 4 $).
  * Inflation occurs near $ \phi = 0 $, where the potential is nearly flat.

**Slow-Roll Parameters:**

Depend on $ \phi/\mu $ and $ p $. For $ p = 2 $:

* $ \epsilon_v \approx \frac{M_{Pl}^2}{2} \left( \frac{2 p (\phi/\mu)^{p-1}}{\mu (1 - (\phi/\mu)^p)} \right)^2 $
* $ \eta_v $: More complex, but dominated by $ V'' $.
* $ N \approx \frac{V_0}{M_{Pl}^2} \int_{\phi_f}^{\phi_0} \frac{d\phi}{V'(\phi)} $, evaluated numerically.
* Can produce small $ r $, more compatible with Planck.

**Comments**

* Advantages: Hilltop models are favored by recent data due to their low $ r $.
* Useful for studying inflation in small-field regions.
* Limitations: Analytical calculations are more complex; requires accurate numerical simulation.


### B.4. $\alpha$-attractor Potential (Starobinsky-like)

---

**Alpha Potential (Starobinsky-like)**
$$V(\phi) = V_0 \left( 1 - e^{-\sqrt{\frac{2}{3\alpha}} \phi} \right)^2$$
**Derivative:**
$$V'(\phi) = 2 V_0 \left( 1 - e^{-\sqrt{\frac{2}{3\alpha}} \phi} \right) \sqrt{\frac{2}{3\alpha}} e^{-\sqrt{\frac{2}{3\alpha}} \phi}$$
**Physical Description**

* **Model**: Inspired by Starobinsky-type inflation or $\alpha$-attractor models, common in string theory.

* **Characteristics:**

  * $ V_0 $: Energy scale.

  * $ \alpha $: Parameter controlling the curvature of the potential (typically $ \alpha \sim 1 $).

  * The exponential term makes the potential flat for $ \phi \gg \sqrt{\frac{3\alpha}{2}} M_{Pl} $, favoring inflation.

**Slow-Roll Parameters:**

* $ \epsilon_v \approx \frac{4 M_{Pl}^2}{3\alpha (e^{\sqrt{\frac{2}{3\alpha}} \phi} - 1)^2} $
* $ \eta_v $: Negative and small for large $ \phi $.
* $ N \approx \frac{3\alpha}{4 M_{Pl}^2} (e^{\sqrt{\frac{2}{3\alpha}} \phi_0} - e^{\sqrt{\frac{2}{3\alpha}} \phi_f}) $.
* Produces very small $ r $, compatible with observations.

**Comments**

* Advantages: Strongly favored by data (low $ r $), relevant for modern models.
* Limitations: Requires numerical care due to the exponential term.

```python
def get_Params(units=UNITS, potential=POTENTIAL, CMB_fit = True):
    """
    Computes the physical constants for cosmic inflation according to the unit system.
    
    Args:
        units (str): Unit system ('NATURAL' or 'SI').
        c (float, optional): Speed of light (m/s). Used only in SI.
        hbar (float, optional): Reduced Planck constant (J·s). Used only in SI.
    
    Returns:
        dict: Dictionary with physical constants:
            - M_Pl: Reduced Planck mass
            - m: Inflaton mass (for quadratic potential)
            - lambda_: Coupling (for quartic potential)
            - V0: Energy scale (for hilltop and alpha potentials)
            - alpha: Parameter for alpha potential
            - mu: Parameter for hilltop potential
            - p: Parameter for hilltop potential
            - G: Gravitational constant
    
    Raises:
        ValueError: If the units are not valid.
    """
    
    # Fundamental physical constants (in SI)
    c = 2.99792458e8  # m/s
    hbar = 1.0545718e-34  # J·s
    G = 6.67430e-11  # m³ kg⁻¹ s⁻²
    GeV = 1.60217662e-10 # J (Unit conversion factor GeV --> J)
   
    if units == 'SI':
        argument = hbar * c / (8 * torch.pi * G)
        M_Pl = torch.sqrt(torch.as_tensor(argument))  # kg
            
    if units == 'NATURAL':        # In natural units: ħ = c = 1
        # Conversion factors. Values in GeV are used for consistency with cosmology literature
        hbar_GeV  = hbar/GeV       # m^3 Kg^-1 s^-2 --> GeV s
        metro_GeV = GeV/(hbar*c)   # meter --> GeV^-1
        Kg_GeV    = c**2/GeV       # Kg --> GeV
        seg_GeV   = GeV/hbar    # Seconds -->  GeV^-1
        # Constant G in GeV^-2 units
        G = G*metro_GeV**3/(Kg_GeV*seg_GeV**2)  # GeV^-2
        # Redefine  ħ = c = 1
        c = 1                       # Natural Units
        hbar = 1                    # Natural Units
        argument = hbar * c / (8 * torch.pi * G)
        M_Pl = torch.sqrt(torch.as_tensor(argument))  # GeV
        
    elif units == 'PLANCK':
        # In Planck units: M_Pl = 1
        hbar = 1
        c = 1        
        G = 1
        M_Pl = 1
    
    # Planck energy
    E_Pl = M_Pl * c**2
    
    # Inflaton mass 
    m_natural = 1.0
    if CMB_fit:  # m ≈ 6e-6 M_Pl fit to compare with CMB data, A_s ≈ 2.1e-9)
        m_natural = m_natural * 6e-6
    m = m_natural * M_Pl  # kg
    
    
    # Coupling λ (dimensionless, adjusted for quartic potential)
    lambda_ = 1e-13  # Dimensionless
        
    # Energy scale V₀ (adjusted for small-field models)
    # V₀ ≈ 1e-10 M_Pl⁴ in natural units
    V0_natural = 1e-10  # Relation V₀/M_Pl⁴
    # In SI: V₀ = V0_natural × (M_Pl c²)⁴ / (ħ c)³ for J/m³
    V0 = V0_natural * (M_Pl*(c**2))**4  /((hbar*c)**3)   # J/m³ (in SI)
    
    # Parameter mu used in the Hilltop potential
    mu = 5.0*M_Pl
    # Parameter α (dimensionless)
    alpha = 1.0  # Dimensionless
    
    param = {
        'hbar': torch.as_tensor(hbar),      # J·s
        'c': torch.as_tensor(c),            # m/s
        'G': torch.as_tensor(G),            # m³ kg⁻¹ s⁻²
        'M_Pl': torch.as_tensor(M_Pl),      # kg
        'E_Pl': torch.as_tensor(E_Pl),      # J
        'm': torch.as_tensor(m),            # 
        'lambda_': torch.as_tensor(lambda_),# dimensionless
        'V0': torch.as_tensor(V0),          # 
        'mu': torch.as_tensor(mu),          # 
        'p': torch.as_tensor(5.0),          # dimensionless 
        'alpha': torch.as_tensor(alpha),    # dimensionless
        'units': units,    # Unit system
        'potential': potential,  # Potential type
        'use_epsilon_v':False    
    }        
   
    return param

def print_param(param):
    """
    Prints the physical constants in a readable format.
    
    Args:
        constants (dict): Dictionary of physical constants.
        units (str): Unit system used.
    """
    print(f"\n=== Physical Constants ({param['units']}) and Parameters\n------------------------------------------")
    for key, value in param.items():
        if key == 'hbar':
            print(f"{key}:  {value:.3e}  {'J·s' if param['units'] == 'SI' else 'hbar'}")
        elif key == 'c': 
            print(f"{key}:  {value:.3e}  {'m/s' if param['units'] == 'SI' else 'c'}")
        elif key == 'G':
            print(f"{key}:  {value:.3e}  {'m³ kg⁻¹ s⁻²' if param['units'] == 'SI' else 'GeV⁻²'}")
        elif key == 'E_Pl':
            print(f"{key}:  {value:.3e}  {'J' if param['units'] == 'SI' else 'GeV'}")
        elif key in ['M_Pl', 'm', 'mu']:
            print(f"{key}:  {value:.3e}  {'kg' if param['units'] == 'SI' else 'GeV'}")
        elif key == 'V0':
            print(f"{key}:  {value:.3e}  {'J/m³' if param['units'] == 'SI' else 'GeV⁴'}")
        elif key in ['lambda_', 'alpha', 'p']:
            print(f"{key}:  {value} (dimensionless)")
        else:
            print(f"{key}:  {value}")
        
UNITS = 'SI'
param = get_Params(units=UNITS, potential= POTENTIAL, CMB_fit = True)
print_param(param)
    
UNITS = 'NATURAL'
param = get_Params(units=UNITS, potential= POTENTIAL, CMB_fit = True)
print_param(param)

UNITS = 'PLANCK'
param = get_Params(units=UNITS, potential= POTENTIAL, CMB_fit = False)
print_param(param)
```

### 1.2. DeepLearning VNet Potencial
---

```python
# === Potential neural network ===
class VNet(nn.Module):
    def __init__(self, hidden=128, depth=4, Vmin=0.0, act="silu"):
        super().__init__()
        acts = {
            "silu": nn.SiLU(),
            "tanh": nn.Tanh(),
            "gelu": nn.GELU(),
        }
        activation = acts.get(act, nn.SiLU())

        layers = [nn.Linear(1, hidden), activation]
        for _ in range(depth - 1):
            layers += [nn.Linear(hidden, hidden), activation]
        layers += [nn.Linear(hidden, 1)]  # scalar output f_θ(φ)

        self.net = nn.Sequential(*layers)
        self.softplus = nn.Softplus(beta=1.0)  # β=1 -> smooth and stable shape
        # Non-trainable Vmin (if you want to fix an offset ≥0)
        self.register_buffer("Vmin", torch.tensor(Vmin, dtype=torch.get_default_dtype()))

    def forward(self, phi):
        """
        phi: tensor of shape (N,1) or (1,1)
        return: V(phi) with shape (N,1)
        """
        f = self.net(phi)
        return self.softplus(f) + self.Vmin

# === Load trained weights ===
checkpoint = torch.load("./V_inflaton_DL.pt", map_location="cpu")
VNet_model = VNet(
    hidden=checkpoint['config']['hidden'],
    depth=checkpoint['config']['depth'],
    Vmin=checkpoint['config']['Vmin'],
    act=checkpoint['config']['act'],
    #fourier_features=checkpoint['config']['fourier_features'],
    #n_freqs=checkpoint['config']['n_freqs'],
    #spectral_norm=checkpoint['config']['spectral_norm']
)
VNet_model.load_state_dict(checkpoint['model_state_dict'])
VNet_model = VNet_model.double()
VNet_model.eval()
```

```python
def V_VNet(phi):
    """
    Neural potential V(φ) as a torch tensor.
    φ can be a float or a 1D tensor.
    """
    if not torch.is_tensor(phi):
        phi = torch.tensor([[phi]], dtype=torch.float64)
    else:
        phi = phi.reshape(-1, 1).to(torch.float64)
    V = VNet_model(phi).item()
    return torch.tensor(V)

def dV_VNet(phi):
    """
    First derivative V'(φ) using autograd.
    """
    phi_t = torch.tensor([[phi]], dtype=torch.float64, requires_grad=True)
    V = VNet_model(phi_t)
    Vp = autograd.grad(V, phi_t, grad_outputs=torch.ones_like(V),
                       create_graph=False, retain_graph=False)[0]
    return torch.tensor(Vp.detach().cpu()[0].item())

def d2V_VNet(phi):
    """
    Second derivative V''(φ).
    """
    phi_t = torch.tensor([[phi]], dtype=torch.float64, requires_grad=True)
    V = VNet_model(phi_t)
    Vp = autograd.grad(V, phi_t, grad_outputs=torch.ones_like(V),
                       create_graph=True, retain_graph=True)[0]
    Vpp = autograd.grad(Vp, phi_t, grad_outputs=torch.ones_like(Vp),
                        create_graph=False, retain_graph=False)[0]
    return torch.tensor(Vpp.detach().cpu()[0].item())
```

### 1.3. Potential and derivative
---


```python
def V(phi, param):
    """
    Computes the potential V(phi) for a scalar field given the potential type.
    
    Args:
        phi (torch.Tensor): Value of the scalar field.
        param(dict): Parameter dictionary
    
    Returns:
        torch.Tensor: Value of the potential V(phi).
    
    Raises:
        ValueError: If the potential type is not valid.
    """
    pot = param['potential']
    phi = torch.as_tensor(phi)
    if pot == 'quad':
        m = param['m']
        return 0.5 * m**2 * phi**2
    elif pot == 'quartic':
        lam = param['lambda_']
        return 0.25 * lam * phi**4
    elif pot == 'six':
        lam = param['lambda_']
        return (1/6) * lam * phi**6
    elif pot == 'hilltop':
        V0 = param['V0']
        mu = param['mu']
        p  = param['p']
        x  = (phi / mu).pow(p)
        return V0 * (1.0 - x)**2
    elif pot == 'alpha':
        V0    = param['V0']
        alpha = max(param['alpha'], 1e-12)
        s     = torch.sqrt(torch.tensor(2.0) / (torch.tensor(3.0) * alpha))
        return V0 * (1.0 - torch.exp(-s * phi))**2
    elif pot == 'VNet':
        return V_VNet(phi)
    else:
        raise ValueError("Unknown potential")

def dV_dphi(phi, param):
    """
    Computes the derivative of the potential V'(phi) = dV/dphi.
    
    Args:
        phi (torch.Tensor): Value of the scalar field (in units of M_Pl).
        param(dict): Parameter dictionary
    
    Returns:
        torch.Tensor: Value of the derivative V'(phi).
    
    Raises:
        ValueError: If the potential type is not valid.
    """
    # Potential type ('quad', 'quartic', 'hilltop', 'alpha').
    pot = param['potential']
    phi = torch.as_tensor(phi)
    if pot == 'quad':
        m = param['m']
        return (m**2) * phi
    elif pot == 'quartic':
        lam = param['lambda_']
        return lam * phi**3
    elif pot == 'six':
        lam = param['lambda_']
        return lam * phi**5
    elif pot == 'hilltop':
        V0 = param['V0']
        mu = param['mu']
        p  = param['p']
        # d/dphi [ V0 (1 - (phi/mu)^p)^2 ] = -2 V0 (1 - x) * p * phi^{p-1}/mu^p
        x  = (phi / mu).pow(p)
        return -2.0 * V0 * (1.0 - x) * p * (phi.pow(p-1)) / (mu.pow(p))
    elif pot == 'alpha':
        V0    = param['V0']
        alpha = max(param['alpha'], 1e-12)
        s     = torch.sqrt(torch.tensor(2.0) / (3.0 * alpha))
        e     = torch.exp(-s * phi)
        return 2.0 * V0 * (1.0 - e) * s * e
    elif pot == 'VNet':
        return dV_VNet(phi)
    else:
        raise ValueError("Unknown potential")

def d2V_dphi(phi, param):
    # Second derivative, useful for eta_V
    pot = param['potential']
    phi = torch.as_tensor(phi)
    if pot == 'quad':
        m = param['m']
        return (m**2) * torch.ones_like(torch.as_tensor(phi))
    elif pot == 'quartic':
        lam = param['lambda_']
        return 3.0 * lam * phi**2
    elif pot == 'hilltop':
        V0 = param['V0']
        mu = param['mu']
        p  = param['p']
        # d/dphi of the previous expression:
        # V' = -2 V0 (1 - x) p phi^{p-1} / mu^p, with x=(phi/mu)^p
        # V'' = -2 V0 [ (-x') p phi^{p-1}/mu^p + (1 - x) p (p-1) phi^{p-2}/mu^p ]
        x   = (phi / mu).pow(p)
        xp  = p * (phi.pow(p-1)) / (mu.pow(p))
        term1 = -xp * p * phi.pow(p-1) / (mu.pow(p))
        term2 = (1 - x) * p * (p-1) * phi.pow(p-2) / (mu.pow(p))
        return -2.0 * V0 * (term1 + term2)
    elif pot == 'alpha':
        V0    = param['V0']
        alpha = max(param['alpha'], 1e-12)
        s     = torch.sqrt(torch.tensor(2.0) / (3.0 * alpha))
        e     = torch.exp(-s * phi)
        # V' = 2 V0 s e (1 - e) = 2 V0 s (e - e^2)
        # V'' = 2 V0 s (-s e) - 4 V0 s (-s e) e = 2 V0 s^2 ( -e + 2 e^2 )
        return 2.0 * V0 * (s**2) * (-e + 2.0*e**2)
    elif pot == 'VNet':
        return d2V_VNet(phi)
    else:
        raise ValueError("Unknown potential")
```

### 2.1. Friedmann and Klein-Gordon Equations

---

Cosmological equations, including the Friedmann equation, the slow-roll parameters ($   \epsilon_H   $, $   \epsilon_V   $, $   \eta_V   $), the Klein-Gordon equation, and the condition for the end of inflation.

**1. FRIEDMANN EQUATION**
This function computes the Hubble parameter $ H $, which describes the expansion rate of the universe.

* **Energy density:**
  $$\rho = \frac{1}{2} \frac{\dot{\phi}^2}{c^2} + V(\phi)$$
  where:

$ \rho $: Energy density
$ \dot{\phi} $: Time derivative of the scalar field (phi_dot)
$ c $: Speed of light
$ V(\phi) $: Potential evaluated at $ \phi $ (computed by $ V(\phi, param) $)

* **Constant factor:**
  $$\text{factor}*H = \frac{\hbar}{c} \cdot \frac{1}{3 M*{\text{Pl}}^2}$$
  where:

$ \hbar $: Reduced Planck constant (hbar)
$ M_{\text{Pl}} $: Planck mass

* **Friedmann equation:**
  $$H = \sqrt{\text{factor}*H \cdot \rho} = \sqrt{\frac{\hbar}{c} \cdot \frac{1}{3 M*{\text{Pl}}^2} \cdot \left( \frac{1}{2} \frac{\dot{\phi}^2}{c^2} + V(\phi) \right)}$$

**2. Slow-Roll Parameter $ \epsilon_H $ (epsilon_H_Equation)**
This function computes the slow-roll parameter $ \epsilon_H $, which quantifies the rate of change of the Hubble parameter.

* Epsilon factor:
  $$\text{factor}*\epsilon = \frac{1}{2} \cdot \frac{\hbar}{c^3} \cdot \frac{1}{M*{\text{Pl}}^2}$$

* Slow-roll parameter $ \epsilon_H $:
  $$\epsilon_H = \text{factor}*\epsilon \cdot \frac{\dot{\phi}^2}{H^2} = \frac{1}{2} \cdot \frac{\hbar}{c^3} \cdot \frac{1}{M*{\text{Pl}}^2} \cdot \frac{\dot{\phi}^2}{H^2}$$

**3. Slow-Roll Parameter $ \epsilon_V $ (epsilon_V_Equation)**
This function computes the slow-roll parameter based on the potential, $ \epsilon_V $.

* **Potential and its derivative:**

  * $ V(\phi) $: Potential evaluated at $ \phi $ (computed by $ V(\phi, param) $)

  * $ \frac{dV}{d\phi} $: First derivative of the potential with respect to $ \phi $ (computed by $ dV_dphi(\phi, param) $)

* **Epsilon factor:**
  $$\text{factor}*\epsilon = \frac{1}{2} M*{\text{Pl}}^2$$

* **Slow-roll parameter $ \epsilon_V $:**
  $$\epsilon_V = \text{factor}*\epsilon \cdot \left( \frac{\frac{dV}{d\phi}}{V(\phi)} \right)^2 = \frac{1}{2} M*{\text{Pl}}^2 \cdot \left( \frac{\frac{dV}{d\phi}}{V(\phi)} \right)^2$$

**4. Slow-Roll Parameter $ \eta_V $ (eta_V_Equation)**
This function computes the second slow-roll parameter $ \eta_V $, related to the curvature of the potential.

* **Potential and its second derivative:**

  * $ V(\phi) $: Potential evaluated at $ \phi $

  * $ \frac{d^2V}{d\phi^2} $: Second derivative of the potential with respect to $ \phi $ (computed by $ d2V_dphi(\phi, param) $)

* **Eta factor:**
  $$\text{factor}*\eta = M*{\text{Pl}}^2$$

* **Slow-roll parameter $ \eta_V $:**
  $$\eta_V = \text{factor}*\eta \cdot \frac{\frac{d^2V}{d\phi^2}}{V(\phi)} = M*{\text{Pl}}^2 \cdot \frac{\frac{d^2V}{d\phi^2}}{V(\phi)}$$

**5. Equations of Motion (equations)**
This function defines the system of differential equations for the scalar field $ \phi $ and its time derivative $ \dot{\phi} $.

* **Input:**

  * $ y = [\phi, \dot{\phi}] $: State vector containing the scalar field and its first time derivative

  * $ \frac{dV}{d\phi} $: First derivative of the potential (computed by $ dV_dphi(\phi, param) $)

  * $ H $: Hubble parameter from the Friedmann equation

* **Klein-Gordon equation:**
  $$\ddot{\phi} = -3 H \dot{\phi} - c^2 \frac{dV}{d\phi}$$

* **System of equations:**
  $$\frac{d}{dt} \begin{bmatrix} \phi \ \dot{\phi} \end{bmatrix} = \begin{bmatrix} \dot{\phi} \ -3 H \dot{\phi} - c^2 \frac{dV}{d\phi} \end{bmatrix}$$

**6. End of Inflation Condition (end_inflation)**
This function defines the condition for the end of inflation, when the slow-roll parameter $ \epsilon $ reaches the value 1.

* **Slow-roll parameter:**
  $$\epsilon = \min(\epsilon_V, \epsilon_H)$$
  where:

  * $ \epsilon_V $: From epsilon_V_Equation

  * $ \epsilon_H $: From epsilon_H_Equation

* **End of inflation condition:**
  $$\epsilon - 1 = 0$$

  * Integration stops when $ \epsilon = 1 $.

  * The condition is configured to detect the transition from $ \epsilon < 1 $ to $ \epsilon > 1 $ (direction = 1).


```python
# Equations of motion (Klein-Gordon + Friedmann)

def friedmann_Equation(phi, phi_dot, param):
    hbar = param['hbar']
    c = param['c']
    M_Pl = param['M_Pl']
    
    V_phi = V(phi, param)                               # Potential at Phi
    energy_density = 0.5 * phi_dot**2/c**2 + V_phi      # Energy density
    factor_H = (hbar/c) /(3 * M_Pl**2)                  # Friedmann equation factor
    
    H = torch.sqrt(factor_H * energy_density)              # Friedmann equation

    return H

def epsilon_H_Equation(H, phi_dot, param):
    hbar = param['hbar']
    c = param['c']
    M_Pl = param['M_Pl']
    
    factor_epsilon = 0.5 *(hbar/c**3)/ M_Pl**2         # Epsilon_H parameter factor
    epsilon_H = factor_epsilon * phi_dot**2 / H**2       # Slow-roll parameter

    return epsilon_H

def epsilon_V_Equation( phi, param):
    hbar = param['hbar']
    c = param['c']
    M_Pl = param['M_Pl']
    
    V_phi = V(phi, param)                               # Potential at Phi
    dV_dphi_ = dV_dphi(phi, param)                      # Derivative of the potential with respect to Phi at Phi
    
    factor_epsilon = 0.5 * M_Pl**2                  # Epsilon_V parameter factor
        
    epsilon_V = factor_epsilon * (dV_dphi_/V_phi)**2    # Slow-roll parameter

    return epsilon_V

def eta_V_Equation( phi, param):
    hbar = param['hbar']
    c = param['c']
    M_Pl = param['M_Pl']
    
    V_phi = V(phi, param)                               # Potential at Phi
    d2V_dphi_ = d2V_dphi(phi, param)                    # Second derivative of the potential with respect to Phi at Phi
    
    factor_eta =  M_Pl**2                  # Epsilon_V parameter factor
        
    eta_V = factor_eta * (d2V_dphi_/V_phi)     # Eta_V parameter

    return eta_V

def equations(t, y, param):
    phi, phi_dot = y
    c = param['c']
    dV_dphi_ = dV_dphi(phi, param)                   # Derivative of the potential with respect to Phi at Phi
    
    H = friedmann_Equation(phi, phi_dot, param)      # Friedmann equation
    phi_ddot = -3 * H * phi_dot - (c**2)*dV_dphi_    # Klein-Gordon equation
    return [phi_dot, phi_ddot]

# Function to detect the end of inflation (epsilon = 1)
def end_inflation(t, y, param):
    phi, phi_dot = y
        
    H = friedmann_Equation(phi, phi_dot, param)          # Friedmann equation
    epsilon_V = epsilon_V_Equation(phi, param)    # Slow-roll parameter
    epsilon_H = epsilon_H_Equation(H, phi_dot, param)      # Slow-roll parameter
    epsilon =torch.min(torch.Tensor(epsilon_V), torch.Tensor(epsilon_H))
    return epsilon - 1  # Crosses 0 when epsilon = 1

end_inflation.terminal = True  # Stops integration when epsilon = 1 is reached
end_inflation.direction = 1    # Only detects crossing from <1 to >1
```

### 2.2. Initial Conditions

---

* Initial value of the field:$$ \phi_0=15,M_{Pl}, \quad \dot\phi_0 = \frac{-dV(\phi_0)}{3,M^2_{Pl}}$$
* Initial approximation of $ H $ using the Slow-Roll approximation: $$ H_0 = \sqrt{\frac{V(\phi_0}{3,M^2_{Pl}}}   $$-

Equations for $ \phi_0 $ according to the type of potential

**1. Quadratic Potential (quad)**
For the quadratic potential, the value of $ \phi_0 $ is computed from an analytical approximation based on the number of e-folds $ N_{\text{star}} $ and the value of $ \phi_{\text{end}} $ (end of inflation).

* **Value of $ \phi_{\text{end}} $:**
  $$\phi_{\text{end}} = \sqrt{2} M_{\text{Pl}}$$

* **Argument:**
  $$\text{argument} = 4 N_{\text{star}} + \left( \frac{\phi_{\text{end}}}{M_{\text{Pl}}} \right)^2$$

* **Initial value $ \phi_0 $:**
  $$\phi_0 = \sqrt{\text{argument}} \cdot M_{\text{Pl}} = \sqrt{4 N_{\text{star}} + \frac{\phi_{\text{end}}^2}{M_{\text{Pl}}^2}} \cdot M_{\text{Pl}}$$

**2. Quartic Potential (quartic)**
For the quartic potential, the computation of $ \phi_0 $ follows a similar form, adjusted for the fourth-order potential.

* **Value of $ \phi_{\text{end}} $:**
  $$\phi_{\text{end}} = 2 \sqrt{2} M_{\text{Pl}}$$

* **Argument:**
  $$\text{argument} = 8 N_{\text{star}} + \left( \frac{\phi_{\text{end}}}{M_{\text{Pl}}} \right)^2$$

* **Initial value $ \phi_0 $:**
  $$\phi_0 = \sqrt{\text{argument}} \cdot M_{\text{Pl}} = \sqrt{8 N_{\text{star}} + \frac{\phi_{\text{end}}^2}{M_{\text{Pl}}^2}} \cdot M_{\text{Pl}}$$

**3. Hilltop Potential (hilltop)**
For the hilltop potential, an analytical approximation near the maximum of the potential is used, based on the parameter $ \mu $.

* **Value of $ \phi_{\text{end}} $ (estimated with $ \epsilon_V \approx 1 $):**
  $$\phi_{\text{end}} = \left( \frac{\mu^8}{32 M_{\text{Pl}}^2} \right)^{1/6}$$

* **Denominator:**
  $$\text{denom} = 8 N_{\text{star}} M_{\text{Pl}}^2 + \frac{\mu^4}{\phi_{\text{end}}^2}$$

* **Initial value $ \phi_0 $:**
  $$\phi_0 = \frac{\mu^2}{\sqrt{\text{denom}}} = \frac{\mu^2}{\sqrt{8 N_{\text{star}} M_{\text{Pl}}^2 + \frac{\mu^4}{\phi_{\text{end}}^2}}}$$

**4. Alpha Potential (alpha)**
For the alpha potential, $ \phi_0 $ is computed using a logarithmic approximation that depends on the parameter $ \alpha $.

* **Value of $ \phi_{\text{end}} $:**
  $$\phi_{\text{end}} = \frac{4}{3 \alpha}$$

* **Parameter $ s $:**
  $$s = \sqrt{\frac{2}{3 \max(\alpha, 10^{-12})}}$$

* **Initial value $ \phi_0 $:**
  $$\phi_0 = \frac{1}{s} \ln \left( \frac{4 N_{\text{star}}}{3 \max(\alpha, 10^{-12})} \right)$$

**Notes:**

* $ N_{\text{star}} $: Number of e-folds, representing the amount of inflationary expansion.
* $ M_{\text{Pl}} $: Planck mass, a fundamental physical constant.
* $ \mu $: Hilltop potential parameter, related to the scale of the potential.
* $ \alpha $: Alpha potential parameter, which controls the shape of the potential.
* The term $ \max(\alpha, 10^{-12}) $ in the alpha potential avoids division by zero for small values of $ \alpha $.
* The equations use PyTorch (for example, torch.sqrt, torch.as_tensor, torch.log) for numerical calculations with tensor support.


```python
# Initial conditions
def get_Initial_Conditions(param, phi0=None, Nstar = 60, kappa_time = 100):
    m = param['m']
    MPl = param['M_Pl']
    
    if phi0 is None:
        if param['potential']=='quad':
            phi_end = torch.sqrt(torch.as_tensor(2.0))* MPl
            argument = 4.0*Nstar + (phi_end/MPl)**2
            phi0 = torch.sqrt(torch.as_tensor(argument)) * MPl
        
        elif param['potential'] == 'quartic':   # φ0 ≈ sqrt(8N + φ_end^2/M_Pl^2) M_Pl
            phi_end = 2.0*torch.sqrt(torch.as_tensor(2.0))*MPl
            argument = 8.0*Nstar + (phi_end/MPl)**2
            phi0 = torch.sqrt(torch.as_tensor(argument)) * MPl
                    
        elif param['potential'] == 'hilltop' :
            # Analytical approximation near the top (see §2c)
            # N ≈ (mu^4/(8 M_Pl^2)) (1/φ0^2 - 1/φ_end^2)  ⇒  φ0 ≈ mu^2 / sqrt(8 N M_Pl^2 + mu^4/φ_end^2)
            # We estimate φ_end with ε_V≈1: φ_end ≈ (mu**(8)/(32 M_Pl**2))**(1/6)
            phi_end = (param['mu']**8/(32.0*MPl**2))**(1.0/6.0)
            denom = 8.0*Nstar*(MPl**2) + (param['mu']**4)/(phi_end**2)
            phi0 =  (param['mu']**2)/torch.sqrt(torch.as_tensor(denom))
    
        elif param['potential'] == 'alpha':
            # φ0 ≈ (1/s) ln(4N/(3α)), s = sqrt(2/(3α))
            phi_end = 4/(3*param['alpha'])
            argument=2.0/(3.0*max(param['alpha'],1e-12))
            s = torch.sqrt(torch.as_tensor(argument))
            phi0 =  (1.0/s)*torch.log(4.0*Nstar/(3.0*max(param['alpha'],1e-12)))
        elif param['potential'] == 'VNet':
            phi_end = 1.64
            phi0 = 5.0
    else:
        phi_end = torch.sqrt(torch.as_tensor(2.0))* MPl
        
    H0 = friedmann_Equation(phi0, 0.0, param)   # Initial approximation for H (slow-roll)
    phi_dot0 = -(param['c']**2)*dV_dphi(phi0, param) / (3 * H0) # Slow-roll approximation for phi_dot    
    y0 = [phi0, phi_dot0]
    V0 = V(phi0,param)
    # Integration time (in units of M_Pl^-1)
    factor = param['hbar']/(param['m']*param['c']**2)
    t_span = (0, kappa_time*factor)
    t_eval = torch.linspace(0, kappa_time*factor, 10000)

    condition = {
        'phi0': phi0,
        'phi_dot0': phi_dot0,
        'phi_end': phi_end,
        'H0': H0,
        'y0': y0,
        'V0': V0,
        't_span': t_span,
        't_eval': t_eval
    }
    return condition

param['potential'] = 'hilltop'
condition = get_Initial_Conditions(param)
print('Initial Conditions\n-----------------------')
print(f'phi0:\t  {condition["phi0"]:.3e}')
print(f'phi_dot0:  {condition["phi_dot0"]:.3e}')
print(f'H0:\t  {condition["H0"]:.3e}')
print(f'V0:\t  {condition["V0"]:.3e}')
print('Potential:', param['potential'])
```

### 2.3. TIME INTEGRATION Solution of the Differential Equations.

---

Steps of the Integration Method

**1. Problem Definition:**

* **Differential Equations:** The function equations provides the ODE system:
  $$\frac{d}{dt} \begin{bmatrix} \phi \ \dot{\phi} \end{bmatrix} = \begin{bmatrix} \dot{\phi} \ -3 H \dot{\phi} - c^2 \frac{dV}{d\phi} \end{bmatrix}$$
  where $ \phi $ is the scalar field, $ \dot{\phi} $ is its time derivative, $ H $ is the Hubble parameter (computed by friedmann_Equation), and $ \frac{dV}{d\phi} $ is the derivative of the potential.

* **Initial Conditions:** The condition dictionary contains:

* t_span: Time interval for the integration, for example, $ [t_{\text{initial}}, t_{\text{final}}] $.

* y0: Initial condition vector $ [\phi_0, \dot{\phi}_0] $, where $ \phi_0 $ is the initial field value and $ \dot{\phi}_0 $ its initial derivative.

* t_eval: Optional time points where the solution is evaluated (if specified).

* phi0 and phi_end: Initial and final (theoretical) values of the scalar field, used for diagnostics.

* **Termination Event:** The events function (for example, end_inflation) defines the condition to stop the integration, such as when $ \epsilon = \min(\epsilon_V, \epsilon_H) = 1 $, indicating the end of inflation.

**2. Integration Method:**

The solve_ivp function solves the ODEs using the method specified in the method parameter (default: RK45). This is an explicit fourth-fifth order Runge-Kutta method, combining a fourth-order method with a fifth-order error control to adaptively adjust the time step size.

**Parameters of solve_ivp:**

* fun: The equations function, which defines the ODE system.
* t_span: Time interval $ [t_{\text{initial}}, t_{\text{final}}] $.
* y0: Initial conditions $ [\phi_0, \dot{\phi}_0] $.
* args: Additional parameters passed to equations, in this case, the param dictionary (containing constants such as $ \hbar $, $ c $, $ M_{\text{Pl}} $).
* t_eval: Time points where the solution is desired (optional).
* events: Function defining stopping conditions for the integration (for example, end_inflation).

The RK45 method is robust and efficient for non-stiff problems, such as those encountered in inflationary models, since it adjusts the time step to maintain numerical accuracy.

**3. Integration Results:**

The solution sol contains:

* sol.t: Times at which the solution was evaluated.
* sol.y[0]: Values of $ \phi $ (phi).
* sol.y[1]: Values of $ \dot{\phi} $ (phi_dot).
* sol.t_events: Times at which the events were triggered (for example, the end of inflation).

From these results, the following quantities are computed:

* **Hubble parameter $ H $:**
  $$H = \sqrt{\frac{\frac{1}{2} \dot{\phi}^2 + V(\phi)}{3}}$$
  Note: This expression simplifies the Friedmann equation by assuming certain units or normalizations (compared with the previous version in friedmann_Equation).

* **Slow-roll parameters:**

  * $ \epsilon_V $: Computed by epsilon_V_Equation.

  * $ \epsilon_H $: Computed by epsilon_H_Equation.

  * $ \eta_V $: Computed by eta_V_Equation.

* **Number of e-folds $ N $:**

  $$N = \int H , dt$$
  Computed numerically using the trapezoidal method (cumulative_trapezoid) over $ H $ with respect to time $ t $, with $ N(0) = 0 $.


```python
def integrate_inflation_by_time(equations, condition, param, events, method='RK45', debug = True):
    # Solve the equations
    sol = solve_ivp(equations, condition['t_span'], condition['y0'], 
                    args=(param,), method=method, 
                    t_eval=condition['t_eval'], events=events,
                   )
    
    # Extract results
    t = sol.t
    phi = sol.y[0]
    phi_dot = sol.y[1]
    
    H = friedmann_Equation(phi, phi_dot, param)
    #epsilon_V = 0.5 * phi_dot**2 / H**2
    epsilon_V = epsilon_V_Equation( phi, param)
    eta_V = eta_V_Equation( phi, param)
    epsilon_H = epsilon_H_Equation( H, phi_dot, param)
    
    # Compute number of e-folds
    N_fold = cumulative_trapezoid(H, t, initial=0)  # N_fold = int H dt
    
    # Print key results
    t_end = sol.t_events[0][0] if sol.t_events[0].size > 0 else t[-1]
    phi_end = sol.y[0][-1]
    N_total = N_fold[-1]
    if debug:
        print(f"Time at the end of inflation:      \t {t_end:.2f}")
        print(f"Initial field value:               \t {condition['phi0']:.3f}")
        print(f"Field value at the end (simul.):   \t {phi_end:.3f}")
        print(f"Field value at the end (theory):   \t {condition['phi_end']:.3f}")
        print(f"Total number of e-folds:           \t {N_total:.2f}")

    results = {
        't' : t,                # Time variable
        'phi' : phi,            # Phi field
        'phi_dot': phi_dot,     # Field derivative
        'H': H,                 # Hubble parameter
        'epsilon_V': epsilon_V,     # epsilon V parameter
        'epsilon_H': epsilon_H,     # epsilon H parameter
        'eta_V': eta_V,     # eta V parameter
        'N': N_fold,            # e-Folds          
        't_end': t_end,         # Final time
        'phi_end': phi_end,     # Final value of the field      
        'N_total': N_total,     # Total number of e-Folds          
    }

    return results
    
param['potential'] = 'quad'
condition  =get_Initial_Conditions(param) 
results = integrate_inflation_by_time(equations, condition=condition, param=param, events=end_inflation, method='RK45')
```

### 3.1. Definition of Functions H2, Epsilon_V, Epsilon_H and Eta_V in terms of the parameter N (e-Folds)

---

These functions use the variable $ u = \frac{d\phi}{dN} $, where $ \phi $ is the scalar field and $ N $ is the number of e-folds.

**1. Square of the Hubble Parameter ($ H^2 $) (H2_from_phi_u)**
This function computes the square of the Hubble parameter $ H^2 $ as a function of the scalar field $ \phi $ and the derivative of the field with respect to the number of e-folds $ u = \frac{d\phi}{dN} $.

$$H^2 = \frac{V(\phi)}{3 M_{\text{Pl}}^2 - \frac{1}{2} u^2}$$

* **Description:**
  This equation is derived from the Friedmann equation in the context of inflation, expressed in terms of $ u = \frac{d\phi}{dN} $. It represents the expansion rate of the universe ($ H^2 $) as a function of the potential $ V(\phi) $ and the derivative of the field with respect to the number of e-folds. The term $ \frac{1}{2} u^2 $ reflects the contribution of the kinetic energy of the scalar field, adjusted by the relation between cosmic time and the number of e-folds ($ \frac{dN}{dt} = H $).

**2. Slow-Roll Parameter $ \epsilon_V $ (epsV)**
This function computes the slow-roll parameter based on the potential, $ \epsilon_V $, which measures the relative slope of the potential.

$$\epsilon_V = \frac{1}{2} M_{\text{Pl}}^2 \left( \frac{V'}{V} \right)^2$$

* **Description:**
  The parameter $ \epsilon_V $ measures the slope of the potential $ V(\phi) $, normalized by the potential itself and the Planck mass. It is a key indicator of slow-roll conditions: small values ($ \epsilon_V \ll 1 $) imply that inflation can be sustained. The equation is identical to that defined in epsilon_V_Equation in the previous functions, but adapted to use PyTorch tensors.

**3. Slow-Roll Parameter $ \eta_V $ (etaV)**
This function computes the slow-roll parameter $ \eta_V $, related to the curvature of the potential.

$$\eta_V = M_{\text{Pl}}^2 \frac{V''}{V}$$

* **Description:**
  The parameter $ \eta_V $ measures the curvature of the potential (second derivative) relative to the potential itself, normalized by $ M_{\text{Pl}}^2 $. Small values of $ \eta_V $ ($ |\eta_V| \ll 1 $) are required to maintain inflation in the slow-roll regime. This equation is consistent with that defined in eta_V_Equation in the previous functions.

**4. Slow-Roll Parameter $ \epsilon_H $ (epsH_from_u)**
This function computes the slow-roll parameter $ \epsilon_H $ in terms of $ u = \frac{d\phi}{dN} $.

$$\epsilon_H = \frac{1}{2} \frac{u^2}{M_{\text{Pl}}^2}$$

* **Description:**
  The parameter $ \epsilon_H $ measures the contribution of the kinetic energy of the scalar field (represented by $ u^2 $) relative to the Planck scale. Since $ u = \frac{d\phi}{dN} = \frac{\dot{\phi}}{H} $ (because $ \frac{dN}{dt} = H $), this equation is equivalent to:
  $$\epsilon_H = \frac{1}{2} \frac{\dot{\phi}^2}{H^2 M_{\text{Pl}}^2}$$
  This matches the definition of $ \epsilon_H $ in epsilon_H_Equation, but expressed in terms of $ u $ instead of $ \dot{\phi} $ and $ H $. Small values of $ \epsilon_H $ indicate sustained inflationary expansion.

**General Notes**

* Use of $ u = \frac{d\phi}{dN} $: The variable $ u $ represents the derivative of the scalar field with respect to the number of e-folds, which is useful in inflationary models because $ N $ directly measures the amount of expansion. The relation between $ u $ and the time derivative is $ u = \frac{\dot{\phi}}{H} $.
* **Numerical Stability**: The functions use torch.clamp to limit denominators close to zero, avoiding numerical errors (for example, $ \text{denom} \geq 10^{-30} $, $ V \geq 10^{-300} $).
* **Dependence on the Potential:** The functions depend on $ V(\phi) $, $ \frac{dV}{d\phi} $, and $ \frac{d^2V}{d\phi^2} $, computed externally by $ V $, $ dV_dphi $, and $ d2V_dphi $. The specific form of the potential (quadratic, quartic, hilltop, alpha, etc.) determines the results.
* **Use of PyTorch:** The equations are implemented with PyTorch tensors (torch.as_tensor, torch.clamp), enabling efficient computation and compatibility with hardware such as GPUs.
* **Physical Constants:** The parameter param includes $ M_{\text{Pl}} $ (Planck mass), and other constants such as $ \hbar $ or $ c $ may be implicitly included in the potential or in normalization factors.

**Summary**
The provided equations compute key quantities for inflationary dynamics:

* $ H^2 $: Square of the Hubble parameter, describing the expansion rate of the universe.
* $ \epsilon_V $: Slow-roll parameter based on the slope of the potential.
* $ \eta_V $: Slow-roll parameter based on the curvature of the potential.
* $ \epsilon_H $: Slow-roll parameter based on the kinetic energy of the scalar field.



```python
# ============================================================
# Auxiliar Magnitudes: H(φ,u), ε_V, η_V, ε_H
# u = dφ/dN
# ============================================================

def H2_from_phi_u(phi, u, param):
    Mpl2  = torch.as_tensor(param['M_Pl'])**2
    V_    = V(phi, param)
    # H^2 = V / (3 Mpl^2 - 0.5 u^2)
    denom = 3.0 * Mpl2 - 0.5 * u**2
    # avoid small denominators
    denom = torch.clamp(denom, min=1e-30)
    return V_ / denom

def epsV(phi, param):
    Mpl2 = torch.as_tensor(param['M_Pl'])**2
    V_   = V(phi, param)
    Vp   = dV_dphi(phi, param)
    return 0.5 * Mpl2 * (Vp / torch.clamp(V_, min=1e-300))**2

def etaV(phi, param):
    Mpl2 = torch.as_tensor(param['M_Pl'])**2
    V_   = V(phi, param)
    Vpp   = d2V_dphi(phi, param)
    return Mpl2 * Vpp / torch.clamp(V_, min=1e-300)

def epsH_from_u(u, param):
    Mpl2 = torch.as_tensor(param['M_Pl'])**2
    return 0.5 * (u**2) / torch.clamp(Mpl2, min=1e-300)
```

#### 3.2. Slow Roll Initial Conditions

---

The functions slowroll_u0 and slowroll_H0 compute approximate initial conditions for the variable $ u_0 = \frac{d\phi}{dN} $ and the Hubble parameter $ H_0 $ in the slow-roll regime for an inflationary model, given an initial value of the scalar field $ \phi_0 $. These functions are essential for initializing the numerical integration of the differential equations describing inflationary dynamics, assuming that slow-roll conditions are valid at the beginning of inflation.

**1. Slow-Roll Initial Condition for $ u_0 $ (slowroll_u0)**

Initial condition $ u_0 $:
$$u_0 = \frac{d\phi}{dN} \approx - M_{\text{Pl}}^2 \frac{V'}{V}$$

* **Purpose:**

* The variable $ u = \frac{d\phi}{dN} $ represents the derivative of the scalar field $ \phi $ with respect to the number of e-folds $ N $, where $ N $ measures the amount of inflationary expansion ($ \frac{dN}{dt} = H $).

* In the slow-roll regime, it is assumed that the acceleration of the scalar field ($ \ddot{\phi} $) is negligible and that cosmic friction dominates ($ 3H\dot{\phi} \approx -V' $). This leads to the approximation:
  $$\dot{\phi} \approx -\frac{V'}{3H}$$
  Since $ u = \frac{d\phi}{dN} = \frac{\dot{\phi}}{H} $, we obtain:
  $$u \approx -\frac{V'}{3H^2}$$
  Using the slow-roll Friedmann equation ($ H^2 \approx \frac{V}{3 M_{\text{Pl}}^2} $), we derive:
  $$u \approx - \frac{V'}{\frac{V}{M_{\text{Pl}}^2}} = - M_{\text{Pl}}^2 \frac{V'}{V}$$

* The function slowroll_u0 computes this quantity at $ \phi = \phi_0 $, providing an approximate initial condition for $ u_0 $, which is the initial derivative of the field with respect to the number of e-folds.

* This equation is used to initialize the numerical integration of the differential equations in terms of $ N $ (instead of time $ t $), since $ u = \frac{d\phi}{dN} $ is the relevant variable in e-fold-based formulations.

* The approximation is valid when the slow-roll parameters ($ \epsilon_V $, $ \eta_V $) are small, ensuring sustained inflation.

**2. Slow-Roll Initial Condition for $ H_0 $ (slowroll_H0)**

Initial condition $ H_0 $:
$$H_0 = \sqrt{\frac{V}{3 M_{\text{Pl}}^2}}$$

* $ H_0 $ is the initial expansion rate of the universe in the inflationary model, computed under the slow-roll approximation.

* It provides the initial value of $ H $, which determines the scale of cosmic expansion and is used to compute other quantities, such as the number of e-folds $ N = \int H , dt $.




```python
# ============================================================
# slow-roll Inicialization  (u0 y H0)
# ============================================================

def slowroll_u0(phi0, param):
    # u_SR = dφ/dN ≈ - Mpl^2 V'/V
    Mpl2 = torch.as_tensor(param['M_Pl'])**2
    V_  = torch.as_tensor(V(phi0, param), dtype=torch.float64)
    Vp  = torch.as_tensor(dV_dphi(phi0, param), dtype=torch.float64)
    return - Mpl2 * Vp / torch.clamp(V_, min=1e-300)

def slowroll_H0(phi0, param):
    Mpl2 = torch.as_tensor(param['M_Pl'])**2
    V_   = V(phi0, param)
    return torch.sqrt(torch.clamp(V_ / (3.0 * Mpl2), min=1e-300))

```

### 3.3. Integration Using Fourth-Order Runge-Kutta over the Number of e-Folds, $ N $

---

**Differential Equations (rhs)**
The rhs function defines the system of differential equations in terms of the number of e-folds $ N $, where the state of the system is $ y = [\phi, u] $, with $ u = \frac{d\phi}{dN} $.

* **Variables:**

  * $ \phi $: Scalar field.

  * $ u = \frac{d\phi}{dN} $: Derivative of the scalar field with respect to the number of e-folds.

  * $ H $: Hubble parameter.

  * $ V' = \frac{dV}{d\phi} $: Derivative of the potential with respect to $ \phi $.

  * $ M_{\text{Pl}} $: Planck mass (from param['M_Pl']).

* **Equations:**

  * Square of the Hubble parameter:
  $$H^2 = \frac{V(\phi)}{3 M_{\text{Pl}}^2 - \frac{1}{2} u^2}$$

  * Slow-roll parameter $ \epsilon_H $:
  $$\epsilon_H = \frac{1}{2} \frac{u^2}{M_{\text{Pl}}^2}$$

  * System of differential equations:
  $$\frac{d}{dN} \begin{bmatrix} \phi \ u \end{bmatrix} = \begin{bmatrix} u \ -(3 - \epsilon_H) u - \frac{\frac{dV}{d\phi}}{H^2} \end{bmatrix}$$

* **Derivation:**

  * The first equation, $ \frac{d\phi}{dN} = u $, is simply the definition of $ u $.

  * The second equation, $ \frac{du}{dN} $, is derived from the Klein-Gordon equation expressed in terms of $ N $.


```python
# ============================================================
# Integraction on N with RK4
# State: y = [φ, u] with u = dφ/dN
# ============================================================

def rhs(phi, u, param):
    # y' = [u, du/dN]
    H2   = H2_from_phi_u(phi, u, param)
    H    = torch.sqrt(torch.clamp(H2, min=1e-300))
    eH   = 0.5 * u**2 / (torch.as_tensor(param['M_Pl'])**2)  # ε_H = u^2/(2 Mpl^2)
    Vp   = dV_dphi(phi, param)
    du   = -(3.0 - eH) * u - Vp / torch.clamp(H2, min=1e-300)
    return u, du, H

def integrate_inflation_by_efolds(
        param,
        condition,
        dN=1e-3,
        N_max=80.0,
        stop_when='epsV',   # 'epsV' o 'epsH'
        record_time=True,
        si_time=False,       # True: return t inn seconds (require H in s^-1)
        debug = True,
        full_time = False
    ):
    # Inicialization
    phi0 = torch.as_tensor(condition['phi0'])
    u0   = slowroll_u0(phi0, param)
    H0   = slowroll_H0(phi0, param)

    steps = int(N_max / dN) + 1
    N  = torch.zeros(steps, dtype=torch.float64)
    phi = torch.zeros_like(N); phi[0] = phi0
    phi_dot  = torch.zeros_like(N); phi_dot[0]  = u0
    H = torch.zeros_like(N); H[0] = H0

    epsilon_V = torch.zeros_like(N); epsilon_V[0] = epsV(phi0, param)
    eta_V = torch.zeros_like(N); eta_V[0] = etaV(phi0, param)
    epsilon_H = torch.zeros_like(N); epsilon_H[0] = epsH_from_u(u0, param)

    t = torch.zeros_like(N)
    if record_time:
        t[0] = torch.tensor(0.0)

    # RK4 Integration on N
    stop_idx = steps-1
    
    for i in range(steps-1):
        phi_i = phi[i]
        u_i   = phi_dot[i]

        # k1
        f1_u, f1_du, H1 = rhs(phi_i, u_i, param)
        # k2
        f2_u, f2_du, _  = rhs(phi_i + 0.5*dN*f1_u, u_i + 0.5*dN*f1_du, param)
        # k3
        f3_u, f3_du, _  = rhs(phi_i + 0.5*dN*f2_u, u_i + 0.5*dN*f2_du, param)
        # k4
        f4_u, f4_du, H4 = rhs(phi_i + dN*f3_u,     u_i + dN*f3_du,     param)

        phi[i+1] = phi_i + (dN/6.0) * (f1_u + 2*f2_u + 2*f3_u + f4_u)
        phi_dot[i+1]  = u_i   + (dN/6.0) * (f1_du + 2*f2_du + 2*f3_du + f4_du)
        N[i+1]  = N[i] + dN

        H_new   = rhs(phi[i+1], phi_dot[i+1], param)[2]
        H[i+1] = H_new

        # Slow-roll params
        epsilon_V[i+1] = epsV(phi[i+1], param)
        eta_V[i+1] = etaV(phi[i+1], param)
        epsilon_H[i+1] = epsH_from_u(phi_dot[i+1], param)

        # time
        if record_time:
            dti = dN / torch.clamp(H_new, min=1e-300)
            t[i+1] = t[i] + dti

        # stop critery
        crit = epsilon_V[i+1] if stop_when == 'epsV' else epsilon_H[i+1]
        if crit >= 1.0 and not full_time:
            stop_idx = i+1
            break
    
    results = {
        't' : t[:stop_idx+1]*5.3e-44,                # time
        'phi' : phi[:stop_idx+1],                    # Phi Field
        'phi_dot': phi_dot[:stop_idx+1],             # Field Derivative (respect N)
        'H': H[:stop_idx+1],                         # Hubble Parámeter
        'epsilon_V': epsilon_V[:stop_idx+1],         # epsilon V parameter
        'epsilon_H': epsilon_H[:stop_idx+1],         # epsilon H parameter
        'eta_V': eta_V[:stop_idx+1],                 # eta V parameter
        'N': N[:stop_idx+1],                         # e-Folds          
        't_end': t[stop_idx].item()*5.3e-44,         # Final time
        'phi_end': phi[stop_idx].item(),             # final value of the field
        'N_total': N[stop_idx].item(),               # total e-Folds          
    }
    if debug:
        print(f"Inflation end time:            \t\t {results['t_end']:.2e} s.")
        print(f"Initial field value:           \t\t {condition['phi0']:.3f}")
        print(f"Final field value (simul.):    \t\t {results['phi_end']:.3f}")
        print(f"Final field value (theory):    \t\t {condition['phi_end']:.3f}")
        print(f"Total Number of e-folds:       \t\t {results['N_total']:.2f}")
        
    return results
```

```python
def plot_Results(results, X = 't', n_sample = 1000):
    if n_sample>0:
        n_samples = 1000
        size = results['t'].shape[0]
        idx = torch.linspace(0, size - 1, n_samples).long()
    else:
        size = results['t'].shape[0]
        idx = torch.linspace(0, size - 1, size).long()
    t = results['t'][idx]
    N = results['N'][idx]
    phi = results['phi'][idx]
    H = results['H'][idx]
    epsilon_V = results['epsilon_V'][idx]
    t_end = results['t_end']
    N_end = results['N_total']
    if X =='t':
        x_plot = t
        end = t_end
        x_label = r'Time $(M_Pl^{-1})$'
    else:
        x_plot = N
        end = N_end
        x_label = r'N'
    
    plt.figure(figsize=(8, 8))
    
    plt.subplot(3, 1, 1)
    plt.plot(x_plot, phi, label=r'$\phi(t)$')
    plt.axvline(end, color='r', linestyle='--', label='End of inflación')
    plt.xlabel(x_label, fontsize=14)
    plt.ylabel(r'$\phi$', fontsize=14)
    plt.legend()
   
    plt.subplot(3, 1, 2)
    plt.plot(x_plot, H, label='H(t)')
    plt.axvline(end, color='r', linestyle='--')
    plt.xlabel(x_label, fontsize=14)
    plt.ylabel('H', fontsize=14)
    plt.legend()
    
    plt.subplot(3, 1, 3)
    plt.plot(x_plot, epsilon_V, label=r'$\epsilon_V(t)$')
    plt.axhline(1, color='r', linestyle='--', label=r'$\epsilon=1$')
    plt.xlabel(x_label, fontsize=14)
    plt.ylabel(r'$\epsilon$', fontsize=14)
    plt.legend()
    
    plt.tight_layout()
    plt.show()
```

### 4.2. Analysis of Results
---

```python
def analisis_Solution(results, param, debug = True):
    N_target = results['N_total'] - 60
    idx = torch.argmin(torch.abs(torch.as_tensor(results['N'] - N_target)))
    epsilon_v = results['epsilon_V'][idx]
    epsilon_h = results['epsilon_H'][idx]
    eta_v = results['eta_V'][idx]
    eta_h = -(-3 * results['H'][idx] * results['phi_dot'][idx] - dV_dphi(results['phi'][idx], param)) / (results['H'][idx] * results['phi_dot'][idx])  # Aproximación
    N = results['N_total']
    p = param['p']
    alpha = param['alpha']
    if param['potential'] == 'quad':
        #n_s = 1 - 6 * epsilon_v + 2 * eta_v
        #r = 16 * epsilon_h
        n_s = 1 - 2/N
        r = 8/N
    elif param['potential'] == 'quartic':
        n_s = 1 - 3/N
        r = 16/N
    elif param['potential'] == 'hilltop':
        n_s = 1 - (2/N)*(p-1)/(p-2)
        r = 16 * epsilon_v
    elif param['potential'] == 'alpha':
        n_s = 1 - 2/N
        r = 12 * alpha/N**2
    else:
        n_s = 1 - 6 * epsilon_v + 2 * eta_v
        r = 16 * epsilon_h
    if debug:
        print(f"Spectral index (n_s) on horizont: \t {n_s:.3f}")
        print(f"tensor-scalar rate (r) on horizont:\t {r:.3f}")
        print()
        if param['potential'] == 'quartic':
            print(f"Parameter lambda  = ", param['lambda_'])
        if param['potential'] == 'hilltop':
            print(f"Parameter p  = {param['p']:.2f}")
            print(f"Parameter mu = {param['mu'].item():.2f}")
            print(f"Parameter V0 = {param['V0'].item():.2e}")
        if param['potential'] == 'alpha':
            print(f"Parameter alpha = {param['alpha'].item():.2f}")
            print(f"Parameter V0 = {param['V0'].item():.2e}")

    param_cosmo ={
        'n_s': n_s,
        'r':r
    }
    #print(epsilon_h.item(), epsilon_v.item(), eta_v.item(), eta_h.item(), idx, N_target, results['N_total'], results['N'].shape)
    return param_cosmo

param_cosmo = analisis_Solution(results, param)
```

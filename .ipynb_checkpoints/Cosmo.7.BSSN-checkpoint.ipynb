## 1.1. Goal and Philosophy of the BSSN Scheme

The purpose of this project is to construct a numerical relativity simulation based on the BSSNOK formulation of Einstein’s equations. The BSSN system rewrites General Relativity as a well-posed initial value problem suitable for stable numerical evolution.

The core idea is the following:

- Specify initial geometric data on a spatial hypersurface $\Sigma_{t=0}$,
- Choose gauge conditions (lapse and shift),
- Impose boundary conditions,
- Evolve the system forward in time by solving a hyperbolic PDE system.

The standard ADM formulation is known to be numerically unstable, as it is only weakly hyperbolic. The BSSN formulation introduces conformal variables and auxiliary fields that render the system strongly hyperbolic and stable for long-term evolutions, including black hole spacetimes.

**Mapping of the Numerical Scheme**

1. Numerical parameters (grid, resolution, time step).
2. Initial data: $\phi$, $\tilde{\gamma}{ij}$, $K$, $\tilde{A}{ij}$, $\tilde{\Gamma}^i$.
3. Spatial discretization and differential operators.
4. BSSN evolution equations and gauge conditions.
5. Time integration (Method of Lines, Runge–Kutta).
6. Constraint monitoring.
7. Boundary conditions.
8. Physical observables.
9. Visualization and output.

**Foliation and Metric Decomposition**

Spacetime is foliated into a family of spatial hypersurfaces $\Sigma_t$, each labeled by a coordinate time $t$. The spacetime metric is decomposed as. $$
ds^2 =
(-\alpha^2 + \beta_i \beta^i)\,dt^2
+
2 \beta_i \, dt \, dx^i
+
\gamma_{ij} \, dx^i dx^j .
$$
where:
- $\gamma_{ij}$ is the spatial metric,
- $\alpha$ is the lapse function,
- $\beta^i$ is the shift vector.

**Extrinsic Curvature**

The extrinsic curvature measures how each hypersurface is embedded in the four-dimensional spacetime:
$$
K_{ij}=
-\frac{1}{2}
\mathcal{L}{\mathbf{n}}
\gamma{ij}.
$$
In coordinate form, the evolution equation for the spatial metric becomes
$$
\partial_t \gamma_{ij}=
-2 \alpha K_{ij}
+
D_i \beta_j
+
D_j \beta_i .
$$

**Einstein Constraints (ADM Constraints)**

Einstein’s equations split into evolution equations plus constraints.
1. Hamiltonian Constraint
$$
\mathcal{H}
\equiv
{}^{(3)}R
+
K^2
+K_{ij} K^{ij}
+ 16\pi \rho
=0
$$
2. Momentum Constraint
$$
\mathcal{M}^i
\equiv
D_j
\left(
K^{ij}
- \gamma^{ij} K
\right)
-\pi j^i
=0
$$
For vacuum spacetimes:
$$
\rho = 0,
\qquad
j^i = 0 .
$$


## 1.2. Motivation for the BSSN Formulation
---
The BSSN formulation improves numerical stability through conformal decomposition and auxiliary variables.

**Conformal Metric and Conformal Factor**

Define the conformal metric
$$
\tilde{\gamma}_{ij}=
\psi^{-4}
\gamma_{ij} .
$$
The conformal factor is written logarithmically as
$$
\phi
\equiv
\ln \psi .
$$
The determinant of the conformal metric is fixed:
$$
\det(\tilde{\gamma}_{ij}) = 1 ,
$$
which implies $$
\phi=
\frac{1}{12}
\ln \gamma .
$$

**Decomposition of the Extrinsic Curvature**

The extrinsic curvature is decomposed into trace and trace-free parts:
$$
K
\equiv
\gamma^{ij} K_{ij} ,
$$
$$
A_{ij}
\equiv
K_{ij}-
\frac{1}{3}
\gamma_{ij} K ,
$$
$$
\tilde{A}{ij}
\equiv
\psi^{-4}
A{ij}=
e^{-4\phi}
A_{ij} .
$$

**Conformal Connection Functions**

Auxiliary variables are introduced:
$$
\tilde{\Gamma}^i
\equiv
\tilde{\gamma}^{jk}
\tilde{\Gamma}^i{}_{jk} .
$$
These variables allow the Ricci tensor to be written in a form dominated by Laplacian operators, which is essential for numerical stability.

## 1.3. Structure of the BSSN Evolution System

The BSSN equations are written using the advective derivative
$$\frac{d}{dt}\equiv \partial_t - \mathcal{L}_\beta$$

The system evolves the following variables:

Conformal metric $\tilde{\gamma}_{ij}$,

Conformal factor $\phi$,

Trace-free extrinsic curvature $\tilde{A}_{ij}$,

Mean curvature $K$,

Conformal connection functions $\tilde{\Gamma}^i$.

Crucially, the evolution equation for $\tilde{\Gamma}^i$ includes terms proportional to the momentum constraint, which significantly improves long-term stability.

## 2.1. Define Numerical Parameters (BSSN Simulation Skeleton)
---
This module defines:
- Spatial domain and grid resolution (1D/2D/3D supported)
- Grid spacings (dx, dy, dz)
- Time step dt from a CFL condition (or user override)
- Total steps Nt and optional dt adjustment so Nt*dt == T_final
- A torch-friendly summary dictionary (scalars as Python floats for clean printing)
- Basic sanity checks

Later blocks will reuse this config to:
- build coordinate tensors (x,y,z) and meshgrids
- define finite-difference stencils
- evolve BSSN variables with RK4 / Method of Lines

### 2.1.1 Grid Configuration

```python

@dataclass
class GridConfig:
    """
    Spatial grid configuration for a Cartesian domain.
     
     Notes
    -----
    - For dim=1: only x is active, Ny=Nz=1 and y/z ranges are ignored.
    - For dim=2: x,y active, Nz=1 and z range ignored.
    - For dim=3: x,y,z active.
    
    Attributes
    ----------
    dim : int
        Spatial dimension (1, 2, or 3).
    x_min, x_max : float
        Domain bounds along x.
    y_min, y_max : float
        Domain bounds along y (only used if dim >= 2).
    z_min, z_max : float
        Domain bounds along z (only used if dim == 3).
    Nx, Ny, Nz : int
        Number of grid points along each axis.
    """
    dim: int = 3

    x_min: float = -10.0
    x_max: float = 10.0
    y_min: float = -10.0
    y_max: float = 10.0
    z_min: float = -10.0
    z_max: float = 10.0

    Nx: int = 129
    Ny: int = 129
    Nz: int = 129

     # Torch dtype/device for all scalar computations in Block 1
    dtype: torch.dtype = torch.float64
    device: torch.device = torch.device("cpu")

    def __post_init__(self):
        # Basic validation (pure Python checks are fine; no numpy/math used)
        assert self.dim in (1, 2, 3), "dim must be 1, 2, or 3"
        assert self.x_max > self.x_min, "x_max must be > x_min"
        assert self.Nx >= 3, "Nx must be >= 3"

        if self.dim >= 2:
            assert self.y_max > self.y_min, "y_max must be > y_min"
            assert self.Ny >= 3, "Ny must be >= 3"
        else:
            self.Ny = 1
            self.y_min = 0.0
            self.y_max = 0.0

        if self.dim == 3:
            assert self.z_max > self.z_min, "z_max must be > z_min"
            assert self.Nz >= 3, "Nz must be >= 3"
        else:
            self.Nz = 1
            self.z_min = 0.0
            self.z_max = 0.0

    def _t(self, x: float) -> torch.Tensor:
        """Create a 0-dim torch scalar on the configured device/dtype."""
        return torch.tensor(x, dtype=self.dtype, device=self.device)

    def spacings(self) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        """
        Compute uniform grid spacings (dx, dy, dz) as torch scalars.

        For inactive dimensions, spacing is set to +inf (torch.inf).
        """
        dx = (self._t(self.x_max) - self._t(self.x_min)) / self._t(self.Nx - 1)

        if self.dim >= 2:
            dy = (self._t(self.y_max) - self._t(self.y_min)) / self._t(self.Ny - 1)
        else:
            dy = torch.tensor(float("inf"), dtype=self.dtype, device=self.device)

        if self.dim == 3:
            dz = (self._t(self.z_max) - self._t(self.z_min)) / self._t(self.Nz - 1)
        else:
            dz = torch.tensor(float("inf"), dtype=self.dtype, device=self.device)

        return dx, dy, dz

    def min_spacing(self) -> torch.Tensor:
        """Return min(dx, dy, dz) over active dimensions, as a torch scalar."""
        dx, dy, dz = self.spacings()
        candidates = [dx]
        if self.dim >= 2:
            candidates.append(dy)
        if self.dim == 3:
            candidates.append(dz)
        return torch.min(torch.stack(candidates))
```

```python
@dataclass
class TimeConfig:
    """
    Time stepping configuration.

    Attributes
    ----------
    CFL : float
        Courant factor for stability (conservative default is ~0.25).
    T_final : float
        Final simulation time.
    dt : Optional[float]
        If provided, dt overrides CFL-based dt.
    """
    CFL: float = 0.25
    T_final: float = 10.0
    dt: Optional[float] = None

    def __post_init__(self):
        assert self.CFL > 0.0, "CFL must be positive"
        assert self.T_final > 0.0, "T_final must be positive"
        if self.dt is not None:
            assert self.dt > 0.0, "dt must be positive if provided"

```

```python
@dataclass
class SimulationConfig:
    """
    Full configuration for Block 1.

    This aggregates the grid and time configuration and derives:
    - dt (torch scalar) from CFL (if dt not supplied)
    - dx, dy, dz (torch scalars)
    - number of time steps Nt (Python int)
    """
    grid: GridConfig
    time: TimeConfig

    # Derived fields (computed later)
    dx: Optional[torch.Tensor] = None
    dy: Optional[torch.Tensor] = None
    dz: Optional[torch.Tensor] = None
    dt: Optional[torch.Tensor] = None
    Nt: Optional[int] = None


    def finalize(self, force_dt_to_hit_Tfinal: bool = True) -> "SimulationConfig":
        """
        Compute derived quantities.

        If force_dt_to_hit_Tfinal=True:
            - choose Nt = ceil(T_final / dt)
            - then reset dt = T_final / Nt so that Nt*dt == T_final exactly.
        """
        dx, dy, dz = self.grid.spacings()
        self.dx, self.dy, self.dz = dx, dy, dz

        # dt from CFL unless overridden
        if self.time.dt is None:
            hmin = self.grid.min_spacing()
            self.dt = self.grid._t(self.time.CFL) * hmin
        else:
            self.dt = self.grid._t(float(self.time.dt))

        # Nt = ceil(T_final / dt) with torch ops (then cast to int)
        T_final_t = self.grid._t(self.time.T_final)
        Nt_t = torch.ceil(T_final_t / self.dt).to(torch.int64)
        Nt_int = int(Nt_t.item())
        self.Nt = max(1, Nt_int)

        if force_dt_to_hit_Tfinal:
            # Reset dt so final time aligns exactly
            self.dt = T_final_t / self.grid._t(self.Nt)

        # Sanity checks
        assert self.dt.item() > 0.0, "dt must be positive"
        assert self.Nt >= 1, "Nt must be >= 1"
        return self

    def summary(self) -> Dict[str, float]:
        """
        Torch-friendly summary for logging/printing.

        Returns Python floats for readability.
        """
        dx = float(self.dx.item()) if self.dx is not None else float("nan")
        dy = float(self.dy.item()) if self.dy is not None else float("nan")
        dz = float(self.dz.item()) if self.dz is not None else float("nan")
        dt = float(self.dt.item()) if self.dt is not None else float("nan")

        return {
            "dim": int(self.grid.dim),
            "Nx": int(self.grid.Nx),
            "Ny": int(self.grid.Ny),
            "Nz": int(self.grid.Nz),
            "x_min": float(self.grid.x_min),
            "x_max": float(self.grid.x_max),
            "y_min": float(self.grid.y_min),
            "y_max": float(self.grid.y_max),
            "z_min": float(self.grid.z_min),
            "z_max": float(self.grid.z_max),
            "dx": dx,
            "dy": dy,
            "dz": dz,
            "CFL": float(self.time.CFL),
            "T_final": float(self.time.T_final),
            "dt": dt,
            "Nt": int(self.Nt) if self.Nt is not None else -1,
            "dtype": str(self.grid.dtype),
            "device": str(self.grid.device),
        }
```

```python

def build_default_config(
    dim: int = 3,
    domain_half_extent: float = 10.0,
    N: int = 129,
    CFL: float = 0.25,
    T_final: float = 10.0,
    dt: Optional[float] = None,
    dtype: torch.dtype = torch.float64,
    device: Optional[torch.device] = None,
) -> SimulationConfig:
    """
    Convenience constructor for a symmetric domain [-L, L]^dim.

    Parameters
    ----------
    dim : int
        Spatial dimension (1/2/3).
    domain_half_extent : float
        L in [-L, L].
    N : int
        Resolution per axis (Nx=Ny=Nz=N for active dimensions).
    CFL : float
        Courant factor (if dt not provided).
    T_final : float
        Final simulation time.
    dt : Optional[float]
        If given, overrides CFL-based dt.

    Returns
    -------
    SimulationConfig
        Finalized configuration (derived quantities computed).
    """
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    L = float(domain_half_extent)

    grid = GridConfig(
        dim=dim,
        x_min=-L, x_max=+L, Nx=N,
        y_min=-L, y_max=+L, Ny=(N if dim >= 2 else 1),
        z_min=-L, z_max=+L, Nz=(N if dim == 3 else 1),
        dtype=dtype,
        device=device,
    )
    time = TimeConfig(CFL=CFL, T_final=T_final, dt=dt)

    cfg = SimulationConfig(grid=grid, time=time).finalize()
    return cfg
```

```python
# Example usage (good for quick validation in a script)
cfg = build_default_config(dim=3, domain_half_extent=20.0, N=129, CFL=0.25, T_final=50.0)

print("=== Block 1: Numerical Parameters Summary ===")
for k, v in cfg.summary().items():
    print(f"{k:>10s}: {v}")
```

Optimization alternatives (practical notes for later blocks)

Adaptive CFL: during evolution, you can reduce dt if constraints blow up (Block 6).

Non-uniform grids / AMR: BSSN binaries typically need refinement near punctures; we can later extend GridConfig to nested boxes.

Higher-order spatial derivatives: using 4th/6th order finite differences forces stricter CFL but improves phase accuracy for waves.

GPU acceleration: once we move to PyTorch tensors, the same config can drive CUDA-based stencils.

Optimization alternatives (still Block 1–relevant)

dtype: start with torch.float64 for stability; once everything works, test float32 + constraint monitoring (Block 6).

device: default picks CUDA if available. For debugging, force device=torch.device("cpu").

CFL tuning: if later (Block 4–5) constraints grow, reduce CFL (e.g. 0.15) or increase spatial resolution.

Exact final time alignment: controlled via force_dt_to_hit_Tfinal; keeping it True helps consistent checkpoints/output.

## Block 2 — Initial Conditions (Torch-first, BSSN variables)
---
This block constructs the initial BSSN state:
    (phi, tilde_gamma_ij, K, tilde_A_ij, tilde_Gamma^i)
plus initial gauge:
    (alpha, beta^i)

We provide two practical initial-data options:

(A) Flat/Minkowski test (debugging baseline)
    psi = 1  -> phi = 0
    tilde_gamma_ij = delta_ij
    K = 0, tilde_A_ij = 0, tilde_Gamma^i = 0
    alpha = 1, beta^i = 0

(B) Brill–Lindquist puncture data (time-symmetric, conformally flat)
    gamma_ij = psi^4 delta_ij
    K_ij = 0  -> K = 0, A_ij = 0 -> tilde_A_ij = 0
    psi = 1 + sum_a m_a / (2 r_a)
    phi = ln(psi)
    tilde_gamma_ij = delta_ij, tilde_Gamma^i = 0 (initially)

Notes:
- This is still "Block 2" (initialization). We are NOT evolving yet.
- Singularities at r=0 (punctures) are handled by a small epsilon in r.
- Later (gauge block) we will switch to 1+log slicing and Gamma-driver shift.
"""

```python

# -----------------------------
# Helper: coordinates on the grid
# -----------------------------
def build_coordinates(cfg, indexing: str = "ij") -> Dict[str, torch.Tensor]:
    """
    Build coordinate tensors (x,y,z) and optionally meshgrid (X,Y,Z).

    Parameters
    ----------
    cfg : SimulationConfigTorch
        From Block 1.
    indexing : str
        "ij" (matrix indexing) is recommended for 3D finite differences.

    Returns
    -------
    coords : dict
        Contains 1D vectors x,y,z and 3D grids X,Y,Z (for dim >= 2/3).
    """
    g = cfg.grid
    device, dtype = g.device, g.dtype

    x = torch.linspace(g.x_min, g.x_max, g.Nx, device=device, dtype=dtype)

    if g.dim >= 2:
        y = torch.linspace(g.y_min, g.y_max, g.Ny, device=device, dtype=dtype)
    else:
        y = torch.zeros(1, device=device, dtype=dtype)

    if g.dim == 3:
        z = torch.linspace(g.z_min, g.z_max, g.Nz, device=device, dtype=dtype)
    else:
        z = torch.zeros(1, device=device, dtype=dtype)

    # Meshgrids are convenient for analytic initial data like punctures.
    # Later, for performance, we can avoid full meshgrids and compute stencils via broadcasting.
    if g.dim == 1:
        X = x
        Y = None
        Z = None
    elif g.dim == 2:
        X, Y = torch.meshgrid(x, y, indexing=indexing)
        Z = None
    else:
        X, Y, Z = torch.meshgrid(x, y, z, indexing=indexing)

    return {"x": x, "y": y, "z": z, "X": X, "Y": Y, "Z": Z}
```

```python

# -----------------------------
# Data container for BSSN state
# -----------------------------
@dataclass
class BSSNState:
    """
    BSSN state + basic gauge variables.

    Tensor shape convention (dim=3):
        scalar fields: (Nx, Ny, Nz)
        vector fields: (3, Nx, Ny, Nz)   index i first
        tensor fields: (3, 3, Nx, Ny, Nz) indices (i,j) first
    """
    # BSSN variables
    phi: torch.Tensor                 # scalar
    tilde_gamma: torch.Tensor         # (3,3,...) conformal metric
    K: torch.Tensor                   # scalar
    tilde_A: torch.Tensor             # (3,3,...) trace-free conformal extrinsic curvature
    tilde_Gamma: torch.Tensor         # (3,...) conformal connection functions

    # Gauge variables (initial)
    alpha: torch.Tensor               # scalar lapse
    beta: torch.Tensor                # (3,...) shift vector

    # Optional: keep psi for diagnostics/IC checks
    psi: Optional[torch.Tensor] = None
```

```python


# -----------------------------
# IC option A: Flat/Minkowski
# -----------------------------
def initial_data_flat(cfg) -> BSSNState:
    """
    Flat spacetime initial data (useful for debugging operators and evolution).

    Returns
    -------
    BSSNState
    """
    g = cfg.grid
    device, dtype = g.device, g.dtype

    # Scalar field shape
    shape_s = (g.Nx, g.Ny, g.Nz) if g.dim == 3 else ((g.Nx, g.Ny) if g.dim == 2 else (g.Nx,))

    zeros_s = torch.zeros(shape_s, device=device, dtype=dtype)
    ones_s  = torch.ones(shape_s,  device=device, dtype=dtype)

    # Build delta_ij conformal metric
    tilde_gamma = torch.zeros((3, 3) + shape_s, device=device, dtype=dtype)
    for i in range(3):
        tilde_gamma[i, i].fill_(1.0)

    # All curvature variables vanish initially
    phi = zeros_s.clone()
    K = zeros_s.clone()
    tilde_A = torch.zeros((3, 3) + shape_s, device=device, dtype=dtype)
    tilde_Gamma = torch.zeros((3,) + shape_s, device=device, dtype=dtype)

    # Gauge: alpha=1, beta=0
    alpha = ones_s.clone()
    beta = torch.zeros((3,) + shape_s, device=device, dtype=dtype)

    # psi = exp(phi) = 1
    psi = ones_s.clone()

    return BSSNState(
        phi=phi,
        tilde_gamma=tilde_gamma,
        K=K,
        tilde_A=tilde_A,
        tilde_Gamma=tilde_Gamma,
        alpha=alpha,
        beta=beta,
        psi=psi,
    )

```

```python

# -----------------------------
# IC option B: Brill–Lindquist punctures (time-symmetric)
# -----------------------------
def initial_data_brill_lindquist_punctures(
    cfg,
    masses: torch.Tensor,
    positions: torch.Tensor,
    eps: float = 1e-6,
    alpha0: float = 1.0,
) -> BSSNState:
    """
    Brill–Lindquist puncture initial data (time-symmetric, conformally flat).

    psi = 1 + sum_a m_a / (2 r_a)
    phi = ln(psi)
    tilde_gamma_ij = delta_ij
    K = 0, tilde_A_ij = 0, tilde_Gamma^i = 0
    alpha = alpha0 (often 1 as a start; later we evolve with 1+log)
    beta^i = 0

    Parameters
    ----------
    cfg : SimulationConfigTorch
    masses : torch.Tensor
        Shape (P,) masses m_a.
    positions : torch.Tensor
        Shape (P,3) positions (x_a, y_a, z_a).
        For dim<3, you can still pass z_a=0 and/or y_a=0.
    eps : float
        Small regularization added under sqrt to avoid division by zero at punctures.
    alpha0 : float
        Initial lapse value.

    Returns
    -------
    BSSNState
    """
    g = cfg.grid
    device, dtype = g.device, g.dtype

    masses = masses.to(device=device, dtype=dtype)
    positions = positions.to(device=device, dtype=dtype)

    coords = build_coordinates(cfg)
    X, Y, Z = coords["X"], coords["Y"], coords["Z"]

    # Build scalar field shape
    if g.dim == 1:
        shape_s = (g.Nx,)
        # For 1D, treat Y=Z=0 and X is already (Nx,)
        Xg = X
        Yg = torch.zeros_like(Xg)
        Zg = torch.zeros_like(Xg)
    elif g.dim == 2:
        shape_s = (g.Nx, g.Ny)
        Xg, Yg = X, Y
        Zg = torch.zeros_like(Xg)
    else:
        shape_s = (g.Nx, g.Ny, g.Nz)
        Xg, Yg, Zg = X, Y, Z

    # Compute psi field
    psi = torch.ones(shape_s, device=device, dtype=dtype)

    # Sum contributions from punctures
    # r_a = sqrt((x-xa)^2 + (y-ya)^2 + (z-za)^2 + eps^2)
    eps_t = torch.tensor(eps, device=device, dtype=dtype)
    for a in range(masses.shape[0]):
        xa, ya, za = positions[a, 0], positions[a, 1], positions[a, 2]
        r = torch.sqrt((Xg - xa) ** 2 + (Yg - ya) ** 2 + (Zg - za) ** 2 + eps_t**2)
        psi = psi + masses[a] / (2.0 * r)

    # Conformal factor variable
    phi = torch.log(psi)

    # Conformal metric tilde_gamma_ij = delta_ij
    tilde_gamma = torch.zeros((3, 3) + shape_s, device=device, dtype=dtype)
    for i in range(3):
        tilde_gamma[i, i].fill_(1.0)

    # Time-symmetric -> K_ij = 0
    K = torch.zeros(shape_s, device=device, dtype=dtype)
    tilde_A = torch.zeros((3, 3) + shape_s, device=device, dtype=dtype)

    # For conformally flat initial data, a common initial choice is tilde_Gamma^i = 0.
    # Later, it will evolve (and the Ricci terms will use derivatives of tilde_gamma and phi).
    tilde_Gamma = torch.zeros((3,) + shape_s, device=device, dtype=dtype)

    # Gauge: initial lapse and shift
    alpha = torch.full(shape_s, float(alpha0), device=device, dtype=dtype)
    beta = torch.zeros((3,) + shape_s, device=device, dtype=dtype)

    return BSSNState(
        phi=phi,
        tilde_gamma=tilde_gamma,
        K=K,
        tilde_A=tilde_A,
        tilde_Gamma=tilde_Gamma,
        alpha=alpha,
        beta=beta,
        psi=psi,
    )
```

```python
# -----------------------------
# Example usage (Notebook-friendly)
# -----------------------------
# Assumes you already ran Block 1 and have `cfg` (SimulationConfigTorch).

# 1) Flat test IC
state_flat = initial_data_flat(cfg)
print("Flat IC: phi mean =", float(state_flat.phi.mean().item()), "psi mean =", float(state_flat.psi.mean().item()))

# 2) Two-puncture Brill–Lindquist IC (simple binary seed)
# Example: equal masses at x=±a, y=z=0
a = 5.0
masses = torch.tensor([1.0, 1.0], dtype=cfg.grid.dtype, device=cfg.grid.device)
positions = torch.tensor([[-a, 0.0, 0.0],
                          [ a, 0.0, 0.0]], dtype=cfg.grid.dtype, device=cfg.grid.device)

state_bl = initial_data_brill_lindquist_punctures(cfg, masses=masses, positions=positions, eps=1e-6, alpha0=1.0)
print("BL puncture IC: phi min/max =", float(state_bl.phi.min().item()), float(state_bl.phi.max().item()))
```

Optimization alternatives (relevant already in Block 2)

Avoid full meshgrids for large 3D runs: you can compute puncture distances using broadcasting with 1D x,y,z vectors (less memory). If you want, I’ll give a “meshgrid-free” version.

Use float64 for IC + first evolutions, then benchmark float32 once constraints are under control.

Regularization eps: smaller gives more accurate puncture near-field but can hurt stability in early debug phases; typical debug values 1e-5–1e-6.

## Block 3 — Spatial Discretization and Differential Operators (Torch-first)
---
This block provides finite-difference operators needed by BSSN:
- First derivatives:  ∂_i f
- Second derivatives: ∂_i∂_i f
- Laplacian:          Δ f = ∂_x^2 f + ∂_y^2 f + ∂_z^2 f
- Gradients for scalar/vector/tensor fields (consistent with our shape conventions)

IMPORTANT SHAPE CONVENTION (dim=3):
- scalar fields: (Nx, Ny, Nz)
- vector fields: (3, Nx, Ny, Nz)         index i first
- tensor fields: (3, 3, Nx, Ny, Nz)      indices (i,j) first

Boundary conditions:
- "periodic" (recommended for early testing): implemented via torch.roll (fast, no padding).
- "neumann0" (zero-normal-derivative): implemented via edge-replication padding.
- "dirichlet0" (zero value at boundary): implemented via constant-zero padding.

Accuracy orders:
- 2nd order centered differences
- 4th order centered differences (more accurate, stricter CFL)

For production BBH runs you often want:
- high-order stencils + Kreiss-Oliger dissipation + radiative outer BCs
Those come later. Here we establish clean, reusable operators.

```python

BCType = Literal["periodic", "neumann0", "dirichlet0"]
OrderType = Literal[2, 4]


def _pad_3d(
    f: torch.Tensor,
    pad: int,
    bc: BCType,
) -> torch.Tensor:
    """
    Pad a scalar field f (Nx,Ny,Nz) by `pad` cells on each side.

    We avoid numpy and use torch.nn.functional.pad.
    """
    import torch.nn.functional as F

    if pad == 0:
        return f

    # torch.nn.functional.pad for 3D expects padding as:
    # (pad_z_left, pad_z_right, pad_y_left, pad_y_right, pad_x_left, pad_x_right)
    p = (pad, pad, pad, pad, pad, pad)

    if bc == "neumann0":
        # replicate boundary values -> approx zero normal derivative at boundary
        return F.pad(f, p, mode="replicate")

    if bc == "dirichlet0":
        # constant zero beyond boundary
        return F.pad(f, p, mode="constant", value=0.0)

    raise ValueError("Padding is only used for non-periodic BCs; periodic uses torch.roll.")

```

```python

@dataclass
class FDOperators:
    """
    Finite-difference operators bound to a given simulation config (from Block 1).

    Parameters
    ----------
    cfg : SimulationConfigTorch
        Must contain dx,dy,dz and dim.
    order : 2 or 4
        Centered finite difference order.
    bc : "periodic" | "neumann0" | "dirichlet0"
        Boundary condition model used by the discrete derivative.

    Notes
    -----
    - For bc="periodic", derivatives use torch.roll and are naturally well-defined everywhere.
    - For bc!="periodic", we pad then apply centered stencils; this introduces boundary modeling.
    """
    cfg: object
    order: OrderType = 4
    bc: BCType = "periodic"

    def __post_init__(self):
        assert self.order in (2, 4), "Only 2nd and 4th order stencils are supported."
        assert self.bc in ("periodic", "neumann0", "dirichlet0")
        assert self.cfg.dx is not None, "Run Block 1 finalize() first so dx/dy/dz exist."
        # Stencil half-width:
        self.r = 1 if self.order == 2 else 2

    # -------------------------
    # Low-level 1D derivative along an axis for a scalar field
    # -------------------------
    def d1(self, f: torch.Tensor, axis: int) -> torch.Tensor:
        """
        First derivative ∂_axis f with centered stencils.

        axis: 0->x, 1->y, 2->z for dim=3 scalar (Nx,Ny,Nz)
        """
        dx = self._spacing(axis)

        if self.bc == "periodic":
            if self.order == 2:
                # (f_{i+1} - f_{i-1}) / (2h)
                return (torch.roll(f, shifts=-1, dims=axis) - torch.roll(f, shifts=+1, dims=axis)) / (2.0 * dx)

            # 4th order:
            # (-f_{i+2} + 8 f_{i+1} - 8 f_{i-1} + f_{i-2}) / (12h)
            return (
                -torch.roll(f, shifts=-2, dims=axis)
                + 8.0 * torch.roll(f, shifts=-1, dims=axis)
                - 8.0 * torch.roll(f, shifts=+1, dims=axis)
                + torch.roll(f, shifts=+2, dims=axis)
            ) / (12.0 * dx)

        # Non-periodic: pad then slice back
        fp = _pad_3d(f, pad=self.r, bc=self.bc)
        ax = axis

        # shift operations on padded field
        if self.order == 2:
            num = torch.roll(fp, shifts=-1, dims=ax) - torch.roll(fp, shifts=+1, dims=ax)
            out = num / (2.0 * dx)
        else:
            num = (
                -torch.roll(fp, shifts=-2, dims=ax)
                + 8.0 * torch.roll(fp, shifts=-1, dims=ax)
                - 8.0 * torch.roll(fp, shifts=+1, dims=ax)
                + torch.roll(fp, shifts=+2, dims=ax)
            )
            out = num / (12.0 * dx)

        # remove padding
        return out[self.r:-self.r, self.r:-self.r, self.r:-self.r]

    def d2(self, f: torch.Tensor, axis: int) -> torch.Tensor:
        """
        Second derivative ∂_axis^2 f with centered stencils.
        """
        dx = self._spacing(axis)

        if self.bc == "periodic":
            if self.order == 2:
                # (f_{i+1} - 2 f_i + f_{i-1}) / h^2
                return (
                    torch.roll(f, shifts=-1, dims=axis)
                    - 2.0 * f
                    + torch.roll(f, shifts=+1, dims=axis)
                ) / (dx * dx)

            # 4th order:
            # (-f_{i+2} + 16 f_{i+1} - 30 f_i + 16 f_{i-1} - f_{i-2}) / (12 h^2)
            return (
                -torch.roll(f, shifts=-2, dims=axis)
                + 16.0 * torch.roll(f, shifts=-1, dims=axis)
                - 30.0 * f
                + 16.0 * torch.roll(f, shifts=+1, dims=axis)
                - torch.roll(f, shifts=+2, dims=axis)
            ) / (12.0 * dx * dx)

        # Non-periodic: pad then slice back
        fp = _pad_3d(f, pad=self.r, bc=self.bc)
        ax = axis

        if self.order == 2:
            out = (
                torch.roll(fp, shifts=-1, dims=ax)
                - 2.0 * fp
                + torch.roll(fp, shifts=+1, dims=ax)
            ) / (dx * dx)
        else:
            out = (
                -torch.roll(fp, shifts=-2, dims=ax)
                + 16.0 * torch.roll(fp, shifts=-1, dims=ax)
                - 30.0 * fp
                + 16.0 * torch.roll(fp, shifts=+1, dims=ax)
                - torch.roll(fp, shifts=+2, dims=ax)
            ) / (12.0 * dx * dx)

        return out[self.r:-self.r, self.r:-self.r, self.r:-self.r]

    def laplacian(self, f: torch.Tensor) -> torch.Tensor:
        """
        Laplacian Δf = ∂_x^2 f + ∂_y^2 f (+ ∂_z^2 f).
        Works for dim=1/2/3; expects scalar field shape consistent with dim.
        """
        dim = self.cfg.grid.dim
        if dim == 1:
            # Interpret scalar as (Nx,) and axis=0
            return self._laplacian_1d(f)

        if dim == 2:
            return self._laplacian_2d(f)

        return self.d2(f, axis=0) + self.d2(f, axis=1) + self.d2(f, axis=2)

    # -------------------------
    # Convenience: gradients for BSSN fields
    # -------------------------
    def grad_scalar(self, f: torch.Tensor) -> torch.Tensor:
        """
        Gradient of scalar: (∂_x f, ∂_y f, ∂_z f)
        Output shape: (3, ...)
        """
        dim = self.cfg.grid.dim
        if dim == 1:
            gx = self._d1_1d(f, axis=0)
            out = torch.zeros((3,) + f.shape, device=f.device, dtype=f.dtype)
            out[0] = gx
            return out

        if dim == 2:
            gx = self._d1_2d(f, axis=0)
            gy = self._d1_2d(f, axis=1)
            out = torch.zeros((3,) + f.shape, device=f.device, dtype=f.dtype)
            out[0], out[1] = gx, gy
            return out

        return torch.stack([self.d1(f, 0), self.d1(f, 1), self.d1(f, 2)], dim=0)

    def grad_vector(self, v: torch.Tensor) -> torch.Tensor:
        """
        Gradient of vector v_i:
        Output: ∂_j v_i with shape (3,3,...) where [i,j,...] = ∂_j v_i.

        Input shape must be (3, ...).
        """
        assert v.shape[0] == 3, "Vector must have leading dimension 3."
        dim = self.cfg.grid.dim

        if dim == 3:
            out = torch.zeros((3, 3) + v.shape[1:], device=v.device, dtype=v.dtype)
            for i in range(3):
                out[i, 0] = self.d1(v[i], 0)
                out[i, 1] = self.d1(v[i], 1)
                out[i, 2] = self.d1(v[i], 2)
            return out

        # For dim<3, we compute only active axes; inactive axes are zeros
        out = torch.zeros((3, 3) + v.shape[1:], device=v.device, dtype=v.dtype)
        active_axes = [0] if dim == 1 else [0, 1]
        for i in range(3):
            for ax in active_axes:
                out[i, ax] = self._d1_dim(f=v[i], axis=ax)
        return out

    def grad_tensor2(self, T: torch.Tensor) -> torch.Tensor:
        """
        Gradient of rank-2 tensor T_{ij}:
        Output: ∂_k T_{ij} with shape (3,3,3,...) where [i,j,k,...] = ∂_k T_{ij}.

        Input shape must be (3,3,...).
        """
        assert T.shape[0] == 3 and T.shape[1] == 3, "Tensor must have leading dims (3,3)."
        dim = self.cfg.grid.dim

        out = torch.zeros((3, 3, 3) + T.shape[2:], device=T.device, dtype=T.dtype)
        active_axes = [0] if dim == 1 else ([0, 1] if dim == 2 else [0, 1, 2])

        for i in range(3):
            for j in range(3):
                Tij = T[i, j]
                for k in active_axes:
                    out[i, j, k] = self._d1_dim(Tij, k)
        return out

    # -------------------------
    # Internal: spacing selection
    # -------------------------
    def _spacing(self, axis: int) -> torch.Tensor:
        """
        Return spacing tensor along axis (0=x,1=y,2=z).
        """
        if axis == 0:
            return self.cfg.dx
        if axis == 1:
            return self.cfg.dy
        if axis == 2:
            return self.cfg.dz
        raise ValueError("axis must be 0,1,2")

    # -------------------------
    # Internal: dim-specialized helpers for 1D/2D (avoid forcing (Nx,Ny,Nz))
    # -------------------------
    def _d1_dim(self, f: torch.Tensor, axis: int) -> torch.Tensor:
        dim = self.cfg.grid.dim
        if dim == 1:
            return self._d1_1d(f, axis=0)
        if dim == 2:
            return self._d1_2d(f, axis=axis)
        return self.d1(f, axis=axis)

    def _laplacian_1d(self, f: torch.Tensor) -> torch.Tensor:
        dx = self.cfg.dx
        if self.bc == "periodic":
            if self.order == 2:
                return (torch.roll(f, -1, 0) - 2.0 * f + torch.roll(f, +1, 0)) / (dx * dx)
            return (-torch.roll(f, -2, 0) + 16.0 * torch.roll(f, -1, 0) - 30.0 * f
                    + 16.0 * torch.roll(f, +1, 0) - torch.roll(f, +2, 0)) / (12.0 * dx * dx)

        # Non-periodic: pad 1D by replicate/zero using 1D pad
        import torch.nn.functional as F
        pad = self.r
        if self.bc == "neumann0":
            fp = F.pad(f, (pad, pad), mode="replicate")
        else:
            fp = F.pad(f, (pad, pad), mode="constant", value=0.0)

        if self.order == 2:
            out = (torch.roll(fp, -1, 0) - 2.0 * fp + torch.roll(fp, +1, 0)) / (dx * dx)
        else:
            out = (-torch.roll(fp, -2, 0) + 16.0 * torch.roll(fp, -1, 0) - 30.0 * fp
                   + 16.0 * torch.roll(fp, +1, 0) - torch.roll(fp, +2, 0)) / (12.0 * dx * dx)

        return out[pad:-pad]

    def _laplacian_2d(self, f: torch.Tensor) -> torch.Tensor:
        # f shape (Nx,Ny)
        dx, dy = self.cfg.dx, self.cfg.dy
        if self.bc == "periodic":
            # reuse d2 logic inline for 2D to avoid 3D padding assumptions
            def d2p(g, ax, h):
                if self.order == 2:
                    return (torch.roll(g, -1, ax) - 2.0 * g + torch.roll(g, +1, ax)) / (h * h)
                return (-torch.roll(g, -2, ax) + 16.0 * torch.roll(g, -1, ax) - 30.0 * g
                        + 16.0 * torch.roll(g, +1, ax) - torch.roll(g, +2, ax)) / (12.0 * h * h)
            return d2p(f, 0, dx) + d2p(f, 1, dy)

        # Non-periodic: pad 2D
        import torch.nn.functional as F
        pad = self.r
        # pad format for 2D: (pad_y_left,pad_y_right,pad_x_left,pad_x_right)
        p = (pad, pad, pad, pad)
        if self.bc == "neumann0":
            fp = F.pad(f, p, mode="replicate")
        else:
            fp = F.pad(f, p, mode="constant", value=0.0)

        def d2c(g, ax, h):
            if self.order == 2:
                return (torch.roll(g, -1, ax) - 2.0 * g + torch.roll(g, +1, ax)) / (h * h)
            return (-torch.roll(g, -2, ax) + 16.0 * torch.roll(g, -1, ax) - 30.0 * g
                    + 16.0 * torch.roll(g, +1, ax) - torch.roll(g, +2, ax)) / (12.0 * h * h)

        out = d2c(fp, 0, dx) + d2c(fp, 1, dy)
        return out[pad:-pad, pad:-pad]

    def _d1_1d(self, f: torch.Tensor, axis: int = 0) -> torch.Tensor:
        dx = self.cfg.dx
        if self.bc == "periodic":
            if self.order == 2:
                return (torch.roll(f, -1, 0) - torch.roll(f, +1, 0)) / (2.0 * dx)
            return (-torch.roll(f, -2, 0) + 8.0 * torch.roll(f, -1, 0)
                    - 8.0 * torch.roll(f, +1, 0) + torch.roll(f, +2, 0)) / (12.0 * dx)

        import torch.nn.functional as F
        pad = self.r
        if self.bc == "neumann0":
            fp = F.pad(f, (pad, pad), mode="replicate")
        else:
            fp = F.pad(f, (pad, pad), mode="constant", value=0.0)

        if self.order == 2:
            out = (torch.roll(fp, -1, 0) - torch.roll(fp, +1, 0)) / (2.0 * dx)
        else:
            out = (-torch.roll(fp, -2, 0) + 8.0 * torch.roll(fp, -1, 0)
                   - 8.0 * torch.roll(fp, +1, 0) + torch.roll(fp, +2, 0)) / (12.0 * dx)

        return out[pad:-pad]

    def _d1_2d(self, f: torch.Tensor, axis: int) -> torch.Tensor:
        dx = self.cfg.dx if axis == 0 else self.cfg.dy
        if self.bc == "periodic":
            if self.order == 2:
                return (torch.roll(f, -1, axis) - torch.roll(f, +1, axis)) / (2.0 * dx)
            return (-torch.roll(f, -2, axis) + 8.0 * torch.roll(f, -1, axis)
                    - 8.0 * torch.roll(f, +1, axis) + torch.roll(f, +2, axis)) / (12.0 * dx)

        import torch.nn.functional as F
        pad = self.r
        p = (pad, pad, pad, pad)  # (yL,yR,xL,xR)
        if self.bc == "neumann0":
            fp = F.pad(f, p, mode="replicate")
        else:
            fp = F.pad(f, p, mode="constant", value=0.0)

        if self.order == 2:
            out = (torch.roll(fp, -1, axis) - torch.roll(fp, +1, axis)) / (2.0 * dx)
        else:
            out = (-torch.roll(fp, -2, axis) + 8.0 * torch.roll(fp, -1, axis)
                   - 8.0 * torch.roll(fp, +1, axis) + torch.roll(fp, +2, axis)) / (12.0 * dx)

        return out[pad:-pad, pad:-pad]
```

```python

# -------------------------------------------------------
# Example usage (Notebook-friendly sanity checks)
# -------------------------------------------------------
# Assumes you already ran Block 1 (cfg) and Block 2 (state_bl or state_flat).

fd = FDOperators(cfg=cfg, order=4, bc="periodic")

# Sanity test 1: derivative of a linear function should be ~constant (periodic domain caveat!)
# We'll test on a sinusoid instead, which is perfectly compatible with periodic BCs.
coords = build_coordinates(cfg)
if cfg.grid.dim == 3:
    X, Y, Z = coords["X"], coords["Y"], coords["Z"]
    k = 2.0
    f = torch.sin(k * X)  # depends only on x
    dfx = fd.d1(f, axis=0)
    dfx_true = k * torch.cos(k * X)
    rel_err = (dfx - dfx_true).abs().mean() / (dfx_true.abs().mean() + 1e-30)
    print("Mean relative error for ∂x sin(kx):", float(rel_err.item()))

    lapf = fd.laplacian(f)
    lapf_true = -(k**2) * torch.sin(k * X)
    rel_err_lap = (lapf - lapf_true).abs().mean() / (lapf_true.abs().mean() + 1e-30)
    print("Mean relative error for Δ sin(kx):", float(rel_err_lap.item()))

# Sanity test 2: gradient of phi (from puncture IC) should exist and be finite (except near punctures)
if 'state_bl' in globals():
    grad_phi = fd.grad_scalar(state_bl.phi)  # shape (3,Nx,Ny,Nz)
    print("grad_phi shape:", tuple(grad_phi.shape), " | grad_phi mean:", float(grad_phi.abs().mean().item()))
```

Notes + optimization alternatives for Block 3

Performance: torch.roll is very fast and GPU-friendly. For non-periodic BCs, padding adds overhead.

Accuracy vs stability: 4th order reduces phase error for gravitational waves but usually forces a smaller CFL.

Memory: this block works without allocating huge temporary tensors beyond rolls; still, 3D grids are expensive.

Next stability tool: when we start evolving (Block 5), we’ll add Kreiss–Oliger dissipation (very standard in NR).

Block 3.5 — Kreiss–Oliger (KO) Dissipation (Torch-first)
---
Purpose
-------
High-order finite differences (especially 4th order) often benefit from explicit numerical
dissipation to damp high-frequency grid-scale noise that can destabilize BSSN evolutions.

We implement a standard KO-like dissipation operator Q[f] and show how to apply it to:
- scalar fields (phi, K, alpha, etc.)
- vector fields ($\beta^i$, $\tilde\Gamma^i$)
- tensor fields (tilde_gamma_ij, tilde_A_ij)

Usage pattern (later in Block 5):
-------------------------------
Given evolution RHS:  ∂_t u = RHS(u)
we modify:
    RHS(u) <- RHS(u) + Q[u]
where Q[u] is small and controlled by a parameter eps_ko.

BC support:
-----------
- periodic: implemented with torch.roll (best for early tests)
- neumann0/dirichlet0: supported via padding, but in practice KO is most common with periodic
  tests or radiative outer boundaries (we will add radiative BCs later).

Notes on scaling
----------------
A common form is:
    Q[u] = - eps_ko * h^(2p-1) * D_+^p D_-^p u
where:
- p is the KO order (often p=3 for 5th derivative-like dissipation used with 4th order FD)
- h is grid spacing
In practice many implementations use:
    Q[u] = - eps_ko * (h) * (D_+^3 D_-^3 u)
or similar, depending on conventions.

Here we implement the classic 1D stencil for p=3:
    D_+^3 D_-^3 u  ~  (u_{i-3} - 6 u_{i-2} + 15 u_{i-1} - 20 u_i + 15 u_{i+1} - 6 u_{i+2} + u_{i+3})

Then we scale by (1 / 64) and optionally by h to keep dimensions sensible.

```python
@dataclass
class KODissipation:
    """
    Kreiss–Oliger dissipation operator.

    Parameters
    ----------
    cfg : SimulationConfigTorch
        Must contain dx,dy,dz and dim.
    eps_ko : float
        Dissipation strength. Typical starting range: 0.05 to 0.2 for debugging.
        For production, you tune it carefully to avoid overdamping physical waves.
    bc : "periodic" | "neumann0" | "dirichlet0"
        Boundary model (periodic recommended early).
    scale_with_h : bool
        If True, multiply each 1D dissipation by the spacing h of that axis.
        This keeps Q[u] roughly consistent under refinement.
    """
    cfg: object
    eps_ko: float = 0.1
    bc: Literal["periodic", "neumann0", "dirichlet0"] = "periodic"
    scale_with_h: bool = True

    def __post_init__(self):
        assert self.cfg.dx is not None, "Run Block 1 finalize() first."
        assert self.bc in ("periodic", "neumann0", "dirichlet0")
        self.device = self.cfg.grid.device
        self.dtype = self.cfg.grid.dtype

        # p=3 KO stencil half-width = 3
        self.r = 3

    def _pad_3d(self, f: torch.Tensor, pad: int) -> torch.Tensor:
        import torch.nn.functional as F
        if pad == 0:
            return f

        p = (pad, pad, pad, pad, pad, pad)
        if self.bc == "neumann0":
            return F.pad(f, p, mode="replicate")
        if self.bc == "dirichlet0":
            return F.pad(f, p, mode="constant", value=0.0)
        raise ValueError("Padding is only used for non-periodic; periodic uses roll.")

    def _ko_1d_stencil_periodic(self, f: torch.Tensor, axis: int) -> torch.Tensor:
        """
        Apply the p=3 KO stencil along one axis (periodic).
        """
        # S[u]_i = (u_{i-3} - 6 u_{i-2} + 15 u_{i-1} - 20 u_i + 15 u_{i+1} - 6 u_{i+2} + u_{i+3})
        S = (
            torch.roll(f, shifts=+3, dims=axis)
            - 6.0 * torch.roll(f, shifts=+2, dims=axis)
            + 15.0 * torch.roll(f, shifts=+1, dims=axis)
            - 20.0 * f
            + 15.0 * torch.roll(f, shifts=-1, dims=axis)
            - 6.0 * torch.roll(f, shifts=-2, dims=axis)
            + torch.roll(f, shifts=-3, dims=axis)
        )
        # Classic normalization often uses 1/64
        return S / 64.0

    def _ko_1d_stencil_nonperiodic(self, f: torch.Tensor, axis: int) -> torch.Tensor:
        """
        Apply the p=3 KO stencil along one axis (non-periodic) using padding.
        """
        fp = self._pad_3d(f, pad=self.r)
        S = (
            torch.roll(fp, shifts=+3, dims=axis)
            - 6.0 * torch.roll(fp, shifts=+2, dims=axis)
            + 15.0 * torch.roll(fp, shifts=+1, dims=axis)
            - 20.0 * fp
            + 15.0 * torch.roll(fp, shifts=-1, dims=axis)
            - 6.0 * torch.roll(fp, shifts=-2, dims=axis)
            + torch.roll(fp, shifts=-3, dims=axis)
        ) / 64.0
        return S[self.r:-self.r, self.r:-self.r, self.r:-self.r]

    def ko_1d(self, f: torch.Tensor, axis: int) -> torch.Tensor:
        """
        1D KO dissipation contribution along one axis.

        Returns Q_axis[f] = -eps_ko * (h * S[f]) or -eps_ko * S[f] depending on scale_with_h.
        """
        h = self._spacing(axis)

        if self.bc == "periodic":
            S = self._ko_1d_stencil_periodic(f, axis=axis)
        else:
            S = self._ko_1d_stencil_nonperiodic(f, axis=axis)

        eps = torch.tensor(self.eps_ko, dtype=f.dtype, device=f.device)

        if self.scale_with_h:
            return -eps * h * S
        else:
            return -eps * S

    def ko_scalar(self, f: torch.Tensor) -> torch.Tensor:
        """
        Full KO dissipation for a scalar field: sum over active axes.
        """
        dim = self.cfg.grid.dim
        if dim == 1:
            return self._ko_scalar_1d(f)

        if dim == 2:
            return self._ko_scalar_2d(f)

        return self.ko_1d(f, axis=0) + self.ko_1d(f, axis=1) + self.ko_1d(f, axis=2)

    def ko_vector(self, v: torch.Tensor) -> torch.Tensor:
        """
        KO dissipation for vector field v_i with shape (3,...).
        """
        assert v.shape[0] == 3
        out = torch.zeros_like(v)
        for i in range(3):
            out[i] = self.ko_scalar(v[i])
        return out

    def ko_tensor2(self, T: torch.Tensor) -> torch.Tensor:
        """
        KO dissipation for rank-2 tensor T_ij with shape (3,3,...).
        """
        assert T.shape[0] == 3 and T.shape[1] == 3
        out = torch.zeros_like(T)
        for i in range(3):
            for j in range(3):
                out[i, j] = self.ko_scalar(T[i, j])
        return out

    # -------------------------
    # Internal helpers for dim<3 (avoid forcing 3D padding)
    # -------------------------
    def _ko_scalar_1d(self, f: torch.Tensor) -> torch.Tensor:
        # f shape (Nx,)
        dx = self.cfg.dx
        eps = torch.tensor(self.eps_ko, dtype=f.dtype, device=f.device)

        if self.bc == "periodic":
            S = (
                torch.roll(f, +3, 0) - 6.0 * torch.roll(f, +2, 0) + 15.0 * torch.roll(f, +1, 0)
                - 20.0 * f
                + 15.0 * torch.roll(f, -1, 0) - 6.0 * torch.roll(f, -2, 0) + torch.roll(f, -3, 0)
            ) / 64.0
            return -eps * (dx * S if self.scale_with_h else S)

        # Non-periodic: pad 1D
        import torch.nn.functional as F
        pad = 3
        fp = F.pad(f, (pad, pad),
                   mode="replicate" if self.bc == "neumann0" else "constant",
                   value=0.0 if self.bc == "dirichlet0" else 0.0)
        S = (
            torch.roll(fp, +3, 0) - 6.0 * torch.roll(fp, +2, 0) + 15.0 * torch.roll(fp, +1, 0)
            - 20.0 * fp
            + 15.0 * torch.roll(fp, -1, 0) - 6.0 * torch.roll(fp, -2, 0) + torch.roll(fp, -3, 0)
        ) / 64.0
        S = S[pad:-pad]
        return -eps * (dx * S if self.scale_with_h else S)

    def _ko_scalar_2d(self, f: torch.Tensor) -> torch.Tensor:
        # f shape (Nx,Ny)
        dx, dy = self.cfg.dx, self.cfg.dy
        eps = torch.tensor(self.eps_ko, dtype=f.dtype, device=f.device)

        def S_axis(g, axis):
            return (
                torch.roll(g, +3, axis) - 6.0 * torch.roll(g, +2, axis) + 15.0 * torch.roll(g, +1, axis)
                - 20.0 * g
                + 15.0 * torch.roll(g, -1, axis) - 6.0 * torch.roll(g, -2, axis) + torch.roll(g, -3, axis)
            ) / 64.0

        if self.bc == "periodic":
            Sx = S_axis(f, 0)
            Sy = S_axis(f, 1)
            Qx = -eps * (dx * Sx if self.scale_with_h else Sx)
            Qy = -eps * (dy * Sy if self.scale_with_h else Sy)
            return Qx + Qy

        # Non-periodic: pad 2D
        import torch.nn.functional as F
        pad = 3
        p = (pad, pad, pad, pad)
        fp = F.pad(f, p, mode="replicate" if self.bc == "neumann0" else "constant", value=0.0)
        Sx = S_axis(fp, 0)
        Sy = S_axis(fp, 1)
        Sx = Sx[pad:-pad, pad:-pad]
        Sy = Sy[pad:-pad, pad:-pad]
        Qx = -eps * (dx * Sx if self.scale_with_h else Sx)
        Qy = -eps * (dy * Sy if self.scale_with_h else Sy)
        return Qx + Qy

    def _spacing(self, axis: int) -> torch.Tensor:
        if axis == 0:
            return self.cfg.dx
        if axis == 1:
            return self.cfg.dy
        if axis == 2:
            return self.cfg.dz
        raise ValueError("axis must be 0,1,2")
```

How to choose eps_ko (practical tuning)

Start with eps_ko ~ 0.1 for early stability tests.

If you see high-frequency noise / “checkerboarding” in constraints: increase to 0.15–0.2.

If waves look overdamped (later when extracting GWs): reduce to 0.02–0.08.

Block 4 — BSSNOK + Gauge: Right-Hand Sides (RHS) in Torch
---
We now build the *continuous* evolution system (still discretized in space via Block 3).
This block constructs the RHS for the variables:

BSSN variables:
    - phi                  (scalar)
    - tilde_gamma_ij        (3x3 tensor)
    - K                    (scalar)
    - tilde_A_ij            (3x3 tensor, trace-free)
    - tilde_Gamma^i         (vector)

Gauge variables (moving punctures standard):
    - alpha                 (lapse) : 1+log slicing
    - beta^i                (shift) : Gamma-driver
    - B^i                   (aux shift driver)

Conventions (dim=3):
    scalar: (Nx,Ny,Nz)
    vector: (3,Nx,Ny,Nz)
    tensor: (3,3,Nx,Ny,Nz)

This block assumes you already have:
    - cfg (Block 1)
    - state (Block 2) containing at least: phi, tilde_gamma, K, tilde_A, tilde_Gamma, alpha, beta
    - fd = FDOperators(cfg, order=?, bc=?) (Block 3)

Vacuum case (no matter sources).

IMPORTANT NOTE (pragmatic):
---------------------------
This is a faithful "first complete" BSSN RHS suitable for early experiments.
For serious BBH production runs, you typically add:
    - constraint damping tweaks,
    - better outer boundary conditions,
    - KO dissipation in RHS (Block 3.5),
    - sometimes densitized lapse / alternative gauge.
We’ll integrate those once the baseline runs.


```python

# ============================================================
# 0) Small tensor utilities (Torch-only)
# ============================================================

def symmetrize_ij(T: torch.Tensor) -> torch.Tensor:
    """Symmetrize a rank-2 tensor field: (T_ij + T_ji)/2."""
    return 0.5 * (T + T.transpose(0, 1))

def trace_wrt_metric(T: torch.Tensor, g_inv: torch.Tensor) -> torch.Tensor:
    """
    Trace: tr = g^{ij} T_ij
    T: (3,3,...)
    g_inv: (3,3,...)
    returns: scalar (...)
    """
    return torch.einsum("ij...,ij...->...", g_inv, T)

def tf_wrt_metric(T: torch.Tensor, g: torch.Tensor, g_inv: torch.Tensor) -> torch.Tensor:
    """
    Trace-free projection with respect to metric g:
        TF(T)_ij = T_ij - (1/3) g_ij (g^{kl} T_kl)
    """
    tr = trace_wrt_metric(T, g_inv)  # (...)
    return T - (1.0 / 3.0) * g * tr.unsqueeze(0).unsqueeze(0)

def det_3x3_field(g: torch.Tensor) -> torch.Tensor:
    """
    Determinant of a 3x3 tensor field g_ij.
    g: (3,3,...)
    returns det(g): (...)
    """
    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]
    return a*(e*i - f*h) - b*(d*i - f*g2) + c*(d*h - e*g2)

def inv_3x3_field(g: torch.Tensor, eps: float = 1e-30) -> torch.Tensor:
    """
    Inverse of 3x3 tensor field g_ij using adjugate/det.
    g: (3,3,...)
    returns g_inv: (3,3,...)
    """
    detg = det_3x3_field(g)
    detg_safe = detg + eps

    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]

    # Cofactors (transpose gives adjugate)
    C00 = (e*i - f*h)
    C01 = -(d*i - f*g2)
    C02 = (d*h - e*g2)

    C10 = -(b*i - c*h)
    C11 = (a*i - c*g2)
    C12 = -(a*h - b*g2)

    C20 = (b*f - c*e)
    C21 = -(a*f - c*d)
    C22 = (a*e - b*d)

    adj = torch.stack([
        torch.stack([C00, C10, C20], dim=0),
        torch.stack([C01, C11, C21], dim=0),
        torch.stack([C02, C12, C22], dim=0),
    ], dim=0)  # (3,3,...)

    return adj / detg_safe.unsqueeze(0).unsqueeze(0)


# ============================================================
# 1) Lie derivatives (advection + deformation by shift)
# ============================================================

def lie_scalar(fd, beta: torch.Tensor, f: torch.Tensor) -> torch.Tensor:
    """Lie derivative of scalar: L_beta f = beta^k ∂_k f."""
    grad_f = fd.grad_scalar(f)  # (3,...)
    return torch.einsum("k...,k...->...", beta, grad_f)

def lie_vector_contrav(fd, beta: torch.Tensor, v: torch.Tensor) -> torch.Tensor:
    """
    Lie derivative of contravariant vector:
        (L_beta v)^i = beta^k ∂_k v^i - v^k ∂_k beta^i
    """
    grad_v = fd.grad_vector(v)       # (i,k,...)=∂_k v^i
    grad_beta = fd.grad_vector(beta) # (i,k,...)=∂_k beta^i
    adv = torch.einsum("k...,ik...->i...", beta, grad_v)
    deform = torch.einsum("k...,ik...->i...", v, grad_beta)
    return adv - deform

def lie_tensor_cov2(fd, beta: torch.Tensor, T: torch.Tensor) -> torch.Tensor:
    """
    Lie derivative of covariant rank-2 tensor:
        (L_beta T)_ij = beta^k ∂_k T_ij + T_ik ∂_j beta^k + T_kj ∂_i beta^k
    """
    grad_T = fd.grad_tensor2(T)      # (i,j,k,...)=∂_k T_ij
    grad_beta = fd.grad_vector(beta) # (k, m,...)=∂_m beta^k  but stored as (k,m,...)

    adv = torch.einsum("k...,ijk...->ij...", beta, grad_T)  # beta^k ∂_k T_ij
    term1 = torch.einsum("ik...,kj...->ij...", T, grad_beta)  # T_ik ∂_j beta^k  (note index mapping)
    term2 = torch.einsum("kj...,ik...->ij...", T, grad_beta)  # T_kj ∂_i beta^k
    # Explanation:
    # grad_beta has indices (k, m, ...)=∂_m beta^k
    # For term1: want T_ik * ∂_j beta^k => contract k, and set m=j.
    # torch.einsum("ik...,kj...->ij...", T, grad_beta) achieves that (m=j).
    # Similarly for term2 but swapping i/j.

    return adv + term1 + term2


# ============================================================
# 2) Conformal and physical connections + Ricci
# ============================================================

def christoffel_conformal(fd, tilde_gamma: torch.Tensor, tilde_gamma_inv: torch.Tensor) -> torch.Tensor:
    """
    Compute conformal Christoffel symbols:
        \tilde{Gamma}^i_{jk} = 1/2 \tilde{gamma}^{il} (∂_j \tilde{gamma}_{lk} + ∂_k \tilde{gamma}_{lj} - ∂_l \tilde{gamma}_{jk})

    Returns:
        G: (3,3,3,...) with indices (i,j,k,...)
    """
    d_tg = fd.grad_tensor2(tilde_gamma)  # (l,k,j,...)=∂_j tilde_gamma_{lk} but stored (i,j,k)=∂_k T_ij
    # Our grad_tensor2 gives out[i,j,k] = ∂_k T_ij.
    # So ∂_j tilde_gamma_{lk} is d_tg[l,k,j]
    G = torch.zeros((3,3,3) + tilde_gamma.shape[2:], device=tilde_gamma.device, dtype=tilde_gamma.dtype)

    for i in range(3):
        for j in range(3):
            for k in range(3):
                # sum over l
                term = 0.0
                for l in range(3):
                    d_j_glk = d_tg[l, k, j]
                    d_k_glj = d_tg[l, j, k]
                    d_l_gjk = d_tg[j, k, l]
                    term = term + tilde_gamma_inv[i, l] * (d_j_glk + d_k_glj - d_l_gjk)
                G[i, j, k] = 0.5 * term
    return G

def christoffel_physical(tilde_Gamma: torch.Tensor, tilde_gamma: torch.Tensor, tilde_gamma_inv: torch.Tensor,
                         dphi: torch.Tensor, Gamma_tilde_ijk: torch.Tensor) -> torch.Tensor:
    """
    Physical Christoffel:
        Gamma^i_{jk} = \tilde{Gamma}^i_{jk} + 2(δ^i_j ∂_k φ + δ^i_k ∂_j φ - \tilde{gamma}_{jk} \tilde{gamma}^{il} ∂_l φ)

    Inputs:
        dphi: (3,...) gradient of φ
        Gamma_tilde_ijk: (i,j,k,...) conformal Christoffel
    Returns:
        Gamma_phys: (i,j,k,...)
    """
    Gamma = Gamma_tilde_ijk.clone()
    # Build the correction term
    # corr^i_{jk} = 2(δ^i_j dphi_k + δ^i_k dphi_j - tilde_gamma_{jk} tilde_gamma^{il} dphi_l)
    # precompute V^i = tilde_gamma^{il} dphi_l
    V = torch.einsum("il...,l...->i...", tilde_gamma_inv, dphi)

    for i in range(3):
        for j in range(3):
            for k in range(3):
                delta_ij = 1.0 if i == j else 0.0
                delta_ik = 1.0 if i == k else 0.0
                corr = 2.0 * (delta_ij * dphi[k] + delta_ik * dphi[j] - tilde_gamma[j, k] * V[i])
                Gamma[i, j, k] = Gamma[i, j, k] + corr
    return Gamma

def ricci_from_christoffel(fd, Gamma_ijk: torch.Tensor) -> torch.Tensor:
    """
    Compute Ricci tensor from Christoffels:
        R_ij = ∂_k Γ^k_{ij} - ∂_j Γ^k_{ik} + Γ^k_{ij} Γ^l_{kl} - Γ^k_{il} Γ^l_{kj}

    Gamma_ijk: (k,i,j,...) = Γ^k_{ij}  (we store as (i,j,k,...) usually)
    We'll assume Gamma_ijk is (k,i,j,...) for this function.

    Returns:
        R_ij: (3,3,...)
    """
    # Derivatives:
    # ∂_k Γ^k_{ij} : sum over k of derivative along axis k of Gamma^k_{ij}
    # Our fd.d1 operates on scalar (Nx,Ny,Nz) along axis 0/1/2.
    # So we can compute for each k: d_k(Gamma[k,i,j]) and sum over k.

    shape = Gamma_ijk.shape[3:]
    R = torch.zeros((3,3) + shape, device=Gamma_ijk.device, dtype=Gamma_ijk.dtype)

    # Precompute contraction C_k = Γ^l_{kl} for each k
    C = torch.zeros((3,) + shape, device=Gamma_ijk.device, dtype=Gamma_ijk.dtype)
    for k in range(3):
        # C_k = sum_l Γ^l_{k l}
        tmp = 0.0
        for l in range(3):
            tmp = tmp + Gamma_ijk[l, k, l]
        C[k] = tmp

    # Helper to take derivative along axis k (0=x,1=y,2=z)
    def d_axis(f, ax):
        return fd.d1(f, axis=ax)

    for i in range(3):
        for j in range(3):
            term1 = 0.0
            for k in range(3):
                term1 = term1 + d_axis(Gamma_ijk[k, i, j], k)  # ∂_k Γ^k_{ij}

            # term2 = ∂_j Γ^k_{ik} (sum over k)
            term2_sum = 0.0
            for k in range(3):
                term2_sum = term2_sum + Gamma_ijk[k, i, k]
            term2 = d_axis(term2_sum, j)

            # term3 = Γ^k_{ij} Γ^l_{kl} = sum_k Γ^k_{ij} C_k
            term3 = 0.0
            for k in range(3):
                term3 = term3 + Gamma_ijk[k, i, j] * C[k]

            # term4 = Γ^k_{il} Γ^l_{kj} = sum_{k,l} Γ^k_{il} Γ^l_{kj}
            term4 = 0.0
            for k in range(3):
                for l in range(3):
                    term4 = term4 + Gamma_ijk[k, i, l] * Gamma_ijk[l, k, j]

            R[i, j] = term1 - term2 + term3 - term4

    return R


def ricci_BSSN(fd, phi: torch.Tensor, tilde_gamma: torch.Tensor, tilde_gamma_inv: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
    """
    Compute the physical Ricci tensor using the standard BSSN split:
        R_ij = \tilde{R}_{ij} + R^{phi}_{ij}

    Here we compute:
    - \tilde{R}_{ij} from conformal Christoffel (full formula via Gamma)
    - R^{phi}_{ij} from phi and conformal derivatives

    Returns:
        R_ij: (3,3,...)
        Rtilde_ij: (3,3,...)
    """
    # Conformal Christoffel symbols \tilde{Gamma}^i_{jk}
    Gamma_tilde_ijk = christoffel_conformal(fd, tilde_gamma, tilde_gamma_inv)  # (i,j,k,...)

    # Build Gamma in (k,i,j,...) form for Ricci formula convenience
    Gamma_tilde_kij = Gamma_tilde_ijk.permute(0, 1, 2, *range(3, Gamma_tilde_ijk.ndim))  # already (i,j,k,...)
    # We need (k,i,j,...): just swap indices: Gamma^k_{ij} means first index is k
    Gamma_tilde_kij = Gamma_tilde_ijk  # since i is already "upper" index in our storage

    Rtilde = ricci_from_christoffel(fd, Gamma_tilde_kij)

    # Now R^{phi}_{ij}
    dphi = fd.grad_scalar(phi)  # (3,...)
    # conformal covariant second derivative: \tilde{D}_i \tilde{D}_j phi = ∂_i∂_j phi - \tilde{Gamma}^k_{ij} ∂_k phi
    # compute second partials ∂i∂j phi
    d2phi = torch.zeros((3,3) + phi.shape, device=phi.device, dtype=phi.dtype)
    for i in range(3):
        for j in range(3):
            d2phi[i, j] = fd.d1(fd.d1(phi, axis=j), axis=i)

    Dtilde2_phi = torch.zeros_like(d2phi)
    for i in range(3):
        for j in range(3):
            corr = 0.0
            for k in range(3):
                corr = corr + Gamma_tilde_ijk[k, i, j] * dphi[k]
            Dtilde2_phi[i, j] = d2phi[i, j] - corr

    # \tilde{D}^k \tilde{D}_k phi = \tilde{gamma}^{ij} \tilde{D}_i \tilde{D}_j phi
    Dtilde_lap_phi = torch.einsum("ij...,ij...->...", tilde_gamma_inv, Dtilde2_phi)

    # contractions of gradients
    dphi_sq = torch.einsum("ij...,i...,j...->...", tilde_gamma_inv, dphi, dphi)

    Rphi = torch.zeros((3,3) + phi.shape, device=phi.device, dtype=phi.dtype)
    for i in range(3):
        for j in range(3):
            Rphi[i, j] = (
                -2.0 * Dtilde2_phi[i, j]
                -2.0 * tilde_gamma[i, j] * Dtilde_lap_phi
                +4.0 * dphi[i] * dphi[j]
                -4.0 * tilde_gamma[i, j] * dphi_sq
            )

    R = Rtilde + Rphi
    return R, Rtilde


# ============================================================
# 3) Covariant second derivative of alpha with physical connection
# ============================================================

def D2_alpha_physical(fd, alpha: torch.Tensor,
                      phi: torch.Tensor,
                      tilde_gamma: torch.Tensor,
                      tilde_gamma_inv: torch.Tensor,
                      Gamma_tilde_ijk: torch.Tensor) -> torch.Tensor:
    """
    Compute physical Hessian:
        D_i D_j alpha = ∂_i∂_j alpha - Γ^k_{ij} ∂_k alpha

    We build Γ^k_{ij} from:
        Γ^k_{ij} = \tilde{Γ}^k_{ij} + 2(δ^k_i ∂_j φ + δ^k_j ∂_i φ - \tilde{γ}_{ij} \tilde{γ}^{kl} ∂_l φ)

    Returns:
        D2a: (3,3,...)
    """
    dalpha = fd.grad_scalar(alpha)  # (3,...)
    dphi = fd.grad_scalar(phi)      # (3,...)

    Gamma_phys = christoffel_physical(
        tilde_Gamma=None,  # not needed here
        tilde_gamma=tilde_gamma,
        tilde_gamma_inv=tilde_gamma_inv,
        dphi=dphi,
        Gamma_tilde_ijk=Gamma_tilde_ijk
    )  # (k,i,j,...)=Γ^k_{ij} (same storage layout)

    # second partials
    d2a = torch.zeros((3,3) + alpha.shape, device=alpha.device, dtype=alpha.dtype)
    for i in range(3):
        for j in range(3):
            d2a[i, j] = fd.d1(fd.d1(alpha, axis=j), axis=i)

    # covariant: d2a - Γ^k_{ij} ∂_k alpha
    D2a = torch.zeros_like(d2a)
    for i in range(3):
        for j in range(3):
            corr = 0.0
            for k in range(3):
                corr = corr + Gamma_phys[k, i, j] * dalpha[k]
            D2a[i, j] = d2a[i, j] - corr
    return D2a


# ============================================================
# 4) BSSN + gauge RHS builder
# ============================================================

@dataclass
class GaugeParams:
    """
    Gauge parameters for moving punctures:
    - 1+log slicing: ∂t α = -2 α K + L_β α  (we implement as: RHS = -2αK + β·∇α)
    - Gamma-driver:
        ∂t β^i = (3/4) B^i + L_β β^i
        ∂t B^i = ∂t \tilde{Γ}^i - η B^i + L_β B^i
    """
    eta: float = 2.0
    beta_driver_coeff: float = 0.75  # 3/4


def bssn_rhs_vacuum(
    fd,
    state,
    B: torch.Tensor,
    gauge: GaugeParams = GaugeParams(),
) -> Dict[str, torch.Tensor]:
    """
    Compute RHS for BSSN variables + gauge in vacuum.

    Inputs
    ------
    fd : FDOperators
    state : BSSNState
        Must contain: phi, tilde_gamma, K, tilde_A, tilde_Gamma, alpha, beta
    B : torch.Tensor
        Shift driver auxiliary field, shape (3,...)
    gauge : GaugeParams

    Returns
    -------
    rhs : dict of torch.Tensor
        Keys: 'phi','tilde_gamma','K','tilde_A','tilde_Gamma','alpha','beta','B'
    """
    phi = state.phi
    tg = state.tilde_gamma
    K = state.K
    tA = state.tilde_A
    tG = state.tilde_Gamma
    alpha = state.alpha
    beta = state.beta

    # Inverse conformal metric
    tg_inv = inv_3x3_field(tg)

    # Useful derivatives
    div_beta = trace_wrt_metric(fd.grad_vector(beta), torch.eye(3, device=beta.device, dtype=beta.dtype).view(3,3,1,1,1))
    # The line above uses a crude "trace" with δ^{ij}; for ∂_k β^k we want partial trace:
    # div_beta = ∂_k beta^k = sum_k ∂_k beta^k
    # Better compute explicitly:
    grad_beta = fd.grad_vector(beta)  # (i,k,...)=∂_k beta^i
    div_beta = grad_beta[0,0] + grad_beta[1,1] + grad_beta[2,2]

    # Lie derivatives
    Lb_phi = lie_scalar(fd, beta, phi)
    Lb_K = lie_scalar(fd, beta, K)
    Lb_alpha = lie_scalar(fd, beta, alpha)
    Lb_tG = lie_vector_contrav(fd, beta, tG)
    Lb_beta = lie_vector_contrav(fd, beta, beta)
    Lb_B = lie_vector_contrav(fd, beta, B)

    Lb_tg = lie_tensor_cov2(fd, beta, tg)
    Lb_tA = lie_tensor_cov2(fd, beta, tA)

    # ------------------------------------------------------------
    # Ricci tensor (physical) via BSSN split
    # ------------------------------------------------------------
    R, Rtilde = ricci_BSSN(fd, phi, tg, tg_inv)

    # Conformal Christoffels (need for D_i D_j alpha)
    Gamma_tilde_ijk = christoffel_conformal(fd, tg, tg_inv)

    # ------------------------------------------------------------
    # D_i D_j alpha and Laplacian D^i D_i alpha
    # ------------------------------------------------------------
    D2a = D2_alpha_physical(fd, alpha, phi, tg, tg_inv, Gamma_tilde_ijk)  # (3,3,...)
    # physical inverse metric: γ^{ij} = e^{-4φ} \tilde{γ}^{ij}
    em4phi = torch.exp(-4.0 * phi)
    gamma_inv = tg_inv * em4phi.unsqueeze(0).unsqueeze(0)
    lap_alpha = torch.einsum("ij...,ij...->...", gamma_inv, D2a)  # scalar (...)

    # ------------------------------------------------------------
    # Raise indices on tA: tA^{ij} = \tilde{γ}^{ik}\tilde{γ}^{jl} tA_{kl}
    # ------------------------------------------------------------
    tA_upup = torch.einsum("ik...,jl...,kl...->ij...", tg_inv, tg_inv, tA)

    # A_ij A^{ij} (using conformal metric)
    A2 = torch.einsum("ij...,ij...->...", tA, tA_upup)

    # ------------------------------------------------------------
    # RHS: phi
    # Standard BSSN:
    #   ∂t φ = - (1/6) α K + (1/6) ∂k β^k + β^k ∂k φ
    # We'll implement:
    #   RHS_phi = -(1/6) α K + (1/6) div_beta + L_beta φ
    # ------------------------------------------------------------
    rhs_phi = -(1.0 / 6.0) * alpha * K + (1.0 / 6.0) * div_beta + Lb_phi

    # ------------------------------------------------------------
    # RHS: tilde_gamma_ij
    #   ∂t \tilde{γ}_{ij} = -2α \tilde{A}_{ij} + L_β \tilde{γ}_{ij} - (2/3)\tilde{γ}_{ij}∂kβ^k
    # But our Lie derivative of covariant tensor already includes deformation terms;
    # the BSSN definition includes an extra traceless adjustment:
    # ------------------------------------------------------------
    rhs_tg = -2.0 * alpha.unsqueeze(0).unsqueeze(0) * tA + Lb_tg - (2.0/3.0) * tg * div_beta.unsqueeze(0).unsqueeze(0)

    # ------------------------------------------------------------
    # RHS: K
    #   ∂t K = - D^i D_i α + α( \tilde{A}_{ij}\tilde{A}^{ij} + (1/3)K^2 ) + L_β K
    # ------------------------------------------------------------
    rhs_K = -lap_alpha + alpha * (A2 + (1.0/3.0) * K * K) + Lb_K

    # ------------------------------------------------------------
    # RHS: tilde_A_ij
    #   ∂t \tilde{A}_{ij} =
    #       e^{-4φ} [ -D_iD_j α + α R_{ij} ]^{TF}
    #       + α ( K \tilde{A}_{ij} - 2 \tilde{A}_{ik}\tilde{A}^k{}_j )
    #       + L_β \tilde{A}_{ij} - (2/3)\tilde{A}_{ij} ∂kβ^k
    # ------------------------------------------------------------
    # Compute M_ij = -D_iD_j α + α R_ij
    M = -D2a + alpha.unsqueeze(0).unsqueeze(0) * R

    # TF with respect to conformal metric
    M_TF = tf_wrt_metric(M, tg, tg_inv)

    # Nonlinear term: \tilde{A}_{ik}\tilde{A}^k{}_j
    # First build mixed A^k{}_j = \tilde{γ}^{km}\tilde{A}_{mj}
    tA_mixed = torch.einsum("km...,mj...->kj...", tg_inv, tA)
    AikAkj = torch.einsum("ik...,kj...->ij...", tA, tA_mixed)

    rhs_tA = (torch.exp(-4.0 * phi).unsqueeze(0).unsqueeze(0) * M_TF
              + alpha.unsqueeze(0).unsqueeze(0) * (K.unsqueeze(0).unsqueeze(0) * tA - 2.0 * AikAkj)
              + Lb_tA
              - (2.0/3.0) * tA * div_beta.unsqueeze(0).unsqueeze(0))

    # ------------------------------------------------------------
    # RHS: tilde_Gamma^i
    # Standard BSSN:
    #   ∂t \tilde{Γ}^i =
    #     -2 ∂_j(α \tilde{A}^{ij})
    #     + 2α( \tilde{Γ}^i_{jk} \tilde{A}^{jk} + 6 \tilde{A}^{ij} ∂_j φ - (2/3)\tilde{γ}^{ij} ∂_j K )
    #     + β^j ∂_j \tilde{Γ}^i - \tilde{Γ}^j ∂_j β^i + (2/3)\tilde{Γ}^i ∂_j β^j
    #     + \tilde{γ}^{jk} ∂_j∂_k β^i + (1/3)\tilde{γ}^{ij} ∂_j∂_k β^k
    #
    # We implement this directly with FD ops.
    # ------------------------------------------------------------
    dphi = fd.grad_scalar(phi)  # (3,...)
    dK = fd.grad_scalar(K)      # (3,...)

    # termA: -2 ∂_j (α \tilde{A}^{ij})
    # build \tilde{A}^{ij}
    tA_upup = torch.einsum("ik...,jl...,kl...->ij...", tg_inv, tg_inv, tA)
    alphaA = alpha.unsqueeze(0).unsqueeze(0) * tA_upup
    # divergence over j: ∂_j(alphaA^{ij})
    termA = torch.zeros_like(tG)
    for i in range(3):
        div = 0.0
        for j in range(3):
            div = div + fd.d1(alphaA[i, j], axis=j)
        termA[i] = -2.0 * div

    # termB: 2α( \tilde{Γ}^i_{jk}\tilde{A}^{jk} + 6 \tilde{A}^{ij}∂_j φ - (2/3)\tilde{γ}^{ij}∂_j K )
    # contraction \tilde{Γ}^i_{jk}\tilde{A}^{jk}
    Gamma_tilde = Gamma_tilde_ijk  # (i,j,k,...)=\tilde{Γ}^i_{jk}
    contr1 = torch.zeros_like(tG)
    for i in range(3):
        s = 0.0
        for j in range(3):
            for k in range(3):
                s = s + Gamma_tilde[i, j, k] * tA_upup[j, k]
        contr1[i] = s

    # 6 \tilde{A}^{ij} ∂_j φ
    contr2 = torch.einsum("ij...,j...->i...", tA_upup, dphi)

    # (2/3)\tilde{γ}^{ij} ∂_j K
    contr3 = torch.einsum("ij...,j...->i...", tg_inv, dK)

    termB = 2.0 * alpha * (contr1 + 6.0 * contr2 - (2.0/3.0) * contr3)

    # Advection + shift-gradient terms
    # β^j ∂_j \tilde{Γ}^i is included in Lb_tG? careful:
    # Lb_tG (contravariant) = β^k ∂_k tG^i - tG^k ∂_k β^i
    # so we still need + (2/3) tG^i ∂_k β^k separately.
    termC = Lb_tG + (2.0/3.0) * tG * div_beta.unsqueeze(0)

    # Second derivative shift terms:
    # \tilde{γ}^{jk} ∂_j∂_k β^i + (1/3)\tilde{γ}^{ij} ∂_j∂_k β^k
    # First compute second partials of beta
    d2beta = torch.zeros((3,3,3) + beta.shape[1:], device=beta.device, dtype=beta.dtype)  # (i,j,k,...)=∂_j∂_k beta^i
    for i in range(3):
        for j in range(3):
            for k in range(3):
                d2beta[i, j, k] = fd.d1(fd.d1(beta[i], axis=k), axis=j)

    # termD1_i = \tilde{γ}^{jk} ∂_j∂_k β^i
    termD1 = torch.zeros_like(tG)
    for i in range(3):
        termD1[i] = torch.einsum("jk...,jk...->...", tg_inv, d2beta[i])

    # termD2_i = (1/3)\tilde{γ}^{ij} ∂_j∂_k β^k
    # compute S_j = ∂_j∂_k β^k
    # first div beta is ∂_k β^k; now take ∂_j of that:
    S = torch.zeros((3,) + beta.shape[1:], device=beta.device, dtype=beta.dtype)
    for j in range(3):
        S[j] = fd.d1(div_beta, axis=j)

    termD2 = (1.0/3.0) * torch.einsum("ij...,j...->i...", tg_inv, S)

    rhs_tG = termA + termB + termC + termD1 + termD2

    # ------------------------------------------------------------
    # Gauge RHS
    # 1+log:
    #   ∂t α = -2 α K + β^i ∂_i α  (we already computed Lb_alpha = β·∇α)
    # ------------------------------------------------------------
    rhs_alpha = -2.0 * alpha * K + Lb_alpha

    # Gamma-driver:
    #   ∂t β^i = (3/4) B^i + L_β β^i
    #   ∂t B^i = ∂t \tilde{Γ}^i - η B^i + L_β B^i
    coeff = gauge.beta_driver_coeff
    rhs_beta = coeff * B + Lb_beta
    rhs_B = rhs_tG - gauge.eta * B + Lb_B

    return {
        "phi": rhs_phi,
        "tilde_gamma": rhs_tg,
        "K": rhs_K,
        "tilde_A": rhs_tA,
        "tilde_Gamma": rhs_tG,
        "alpha": rhs_alpha,
        "beta": rhs_beta,
        "B": rhs_B,
    }
```

```python
"""
Block 4 (LOW-MEMORY) — BSSNOK + Gauge RHS (Torch) with streamed derivatives
============================================================================

Goal
----
Same equations as Block 4, but rewritten to reduce peak memory by:

1) NEVER materializing 27-component fields like:
     - grad_tensor2(tilde_gamma)   -> (3,3,3,...)
     - d2beta                      -> (3,3,3,...)
     - Christoffel tensor          -> (3,3,3,...)
2) Computing derivatives "on the fly" and contracting immediately.
3) Deleting large temporaries ASAP.

This greatly reduces memory at the cost of more Python loops (more compute time).
For early correctness + feasibility on modest GPUs/RAM, this is a good trade.

Assumptions
-----------
- dim=3 (full 3D). For dim<3 we can adapt later.
- Vacuum (no matter).
- You already ran:
    Block 1: cfg
    Block 2: state (BSSNState)
    Block 3: fd  (FDOperators)
    Block 3.5: optional KO (not used here but can be added)

Returned RHS keys:
    'phi','tilde_gamma','K','tilde_A','tilde_Gamma','alpha','beta','B'
"""

from dataclasses import dataclass
from typing import Dict
import torch


# ---------------------------------------------------------------------
# Small tensor utilities (same spirit as before, but keep memory small)
# ---------------------------------------------------------------------

def det_3x3_field(g: torch.Tensor) -> torch.Tensor:
    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]
    return a*(e*i - f*h) - b*(d*i - f*g2) + c*(d*h - e*g2)

def inv_3x3_field(g: torch.Tensor, eps: float = 1e-30) -> torch.Tensor:
    detg = det_3x3_field(g)
    detg_safe = detg + eps

    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]

    C00 = (e*i - f*h)
    C01 = -(d*i - f*g2)
    C02 = (d*h - e*g2)

    C10 = -(b*i - c*h)
    C11 = (a*i - c*g2)
    C12 = -(a*h - b*g2)

    C20 = (b*f - c*e)
    C21 = -(a*f - c*d)
    C22 = (a*e - b*d)

    adj = torch.stack([
        torch.stack([C00, C10, C20], dim=0),
        torch.stack([C01, C11, C21], dim=0),
        torch.stack([C02, C12, C22], dim=0),
    ], dim=0)
    return adj / detg_safe.unsqueeze(0).unsqueeze(0)

def trace_wrt_metric(T: torch.Tensor, g_inv: torch.Tensor) -> torch.Tensor:
    return torch.einsum("ij...,ij...->...", g_inv, T)

def tf_wrt_metric(T: torch.Tensor, g: torch.Tensor, g_inv: torch.Tensor) -> torch.Tensor:
    tr = trace_wrt_metric(T, g_inv)
    return T - (1.0/3.0) * g * tr.unsqueeze(0).unsqueeze(0)

def lie_scalar(fd, beta: torch.Tensor, f: torch.Tensor) -> torch.Tensor:
    grad_f = fd.grad_scalar(f)  # (3,...)
    out = torch.einsum("k...,k...->...", beta, grad_f)
    # free
    del grad_f
    return out

def lie_vector_contrav(fd, beta: torch.Tensor, v: torch.Tensor) -> torch.Tensor:
    grad_v = fd.grad_vector(v)       # (i,k,...)
    grad_beta = fd.grad_vector(beta) # (i,k,...)
    adv = torch.einsum("k...,ik...->i...", beta, grad_v)
    deform = torch.einsum("k...,ik...->i...", v, grad_beta)
    out = adv - deform
    del grad_v, grad_beta, adv, deform
    return out

def lie_tensor_cov2(fd, beta: torch.Tensor, T: torch.Tensor) -> torch.Tensor:
    # WARNING: grad_tensor2 is heavy (27 components). We avoid it.
    # We'll stream L_beta(T)_ij = beta^k ∂_k T_ij + T_ik ∂_j beta^k + T_kj ∂_i beta^k
    grad_beta = fd.grad_vector(beta)  # (k,j,...)=∂_j beta^k (stored as (k,j,...) in our convention? Actually grad_vector returns (i,k)=∂_k v^i
    # so grad_beta[k, j] = ∂_j beta^k
    shape = T.shape[2:]
    out = torch.zeros_like(T)

    # precompute advective piece beta^k ∂_k T_ij without building full ∂_k T_ij tensor:
    for i in range(3):
        for j in range(3):
            adv = 0.0
            for k in range(3):
                adv = adv + beta[k] * fd.d1(T[i, j], axis=k)
            term1 = 0.0
            term2 = 0.0
            # T_ik ∂_j beta^k
            for k in range(3):
                term1 = term1 + T[i, k] * grad_beta[k, j]
                term2 = term2 + T[k, j] * grad_beta[k, i]
            out[i, j] = adv + term1 + term2

    del grad_beta
    return out


# ---------------------------------------------------------------------
# Streamed Christoffel computation: \tilde{Gamma}^k_{ij} (scalar field)
# ---------------------------------------------------------------------

def gamma_tilde_kij_stream(fd, tg: torch.Tensor, tg_inv: torch.Tensor, k: int, i: int, j: int) -> torch.Tensor:
    """
    Return conformal Christoffel component \tilde{Gamma}^k_{ij} as a scalar field.
    Computed on the fly:
        1/2 * tg_inv[k,l] ( ∂_i tg[l,j] + ∂_j tg[l,i] - ∂_l tg[i,j] )
    """
    term = 0.0
    for l in range(3):
        d_i_glj = fd.d1(tg[l, j], axis=i)
        d_j_gli = fd.d1(tg[l, i], axis=j)
        d_l_gij = fd.d1(tg[i, j], axis=l)
        term = term + tg_inv[k, l] * (d_i_glj + d_j_gli - d_l_gij)
        # free inner temps quickly
        del d_i_glj, d_j_gli, d_l_gij
    return 0.5 * term


# ---------------------------------------------------------------------
# Streamed conformal Ricci: \tilde{R}_{ij} (BSSN-friendly form)
# ---------------------------------------------------------------------

def ricci_tilde_stream(fd, tg: torch.Tensor, tg_inv: torch.Tensor, tG: torch.Tensor) -> torch.Tensor:
    """
    Compute conformal Ricci tensor \tilde{R}_{ij} in a BSSN-friendly way without
    storing 27-component fields.

    We use a standard identity (used in many BSSN implementations):

    \tilde{R}_{ij} =
      - 1/2 * \tilde{\gamma}^{lm} ∂_l∂_m \tilde{\gamma}_{ij}
      + \tilde{\gamma}_{k(i} ∂_{j)} \tilde{\Gamma}^k
      + \tilde{\Gamma}^k \tilde{\Gamma}_{(ij)k}
      + \tilde{\gamma}^{lm} ( 2 \tilde{\Gamma}^k_{l(i} \tilde{\Gamma}_{j)km} + \tilde{\Gamma}^k_{im} \tilde{\Gamma}_{klj} )

    where \tilde{\Gamma}_{ijk} = \tilde{\gamma}_{il} \tilde{\Gamma}^l_{jk}.

    This is compute-heavy but memory-light.
    """
    shape = tg.shape[2:]
    Rtil = torch.zeros((3,3) + shape, device=tg.device, dtype=tg.dtype)

    # Precompute derivatives of contracted Gamma: ∂_j \tilde{\Gamma}^k  (vector gradient = 9 components only)
    dtG = fd.grad_vector(tG)  # (k, j, ...)=∂_j tG^k  (since grad_vector returns (i,k)=∂_k v^i, so dtG[k,j]=∂_j tG^k)

    # Precompute \tilde{gamma}^{lm} once (already passed)
    # Compute term-by-term for each (i,j)
    for i in range(3):
        for j in range(3):

            # Term1: -1/2 * tg_inv^{lm} ∂_l∂_m tg_ij
            # We'll compute contracted second derivatives directly:
            # S = tg_inv^{lm} ∂_l∂_m tg_ij
            S = 0.0
            for l in range(3):
                for m in range(3):
                    d2 = fd.d1(fd.d1(tg[i, j], axis=m), axis=l)
                    S = S + tg_inv[l, m] * d2
                    del d2
            term1 = -0.5 * S
            del S

            # Term2: tg_{k(i} ∂_{j)} tG^k = 1/2 [ tg_{k i} ∂_j tG^k + tg_{k j} ∂_i tG^k ]
            term2 = 0.0
            for k in range(3):
                term2 = term2 + 0.5 * (tg[k, i] * dtG[k, j] + tg[k, j] * dtG[k, i])

            # Term3: tG^k * \tilde{Gamma}_{(ij)k}
            # First compute Gamma_{ijk} = tg_{i l} Gamma^l_{jk} on the fly.
            # Need sym in (i,j): Gamma_{(ij)k} = 1/2 (Gamma_{ijk} + Gamma_{jik})
            term3 = 0.0
            for k in range(3):
                # Gamma_{ijk}:
                G_ijk = 0.0
                G_jik = 0.0
                for l in range(3):
                    Gam_ljk = gamma_tilde_kij_stream(fd, tg, tg_inv, l, j, k)  # \tilde{Gamma}^l_{jk}
                    Gam_lik = gamma_tilde_kij_stream(fd, tg, tg_inv, l, i, k)  # \tilde{Gamma}^l_{ik}
                    G_ijk = G_ijk + tg[i, l] * Gam_ljk
                    G_jik = G_jik + tg[j, l] * Gam_lik
                    del Gam_ljk, Gam_lik
                G_sym = 0.5 * (G_ijk + G_jik)
                term3 = term3 + tG[k] * G_sym
                del G_ijk, G_jik, G_sym

            # Term4: tg_inv^{lm} ( 2 Γ^k_{l(i} Γ_{j)km} + Γ^k_{im} Γ_{klj} )
            term4 = 0.0
            for l in range(3):
                for m in range(3):
                    # accumulate over k
                    inner = 0.0
                    for k in range(3):
                        # Γ^k_{l i} and Γ^k_{l j}
                        Gam_kli = gamma_tilde_kij_stream(fd, tg, tg_inv, k, l, i)
                        Gam_klj = gamma_tilde_kij_stream(fd, tg, tg_inv, k, l, j)

                        # Γ_{j k m} = tg_{j a} Γ^a_{k m}
                        G_jkm = 0.0
                        G_ikm = 0.0
                        for a in range(3):
                            Gam_akm = gamma_tilde_kij_stream(fd, tg, tg_inv, a, k, m)
                            G_jkm = G_jkm + tg[j, a] * Gam_akm
                            G_ikm = G_ikm + tg[i, a] * Gam_akm
                            del Gam_akm

                        # First piece: 2 Γ^k_{l(i} Γ_{j) k m} = Γ^k_{l i} Γ_{j k m} + Γ^k_{l j} Γ_{i k m}
                        piece1 = Gam_kli * G_jkm + Gam_klj * G_ikm

                        # Second piece: Γ^k_{i m} Γ_{k l j}
                        Gam_kim = gamma_tilde_kij_stream(fd, tg, tg_inv, k, i, m)
                        # Γ_{k l j} = tg_{k a} Γ^a_{l j}
                        G_klj = 0.0
                        for a in range(3):
                            Gam_alj = gamma_tilde_kij_stream(fd, tg, tg_inv, a, l, j)
                            G_klj = G_klj + tg[k, a] * Gam_alj
                            del Gam_alj

                        piece2 = Gam_kim * G_klj

                        inner = inner + piece1 + piece2

                        del Gam_kli, Gam_klj, G_jkm, G_ikm, piece1, Gam_kim, G_klj, piece2

                    term4 = term4 + tg_inv[l, m] * inner
                    del inner

            Rtil[i, j] = term1 + term2 + term3 + term4

            # free per-(i,j) terms
            del term1, term2, term3, term4

    del dtG
    return Rtil


# ---------------------------------------------------------------------
# Phi-curvature piece: R^phi_{ij}
# ---------------------------------------------------------------------

def ricci_phi_stream(fd, phi: torch.Tensor, tg: torch.Tensor, tg_inv: torch.Tensor) -> torch.Tensor:
    """
    Compute:
      R^phi_{ij} = -2 \tilde{D}_i \tilde{D}_j phi
                  -2 \tilde{gamma}_{ij} \tilde{D}^k \tilde{D}_k phi
                  +4 (∂_i phi)(∂_j phi)
                  -4 \tilde{gamma}_{ij} (∂^k phi ∂_k phi)
    without storing full Gamma_tilde tensor.
    """
    shape = phi.shape
    dphi = fd.grad_scalar(phi)  # (3,...)

    # Build \tilde{D}_i \tilde{D}_j phi streamed:
    D2phi = torch.zeros((3,3) + shape, device=phi.device, dtype=phi.dtype)
    for i in range(3):
        for j in range(3):
            # partials
            d2 = fd.d1(fd.d1(phi, axis=j), axis=i)
            # subtract Gamma^k_{ij} ∂_k phi
            corr = 0.0
            for k in range(3):
                Gam_kij = gamma_tilde_kij_stream(fd, tg, tg_inv, k, i, j)
                corr = corr + Gam_kij * dphi[k]
                del Gam_kij
            D2phi[i, j] = d2 - corr
            del d2, corr

    lap_phi = torch.einsum("ij...,ij...->...", tg_inv, D2phi)
    dphi_sq = torch.einsum("ij...,i...,j...->...", tg_inv, dphi, dphi)

    Rphi = torch.zeros_like(D2phi)
    for i in range(3):
        for j in range(3):
            Rphi[i, j] = (
                -2.0 * D2phi[i, j]
                -2.0 * tg[i, j] * lap_phi
                +4.0 * dphi[i] * dphi[j]
                -4.0 * tg[i, j] * dphi_sq
            )

    del dphi, D2phi, lap_phi, dphi_sq
    return Rphi


# ---------------------------------------------------------------------
# Physical Hessian D_i D_j alpha (needs physical Gamma)
# ---------------------------------------------------------------------

def D2_alpha_physical_stream(fd, alpha: torch.Tensor, phi: torch.Tensor, tg: torch.Tensor, tg_inv: torch.Tensor) -> torch.Tensor:
    """
    D_i D_j alpha = ∂_i∂_j alpha - Γ^k_{ij} ∂_k alpha
    with Γ^k_{ij} = \tilde{Γ}^k_{ij} + 2(δ^k_i ∂_j φ + δ^k_j ∂_i φ - \tilde{γ}_{ij}\tilde{γ}^{kl} ∂_l φ)
    """
    dalpha = fd.grad_scalar(alpha)  # (3,...)
    dphi = fd.grad_scalar(phi)      # (3,...)
    V = torch.einsum("kl...,l...->k...", tg_inv, dphi)  # V^k = tg_inv^{kl} dphi_l

    shape = alpha.shape
    D2a = torch.zeros((3,3) + shape, device=alpha.device, dtype=alpha.dtype)

    for i in range(3):
        for j in range(3):
            d2 = fd.d1(fd.d1(alpha, axis=j), axis=i)
            corr = 0.0
            for k in range(3):
                Gam_tilde = gamma_tilde_kij_stream(fd, tg, tg_inv, k, i, j)
                delta_ki = 1.0 if k == i else 0.0
                delta_kj = 1.0 if k == j else 0.0
                Gam_phys = Gam_tilde + 2.0 * (delta_ki * dphi[j] + delta_kj * dphi[i] - tg[i, j] * V[k])
                corr = corr + Gam_phys * dalpha[k]
                del Gam_tilde, Gam_phys
            D2a[i, j] = d2 - corr
            del d2, corr

    del dalpha, dphi, V
    return D2a


# ---------------------------------------------------------------------
# Gauge params
# ---------------------------------------------------------------------

@dataclass
class GaugeParams:
    eta: float = 2.0
    beta_driver_coeff: float = 0.75  # 3/4


# ---------------------------------------------------------------------
# Block 4 low-memory RHS
# ---------------------------------------------------------------------

def bssn_rhs_vacuum_lowmem(fd, state, B: torch.Tensor, gauge: GaugeParams = GaugeParams()) -> Dict[str, torch.Tensor]:
    """
    Low-memory BSSN + moving puncture gauge RHS.
    """
    assert state.phi.ndim == 3, "This lowmem RHS assumes dim=3 scalar fields (Nx,Ny,Nz)."

    phi = state.phi
    tg = state.tilde_gamma
    K = state.K
    tA = state.tilde_A
    tG = state.tilde_Gamma
    alpha = state.alpha
    beta = state.beta

    # Inverse conformal metric
    tg_inv = inv_3x3_field(tg)

    # grad beta for div_beta (only 9 comps; OK)
    grad_beta = fd.grad_vector(beta)  # (i,k,...)=∂_k beta^i
    div_beta = grad_beta[0,0] + grad_beta[1,1] + grad_beta[2,2]

    # Lie terms (memory-light except grad_vector inside some; still small)
    Lb_phi = lie_scalar(fd, beta, phi)
    Lb_K = lie_scalar(fd, beta, K)
    Lb_alpha = lie_scalar(fd, beta, alpha)
    Lb_tG = lie_vector_contrav(fd, beta, tG)
    Lb_beta = lie_vector_contrav(fd, beta, beta)
    Lb_B = lie_vector_contrav(fd, beta, B)

    # Lie of tensors streamed (no 27-component allocation)
    Lb_tg = lie_tensor_cov2(fd, beta, tg)
    Lb_tA = lie_tensor_cov2(fd, beta, tA)

    # -----------------------------------------------------------------
    # Ricci: R_ij = Rtilde_ij + Rphi_ij (streamed, heavy compute, low mem)
    # -----------------------------------------------------------------
    Rtilde = ricci_tilde_stream(fd, tg, tg_inv, tG)
    Rphi = ricci_phi_stream(fd, phi, tg, tg_inv)
    R = Rtilde + Rphi
    del Rphi  # keep Rtilde if you want diagnostics later

    # -----------------------------------------------------------------
    # Physical Hessian of alpha and Laplacian
    # -----------------------------------------------------------------
    D2a = D2_alpha_physical_stream(fd, alpha, phi, tg, tg_inv)
    em4phi = torch.exp(-4.0 * phi)
    gamma_inv = tg_inv * em4phi.unsqueeze(0).unsqueeze(0)
    lap_alpha = torch.einsum("ij...,ij...->...", gamma_inv, D2a)

    # -----------------------------------------------------------------
    # Raise indices on tA and compute A_ij A^{ij}
    # -----------------------------------------------------------------
    tA_upup = torch.einsum("ik...,jl...,kl...->ij...", tg_inv, tg_inv, tA)
    A2 = torch.einsum("ij...,ij...->...", tA, tA_upup)

    # -----------------------------------------------------------------
    # RHS phi
    # -----------------------------------------------------------------
    rhs_phi = -(1.0/6.0) * alpha * K + (1.0/6.0) * div_beta + Lb_phi

    # -----------------------------------------------------------------
    # RHS tilde_gamma
    # -----------------------------------------------------------------
    rhs_tg = -2.0 * alpha.unsqueeze(0).unsqueeze(0) * tA + Lb_tg - (2.0/3.0) * tg * div_beta.unsqueeze(0).unsqueeze(0)

    # -----------------------------------------------------------------
    # RHS K
    # -----------------------------------------------------------------
    rhs_K = -lap_alpha + alpha * (A2 + (1.0/3.0) * K * K) + Lb_K

    # -----------------------------------------------------------------
    # RHS tilde_A
    # -----------------------------------------------------------------
    M = -D2a + alpha.unsqueeze(0).unsqueeze(0) * R
    M_TF = tf_wrt_metric(M, tg, tg_inv)

    tA_mixed = torch.einsum("km...,mj...->kj...", tg_inv, tA)     # A^k{}_j
    AikAkj = torch.einsum("ik...,kj...->ij...", tA, tA_mixed)     # A_{ik} A^k{}_j

    rhs_tA = (
        torch.exp(-4.0 * phi).unsqueeze(0).unsqueeze(0) * M_TF
        + alpha.unsqueeze(0).unsqueeze(0) * (K.unsqueeze(0).unsqueeze(0) * tA - 2.0 * AikAkj)
        + Lb_tA
        - (2.0/3.0) * tA * div_beta.unsqueeze(0).unsqueeze(0)
    )

    # Free big temps early
    del D2a, gamma_inv, lap_alpha, M, M_TF, tA_mixed, AikAkj, R

    # -----------------------------------------------------------------
    # RHS tilde_Gamma^i (avoid d2beta allocation)
    # -----------------------------------------------------------------
    dphi = fd.grad_scalar(phi)  # (3,...)
    dK = fd.grad_scalar(K)      # (3,...)

    # termA: -2 ∂_j(α \tilde{A}^{ij})
    alphaA = alpha.unsqueeze(0).unsqueeze(0) * tA_upup
    termA = torch.zeros_like(tG)
    for i in range(3):
        div = 0.0
        for j in range(3):
            div = div + fd.d1(alphaA[i, j], axis=j)
        termA[i] = -2.0 * div
        del div

    # termB: 2α( \tilde{Γ}^i_{jk}\tilde{A}^{jk} + 6 \tilde{A}^{ij}∂_j φ - (2/3)\tilde{γ}^{ij}∂_j K )
    # contr1_i = \tilde{Γ}^i_{jk}\tilde{A}^{jk} streamed
    contr1 = torch.zeros_like(tG)
    for i in range(3):
        s = 0.0
        for j in range(3):
            for k in range(3):
                Gam_ijk = gamma_tilde_kij_stream(fd, tg, tg_inv, i, j, k)  # \tilde{Γ}^i_{jk}
                s = s + Gam_ijk * tA_upup[j, k]
                del Gam_ijk
        contr1[i] = s
        del s

    contr2 = torch.einsum("ij...,j...->i...", tA_upup, dphi)
    contr3 = torch.einsum("ij...,j...->i...", tg_inv, dK)
    termB = 2.0 * alpha * (contr1 + 6.0 * contr2 - (2.0/3.0) * contr3)

    # termC: L_beta tG + (2/3) tG div_beta
    termC = Lb_tG + (2.0/3.0) * tG * div_beta.unsqueeze(0)

    # termD1: tg_inv^{jk} ∂_j∂_k beta^i  (streamed)
    termD1 = torch.zeros_like(tG)
    for i in range(3):
        acc = 0.0
        for j in range(3):
            for k in range(3):
                d2 = fd.d1(fd.d1(beta[i], axis=k), axis=j)
                acc = acc + tg_inv[j, k] * d2
                del d2
        termD1[i] = acc
        del acc

    # termD2: (1/3) tg_inv^{ij} ∂_j∂_k beta^k
    # Compute S_j = ∂_j(div_beta)
    S = torch.zeros((3,) + div_beta.shape, device=div_beta.device, dtype=div_beta.dtype)
    for j in range(3):
        S[j] = fd.d1(div_beta, axis=j)
    termD2 = (1.0/3.0) * torch.einsum("ij...,j...->i...", tg_inv, S)
    del S

    rhs_tG = termA + termB + termC + termD1 + termD2

    # Free temps
    del dphi, dK, alphaA, termA, contr1, contr2, contr3, termB, termC, termD1, termD2, tA_upup

    # -----------------------------------------------------------------
    # Gauge RHS
    # -----------------------------------------------------------------
    rhs_alpha = -2.0 * alpha * K + Lb_alpha

    coeff = gauge.beta_driver_coeff
    rhs_beta = coeff * B + Lb_beta
    rhs_B = rhs_tG - gauge.eta * B + Lb_B

    # Final cleanup (some of these might still be referenced; safe to keep minimal)
    del tg_inv, grad_beta, div_beta, Lb_phi, Lb_K, Lb_alpha, Lb_tG, Lb_beta, Lb_B, Lb_tg, Lb_tA

    return {
        "phi": rhs_phi,
        "tilde_gamma": rhs_tg,
        "K": rhs_K,
        "tilde_A": rhs_tA,
        "tilde_Gamma": rhs_tG,
        "alpha": rhs_alpha,
        "beta": rhs_beta,
        "B": rhs_B,
        # Optional: expose Rtilde for debugging if you want (commented to reduce memory traffic)
        # "Rtilde": Rtilde,
    }


# ---------------------------------------------------------------------
# Example call
# ---------------------------------------------------------------------
# Ensure you have B_driver initialized (shift driver auxiliary)
if "B_driver" not in globals():
    # choose either state_bl or state_flat depending on what's available
    _state = state_bl if "state_bl" in globals() else state_flat
    B_driver = torch.zeros_like(_state.beta)

state = state_bl if "state_bl" in globals() else state_flat

rhs_lowmem = bssn_rhs_vacuum_lowmem(
    fd=fd,
    state=state,
    B=B_driver,
    gauge=GaugeParams(eta=2.0, beta_driver_coeff=0.75)
)

print("LOWMEM RHS keys:", list(rhs_lowmem.keys()))
print("LOWMEM RHS phi mean |.|  =", float(rhs_lowmem["phi"].abs().mean().item()))
print("LOWMEM RHS K mean |.|    =", float(rhs_lowmem["K"].abs().mean().item()))
print("LOWMEM RHS alpha mean |.|=", float(rhs_lowmem["alpha"].abs().mean().item()))
```

Block 5 — Time Integration (RK4 / Method of Lines) using Block 4 (first version)
---
We integrate the BSSN+gauge system:
    d/dt u = RHS(u)
with the RHS computed by `bssn_rhs_vacuum` (the FIRST / non-lowmem Block 4).

State variables evolved:
    - phi, tilde_gamma_ij, K, tilde_A_ij, tilde_Gamma^i
    - alpha, beta^i
    - B^i   (Gamma-driver auxiliary)

Features included here:
-----------------------
1) Classic RK4 integrator (MoL).
2) Optional KO dissipation injection (from Block 3.5) as:
       RHS <- RHS + KO(u)
3) Lightweight diagnostics each N steps:
    - mean/maximum of |phi|, |K|
    - det(tilde_gamma) deviation from 1 (should stay near 1)
    - trace(tilde_A) (should stay near 0)
4) Safety hooks (optional):
    - NaN/Inf detection
    - early stop if blow-up

Important:
----------
- This integrates one step at a time; you can wrap it in longer loops.
- For initial experiments, use small grids (e.g. 64^3) and float64.
- Boundary conditions: whatever you chose in fd (Block 3). Periodic is easiest.

Prerequisites:
--------------
- Block 1: cfg
- Block 2: state (state_flat or state_bl)
- Block 3: fd = FDOperators(cfg, ...)
- Block 4: bssn_rhs_vacuum(...) (first version) must be defined
- Optional Block 3.5: ko = KODissipation(...) if you want dissipation
"""

```python

# ------------------------------------------------------------
# 0) Helpers: state algebra (RK updates) and projections
# ------------------------------------------------------------

def state_add_scaled(state, k: Dict[str, torch.Tensor], a: float):
    """
    Return a NEW state = state + a * k, for RK stages.
    """
    # Note: we build a lightweight "state-like" object with the same attributes.
    # This avoids rewriting BSSNState dataclass here.
    class _Tmp:
        pass

    out = _Tmp()
    out.phi = state.phi + a * k["phi"]
    out.tilde_gamma = state.tilde_gamma + a * k["tilde_gamma"]
    out.K = state.K + a * k["K"]
    out.tilde_A = state.tilde_A + a * k["tilde_A"]
    out.tilde_Gamma = state.tilde_Gamma + a * k["tilde_Gamma"]
    out.alpha = state.alpha + a * k["alpha"]
    out.beta = state.beta + a * k["beta"]
    # keep psi optional (recomputable)
    out.psi = None
    return out

def vector_add_scaled(v: torch.Tensor, k: torch.Tensor, a: float) -> torch.Tensor:
    """Return v + a*k (used for B-driver)."""
    return v + a * k

def det_3x3_field(g: torch.Tensor) -> torch.Tensor:
    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]
    return a*(e*i - f*h) - b*(d*i - f*g2) + c*(d*h - e*g2)

def inv_3x3_field(g: torch.Tensor, eps: float = 1e-30) -> torch.Tensor:
    detg = det_3x3_field(g)
    detg_safe = detg + eps

    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]

    C00 = (e*i - f*h)
    C01 = -(d*i - f*g2)
    C02 = (d*h - e*g2)

    C10 = -(b*i - c*h)
    C11 = (a*i - c*g2)
    C12 = -(a*h - b*g2)

    C20 = (b*f - c*e)
    C21 = -(a*f - c*d)
    C22 = (a*e - b*d)

    adj = torch.stack([
        torch.stack([C00, C10, C20], dim=0),
        torch.stack([C01, C11, C21], dim=0),
        torch.stack([C02, C12, C22], dim=0),
    ], dim=0)
    return adj / detg_safe.unsqueeze(0).unsqueeze(0)

def trace_wrt_metric(T: torch.Tensor, g_inv: torch.Tensor) -> torch.Tensor:
    return torch.einsum("ij...,ij...->...", g_inv, T)

def enforce_bssn_algebraic_constraints(state, eps_det: float = 1e-30):
    """
    Enforce the algebraic BSSN conditions:
    1) det(tilde_gamma) = 1  (by rescaling)
    2) trace(tilde_A) = 0   (by TF projection)

    This is common practice and helps numerical stability.

    NOTE:
    - This is a "projection" step and can slightly alter the PDE solution,
      but it's standard and very useful early on.
    """
    tg = state.tilde_gamma
    tA = state.tilde_A

    det_tg = det_3x3_field(tg)  # (...)
    # Rescale: tg_ij <- det^{-1/3} tg_ij
    # Because det(s * tg) = s^3 det(tg), choose s = det^{-1/3}.
    s = torch.pow(det_tg + eps_det, -1.0/3.0)
    tg = tg * s.unsqueeze(0).unsqueeze(0)

    tg_inv = inv_3x3_field(tg, eps=eps_det)

    # Trace-free projection of tilde_A
    trA = trace_wrt_metric(tA, tg_inv)
    tA = tA - (1.0/3.0) * tg * trA.unsqueeze(0).unsqueeze(0)

    # Write back
    state.tilde_gamma = tg
    state.tilde_A = tA

    # Free
    del det_tg, s, tg_inv, trA


# ------------------------------------------------------------
# 1) Optional KO injection
# ------------------------------------------------------------

@dataclass
class DissipationConfig:
    enabled: bool = False
    # If enabled, you must have a `ko` object in scope (Block 3.5), or pass it in the integrator.
    apply_to_phi: bool = True
    apply_to_K: bool = True
    apply_to_alpha: bool = True
    apply_to_beta: bool = True
    apply_to_tildeGamma: bool = True
    apply_to_tildeGamma_metric: bool = True
    apply_to_tildeA: bool = True


def add_KO_to_rhs(rhs: Dict[str, torch.Tensor], state, ko, diss_cfg: DissipationConfig):
    """
    Modify RHS in-place by adding KO dissipation terms.
    """
    if not diss_cfg.enabled:
        return rhs

    if diss_cfg.apply_to_phi:
        rhs["phi"] = rhs["phi"] + ko.ko_scalar(state.phi)
    if diss_cfg.apply_to_K:
        rhs["K"] = rhs["K"] + ko.ko_scalar(state.K)
    if diss_cfg.apply_to_alpha:
        rhs["alpha"] = rhs["alpha"] + ko.ko_scalar(state.alpha)
    if diss_cfg.apply_to_beta:
        rhs["beta"] = rhs["beta"] + ko.ko_vector(state.beta)
    if diss_cfg.apply_to_tildeGamma:
        rhs["tilde_Gamma"] = rhs["tilde_Gamma"] + ko.ko_vector(state.tilde_Gamma)
    if diss_cfg.apply_to_tildeGamma_metric:
        rhs["tilde_gamma"] = rhs["tilde_gamma"] + ko.ko_tensor2(state.tilde_gamma)
    if diss_cfg.apply_to_tildeA:
        rhs["tilde_A"] = rhs["tilde_A"] + ko.ko_tensor2(state.tilde_A)

    return rhs


# ------------------------------------------------------------
# 2) RK4 stepper for BSSN
# ------------------------------------------------------------

@dataclass
class IntegratorConfig:
    dt: float
    project_each_step: bool = True
    nan_check: bool = True
    dissipation: DissipationConfig = DissipationConfig()


def rk4_step_bssn(
    fd,
    state,
    B: torch.Tensor,
    cfg_int: IntegratorConfig,
    gauge_params,
    ko: Optional[object] = None,
):
    """
    Perform one RK4 step on (state, B) using the first Block 4 RHS.

    Returns updated (state, B) in-place style:
        state fields modified
        B modified and returned
    """
    dt = torch.tensor(cfg_int.dt, device=state.phi.device, dtype=state.phi.dtype)

    # --- Stage 1 ---
    k1 = bssn_rhs_vacuum(fd=fd, state=state, B=B, gauge=gauge_params)
    if cfg_int.dissipation.enabled:
        assert ko is not None, "KO enabled but ko object not provided."
        k1 = add_KO_to_rhs(k1, state, ko, cfg_int.dissipation)

    # --- Stage 2 ---
    s2 = state_add_scaled(state, k1, 0.5 * dt.item())
    B2 = vector_add_scaled(B, k1["B"], 0.5 * dt.item())
    k2 = bssn_rhs_vacuum(fd=fd, state=s2, B=B2, gauge=gauge_params)
    if cfg_int.dissipation.enabled:
        k2 = add_KO_to_rhs(k2, s2, ko, cfg_int.dissipation)

    # --- Stage 3 ---
    s3 = state_add_scaled(state, k2, 0.5 * dt.item())
    B3 = vector_add_scaled(B, k2["B"], 0.5 * dt.item())
    k3 = bssn_rhs_vacuum(fd=fd, state=s3, B=B3, gauge=gauge_params)
    if cfg_int.dissipation.enabled:
        k3 = add_KO_to_rhs(k3, s3, ko, cfg_int.dissipation)

    # --- Stage 4 ---
    s4 = state_add_scaled(state, k3, 1.0 * dt.item())
    B4 = vector_add_scaled(B, k3["B"], 1.0 * dt.item())
    k4 = bssn_rhs_vacuum(fd=fd, state=s4, B=B4, gauge=gauge_params)
    if cfg_int.dissipation.enabled:
        k4 = add_KO_to_rhs(k4, s4, ko, cfg_int.dissipation)

    # Combine RK4
    with torch.no_grad():
        state.phi = state.phi + (dt / 6.0) * (k1["phi"] + 2.0*k2["phi"] + 2.0*k3["phi"] + k4["phi"])
        state.tilde_gamma = state.tilde_gamma + (dt / 6.0) * (k1["tilde_gamma"] + 2.0*k2["tilde_gamma"] + 2.0*k3["tilde_gamma"] + k4["tilde_gamma"])
        state.K = state.K + (dt / 6.0) * (k1["K"] + 2.0*k2["K"] + 2.0*k3["K"] + k4["K"])
        state.tilde_A = state.tilde_A + (dt / 6.0) * (k1["tilde_A"] + 2.0*k2["tilde_A"] + 2.0*k3["tilde_A"] + k4["tilde_A"])
        state.tilde_Gamma = state.tilde_Gamma + (dt / 6.0) * (k1["tilde_Gamma"] + 2.0*k2["tilde_Gamma"] + 2.0*k3["tilde_Gamma"] + k4["tilde_Gamma"])
        state.alpha = state.alpha + (dt / 6.0) * (k1["alpha"] + 2.0*k2["alpha"] + 2.0*k3["alpha"] + k4["alpha"])
        state.beta = state.beta + (dt / 6.0) * (k1["beta"] + 2.0*k2["beta"] + 2.0*k3["beta"] + k4["beta"])

        B = B + (dt / 6.0) * (k1["B"] + 2.0*k2["B"] + 2.0*k3["B"] + k4["B"])

    # Algebraic constraint projection (recommended)
    if cfg_int.project_each_step:
        enforce_bssn_algebraic_constraints(state)

    # NaN/Inf check
    if cfg_int.nan_check:
        bad = (
            torch.isnan(state.phi).any() or torch.isinf(state.phi).any() or
            torch.isnan(state.K).any() or torch.isinf(state.K).any() or
            torch.isnan(state.alpha).any() or torch.isinf(state.alpha).any()
        )
        if bad:
            raise FloatingPointError("NaN/Inf detected after RK4 step.")

    # Explicitly delete stage objects to help the allocator (especially on GPU)
    del k1, k2, k3, k4, s2, s3, s4, B2, B3, B4, dt

    return state, B


# ------------------------------------------------------------
# 3) Diagnostics
# ------------------------------------------------------------

@torch.no_grad()
def diagnostics_basic(state) -> Dict[str, float]:
    """
    Lightweight diagnostics to monitor stability:
    - det(tilde_gamma) deviation from 1
    - trace(tilde_A) (should be ~0)
    - mean abs of some fields
    """
    tg = state.tilde_gamma
    det_tg = det_3x3_field(tg)
    det_dev = (det_tg - 1.0).abs()

    tg_inv = inv_3x3_field(tg)
    trA = trace_wrt_metric(state.tilde_A, tg_inv).abs()

    out = {
        "phi_mean_abs": float(state.phi.abs().mean().item()),
        "phi_max_abs": float(state.phi.abs().max().item()),
        "K_mean_abs": float(state.K.abs().mean().item()),
        "K_max_abs": float(state.K.abs().max().item()),
        "alpha_mean": float(state.alpha.mean().item()),
        "alpha_min": float(state.alpha.min().item()),
        "alpha_max": float(state.alpha.max().item()),
        "det_tg_mean_dev": float(det_dev.mean().item()),
        "det_tg_max_dev": float(det_dev.max().item()),
        "trA_mean_abs": float(trA.mean().item()),
        "trA_max_abs": float(trA.max().item()),
    }

    del det_tg, det_dev, tg_inv, trA
    return out


# ------------------------------------------------------------
# 4) Driver loop (run Nt steps)
# ------------------------------------------------------------

def evolve_bssn(
    fd,
    state,
    B: torch.Tensor,
    Nt: int,
    dt: float,
    gauge_params,
    print_every: int = 10,
    ko: Optional[object] = None,
    dissipation_enabled: bool = False,
):
    """
    Evolve for Nt steps with RK4.

    Parameters
    ----------
    state : BSSNState
    B : torch.Tensor (3,...)
    Nt : int
    dt : float
    gauge_params : GaugeParams (from Block 4)
    print_every : int
    ko : optional KODissipation
    dissipation_enabled : bool
    """
    int_cfg = IntegratorConfig(
        dt=dt,
        project_each_step=True,
        nan_check=True,
        dissipation=DissipationConfig(enabled=dissipation_enabled),
    )

    t = 0.0
    for n in range(Nt):
        state, B = rk4_step_bssn(fd, state, B, int_cfg, gauge_params, ko=ko)
        t += dt

        if (n % print_every) == 0 or (n == Nt - 1):
            d = diagnostics_basic(state)
            print(f"[step {n:05d}/{Nt-1:05d}] t={t:.6e} | "
                  f"phi|mean={d['phi_mean_abs']:.3e} max={d['phi_max_abs']:.3e} | "
                  f"K|mean={d['K_mean_abs']:.3e} max={d['K_max_abs']:.3e} | "
                  f"alpha[min,max]=({d['alpha_min']:.3e},{d['alpha_max']:.3e}) | "
                  f"det(tg) dev mean/max=({d['det_tg_mean_dev']:.3e},{d['det_tg_max_dev']:.3e}) | "
                  f"trA mean/max=({d['trA_mean_abs']:.3e},{d['trA_max_abs']:.3e})")

    return state, B


# ------------------------------------------------------------
# 5) Example: run a short evolution
# ------------------------------------------------------------

# Choose state (flat or puncture)
state = state_bl if "state_bl" in globals() else state_flat

# Initialize B-driver if not present
if "B_driver" not in globals():
    B_driver = torch.zeros_like(state.beta)

# Use dt from Block 1 (cfg.dt is a torch scalar)
dt_float = float(cfg.dt.item())
Nt_run = min(50, int(cfg.Nt))  # run 50 steps for a quick test

# Optional: enable KO if you defined ko in Block 3.5
use_KO = ("ko" in globals())
diss_on = False  # set True to inject KO into RHS

state, B_driver = evolve_bssn(
    fd=fd,
    state=state,
    B=B_driver,
    Nt=Nt_run,
    dt=dt_float,
    gauge_params=GaugeParams(eta=2.0, beta_driver_coeff=0.75),
    print_every=5,
    ko=(ko if use_KO else None),
    dissipation_enabled=diss_on,
)
```

Interpretación de resultados esperados (para validar que “no explotó”)

En flat (Minkowski):

deberías ver valores cercanos a 0 para phi y K, y alpha cerca de 1.

det(tilde_gamma) debe quedarse muy cerca de 1 y trA cerca de 0 (tras proyección).

En punctures (Brill–Lindquist):

phi no será pequeño (hay singularidades suavizadas por eps), pero el sistema no debería generar NaNs rápido si dt es conservador.

Si explota: baja CFL, activa KO, o empieza con malla menor.

Alternativas de optimización (ya útiles en Block 5)

torch.no_grad() durante la evolución: ya lo usamos en el update RK para ahorrar memoria (no autograd).

KO dissipation: si aparece “noise” de alta frecuencia, activar dissipation_enabled=True.

torch.compile (PyTorch 2): cuando esté estable, compilar bssn_rhs_vacuum + rk4_step_bssn puede acelerar bastante.

Checkpointing/outputs: guarda snapshots cada N pasos para análisis posterior (lo añadimos en el siguiente bloque si quieres).

## Block 6 — Constraint Monitoring (Hamiltonian + Momentum) in Torch
---
=================================================================

This block computes the ADM constraints in vacuum for a BSSN state:

Hamiltonian constraint:
    H = R + K^2 - K_ij K^{ij} = 0

Momentum constraint:
    M^i = D_j (K^{ij} - γ^{ij} K) = 0

where:
- γ_ij is the physical 3-metric
- K_ij is the physical extrinsic curvature
- R is the 3D Ricci scalar of γ_ij
- D is the covariant derivative compatible with γ_ij

We compute R and Γ^i_{jk} using BSSN variables:
    γ_ij = e^{4φ} \tilde{γ}_{ij}
    γ^{ij} = e^{-4φ} \tilde{γ}^{ij}
    A_ij = e^{4φ} \tilde{A}_{ij}
    K_ij = A_ij + (1/3) γ_ij K

We reuse the same "first Block 4" building blocks:
- conformal Christoffels:  \tilde{Γ}^i_{jk}
- physical Christoffels:   Γ^i_{jk} = \tilde{Γ}^i_{jk} + 2(δ^i_j ∂_k φ + δ^i_k ∂_j φ - \tilde{γ}_{jk} \tilde{γ}^{il} ∂_l φ)
- physical Ricci tensor:   R_ij = \tilde{R}_{ij} + R^φ_{ij}
- Ricci scalar:            R = γ^{ij} R_ij

Outputs:
- H: scalar field (Nx,Ny,Nz)
- M: vector field (3,Nx,Ny,Nz) representing M^i
- summary norms and optionally L2-like estimates.

Prerequisites:
- cfg (Block 1)
- state (Block 2) with fields: phi, tilde_gamma, K, tilde_A, tilde_Gamma, alpha, beta
- fd (Block 3) with d1 / grad_scalar / grad_vector / grad_tensor2 etc.
- (Optional) If your Block 4 defined ricci_BSSN/christoffel_conformal/christoffel_physical, we reuse them.
  If not, we re-define the minimal parts here.
"""

```python

# ============================================================
# 0) Basic 3x3 field ops (Torch-only)
# ============================================================

def det_3x3_field(g: torch.Tensor) -> torch.Tensor:
    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]
    return a*(e*i - f*h) - b*(d*i - f*g2) + c*(d*h - e*g2)

def inv_3x3_field(g: torch.Tensor, eps: float = 1e-30) -> torch.Tensor:
    detg = det_3x3_field(g)
    detg_safe = detg + eps

    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]

    C00 = (e*i - f*h)
    C01 = -(d*i - f*g2)
    C02 = (d*h - e*g2)

    C10 = -(b*i - c*h)
    C11 = (a*i - c*g2)
    C12 = -(a*h - b*g2)

    C20 = (b*f - c*e)
    C21 = -(a*f - c*d)
    C22 = (a*e - b*d)

    adj = torch.stack([
        torch.stack([C00, C10, C20], dim=0),
        torch.stack([C01, C11, C21], dim=0),
        torch.stack([C02, C12, C22], dim=0),
    ], dim=0)
    return adj / detg_safe.unsqueeze(0).unsqueeze(0)

def trace_wrt_metric(T: torch.Tensor, g_inv: torch.Tensor) -> torch.Tensor:
    return torch.einsum("ij...,ij...->...", g_inv, T)


# ============================================================
# 1) Christoffels (conformal + physical)
# ============================================================

def christoffel_conformal(fd, tilde_gamma: torch.Tensor, tilde_gamma_inv: torch.Tensor) -> torch.Tensor:
    """
    \tilde{Γ}^i_{jk} = 1/2 \tilde{γ}^{il} (∂_j \tilde{γ}_{lk} + ∂_k \tilde{γ}_{lj} - ∂_l \tilde{γ}_{jk})
    Returns: (i,j,k,...)
    """
    d_tg = fd.grad_tensor2(tilde_gamma)  # out[l,k,j] = ∂_j tilde_gamma_{lk}
    G = torch.zeros((3,3,3) + tilde_gamma.shape[2:], device=tilde_gamma.device, dtype=tilde_gamma.dtype)

    for i in range(3):
        for j in range(3):
            for k in range(3):
                acc = 0.0
                for l in range(3):
                    acc = acc + tilde_gamma_inv[i, l] * (d_tg[l, k, j] + d_tg[l, j, k] - d_tg[j, k, l])
                G[i, j, k] = 0.5 * acc
    return G

def christoffel_physical_from_conformal(
    tilde_gamma: torch.Tensor,
    tilde_gamma_inv: torch.Tensor,
    dphi: torch.Tensor,
    Gamma_tilde_ijk: torch.Tensor
) -> torch.Tensor:
    """
    Γ^i_{jk} = \tilde{Γ}^i_{jk} + 2(δ^i_j ∂_k φ + δ^i_k ∂_j φ - \tilde{γ}_{jk} \tilde{γ}^{il} ∂_l φ)
    Returns: (i,j,k,...)
    """
    V = torch.einsum("il...,l...->i...", tilde_gamma_inv, dphi)  # V^i = tilde_gamma^{il} dphi_l

    Gamma = Gamma_tilde_ijk.clone()
    for i in range(3):
        for j in range(3):
            for k in range(3):
                delta_ij = 1.0 if i == j else 0.0
                delta_ik = 1.0 if i == k else 0.0
                Gamma[i, j, k] = Gamma[i, j, k] + 2.0 * (delta_ij * dphi[k] + delta_ik * dphi[j] - tilde_gamma[j, k] * V[i])
    return Gamma


# ============================================================
# 2) Ricci tensor + Ricci scalar via BSSN split (same as Block 4)
# ============================================================

def ricci_from_christoffel(fd, Gamma_kij: torch.Tensor) -> torch.Tensor:
    """
    R_ij = ∂_k Γ^k_{ij} - ∂_j Γ^k_{ik} + Γ^k_{ij} Γ^l_{kl} - Γ^k_{il} Γ^l_{kj}

    Input Gamma_kij is stored as (k,i,j,...).
    Returns R_ij as (i,j,...).
    """
    shape = Gamma_kij.shape[3:]
    R = torch.zeros((3,3) + shape, device=Gamma_kij.device, dtype=Gamma_kij.dtype)

    # C_k = Γ^l_{kl}
    C = torch.zeros((3,) + shape, device=Gamma_kij.device, dtype=Gamma_kij.dtype)
    for k in range(3):
        tmp = 0.0
        for l in range(3):
            tmp = tmp + Gamma_kij[l, k, l]
        C[k] = tmp

    def d_axis(f, ax):  # ax=0,1,2
        return fd.d1(f, axis=ax)

    for i in range(3):
        for j in range(3):
            term1 = 0.0
            for k in range(3):
                term1 = term1 + d_axis(Gamma_kij[k, i, j], k)

            term2_sum = 0.0
            for k in range(3):
                term2_sum = term2_sum + Gamma_kij[k, i, k]
            term2 = d_axis(term2_sum, j)

            term3 = 0.0
            for k in range(3):
                term3 = term3 + Gamma_kij[k, i, j] * C[k]

            term4 = 0.0
            for k in range(3):
                for l in range(3):
                    term4 = term4 + Gamma_kij[k, i, l] * Gamma_kij[l, k, j]

            R[i, j] = term1 - term2 + term3 - term4

    return R

def ricci_BSSN(fd, phi: torch.Tensor, tilde_gamma: torch.Tensor, tilde_gamma_inv: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
    """
    R_ij = \tilde{R}_{ij} + R^φ_{ij}
    Returns (R_ij, Rtilde_ij).
    """
    Gamma_tilde = christoffel_conformal(fd, tilde_gamma, tilde_gamma_inv)  # (i,j,k,...)
    # Interpret Gamma_tilde as (k,i,j,...) for ricci_from_christoffel:
    # our storage is (i,j,k)=Γ^i_{jk}, i is upper index already -> treat as k=i directly.
    Rtilde = ricci_from_christoffel(fd, Gamma_tilde)

    dphi = fd.grad_scalar(phi)  # (3,...)

    # second partials of phi
    d2phi = torch.zeros((3,3) + phi.shape, device=phi.device, dtype=phi.dtype)
    for i in range(3):
        for j in range(3):
            d2phi[i, j] = fd.d1(fd.d1(phi, axis=j), axis=i)

    # \tilde{D}_i \tilde{D}_j phi = ∂_i∂_j phi - \tilde{Γ}^k_{ij} ∂_k phi
    Dtilde2_phi = torch.zeros_like(d2phi)
    for i in range(3):
        for j in range(3):
            corr = 0.0
            for k in range(3):
                corr = corr + Gamma_tilde[k, i, j] * dphi[k]
            Dtilde2_phi[i, j] = d2phi[i, j] - corr

    Dtilde_lap_phi = torch.einsum("ij...,ij...->...", tilde_gamma_inv, Dtilde2_phi)
    dphi_sq = torch.einsum("ij...,i...,j...->...", tilde_gamma_inv, dphi, dphi)

    Rphi = torch.zeros((3,3) + phi.shape, device=phi.device, dtype=phi.dtype)
    for i in range(3):
        for j in range(3):
            Rphi[i, j] = (
                -2.0 * Dtilde2_phi[i, j]
                -2.0 * tilde_gamma[i, j] * Dtilde_lap_phi
                +4.0 * dphi[i] * dphi[j]
                -4.0 * tilde_gamma[i, j] * dphi_sq
            )

    R = Rtilde + Rphi
    return R, Rtilde


# ============================================================
# 3) Momentum constraint: M^i = D_j (K^{ij} - γ^{ij} K)
# ============================================================

def cov_div_contrav_rank2(fd, Gamma_phys: torch.Tensor, T_upup: torch.Tensor) -> torch.Tensor:
    """
    Compute divergence of a contravariant rank-2 tensor:
        (D_j T^{ij}) = ∂_j T^{ij} + Γ^i_{jk} T^{kj} + Γ^j_{jk} T^{ik}

    Inputs:
        Gamma_phys: (i,j,k,...)  = Γ^i_{jk}
        T_upup:     (i,j,...)    = T^{ij}

    Output:
        divT: (i,...) = D_j T^{ij}
    """
    shape = T_upup.shape[2:]
    divT = torch.zeros((3,) + shape, device=T_upup.device, dtype=T_upup.dtype)

    # Precompute C_k = Γ^j_{jk} (contraction on upper index j)
    # Here: C_k = sum_j Γ^j_{jk}
    C = torch.zeros((3,) + shape, device=T_upup.device, dtype=T_upup.dtype)
    for k in range(3):
        tmp = 0.0
        for j in range(3):
            tmp = tmp + Gamma_phys[j, j, k]
        C[k] = tmp

    for i in range(3):
        # ∂_j T^{ij}
        term_partial = 0.0
        for j in range(3):
            term_partial = term_partial + fd.d1(T_upup[i, j], axis=j)

        # Γ^i_{jk} T^{kj}
        term1 = 0.0
        for j in range(3):
            for k in range(3):
                term1 = term1 + Gamma_phys[i, j, k] * T_upup[k, j]

        # Γ^j_{jk} T^{ik} = C_k T^{ik}
        term2 = 0.0
        for k in range(3):
            term2 = term2 + C[k] * T_upup[i, k]

        divT[i] = term_partial + term1 + term2

    return divT


# ============================================================
# 4) Main constraint computation
# ============================================================

@torch.no_grad()
def compute_constraints(fd, state) -> Dict[str, torch.Tensor]:
    """
    Compute Hamiltonian H and Momentum M^i (vacuum).

    Returns dict with:
        - H : (Nx,Ny,Nz)
        - M : (3,Nx,Ny,Nz)
        - R : Ricci scalar field
        - det_gamma : det(physical metric)
        - det_tilde_gamma : det(conformal metric)
    """
    phi = state.phi
    tg = state.tilde_gamma
    K = state.K
    tA = state.tilde_A

    # Build inverses
    tg_inv = inv_3x3_field(tg)

    # Physical metric and inverse
    e4phi = torch.exp(4.0 * phi)
    em4phi = torch.exp(-4.0 * phi)

    gamma = tg * e4phi.unsqueeze(0).unsqueeze(0)         # γ_ij
    gamma_inv = tg_inv * em4phi.unsqueeze(0).unsqueeze(0) # γ^{ij}

    # Physical extrinsic curvature:
    # A_ij = e^{4φ} \tilde{A}_{ij}
    A = tA * e4phi.unsqueeze(0).unsqueeze(0)
    Kij = A + (1.0/3.0) * gamma * K.unsqueeze(0).unsqueeze(0)

    # Raise indices: K^{ij} = γ^{ik} γ^{jl} K_{kl}
    Kupup = torch.einsum("ik...,jl...,kl...->ij...", gamma_inv, gamma_inv, Kij)

    # K_ij K^{ij}
    KijKij = torch.einsum("ij...,ij...->...", Kij, Kupup)

    # Ricci scalar: R = γ^{ij} R_ij
    R_ij, _ = ricci_BSSN(fd, phi, tg, tg_inv)
    R = torch.einsum("ij...,ij...->...", gamma_inv, R_ij)

    # Hamiltonian constraint
    H = R + K*K - KijKij

    # Momentum constraint:
    # M^i = D_j (K^{ij} - γ^{ij} K) = D_j (A^{ij} - (2/3) γ^{ij} K)
    # First compute A^{ij} = γ^{ik} γ^{jl} A_{kl}
    Aupup = torch.einsum("ik...,jl...,kl...->ij...", gamma_inv, gamma_inv, A)

    Tupup = Aupup - (2.0/3.0) * gamma_inv * K.unsqueeze(0).unsqueeze(0)

    # Physical Christoffels
    dphi = fd.grad_scalar(phi)  # (3,...)
    Gamma_tilde = christoffel_conformal(fd, tg, tg_inv)  # (i,j,k,...)
    Gamma_phys = christoffel_physical_from_conformal(tg, tg_inv, dphi, Gamma_tilde)  # (i,j,k,...)

    M = cov_div_contrav_rank2(fd, Gamma_phys, Tupup)  # (3,...)

    # Extra diagnostics
    det_tg = det_3x3_field(tg)
    det_g = det_3x3_field(gamma)

    return {
        "H": H,
        "M": M,
        "R": R,
        "det_gamma": det_g,
        "det_tilde_gamma": det_tg,
    }


@torch.no_grad()
def constraint_norms(constraints: Dict[str, torch.Tensor]) -> Dict[str, float]:
    """
    Compute simple norms for monitoring:
    - mean abs and max abs for H
    - mean abs and max abs for |M| (Euclidean norm)
    """
    H = constraints["H"]
    M = constraints["M"]

    M_norm = torch.sqrt(M[0]**2 + M[1]**2 + M[2]**2)

    return {
        "H_mean_abs": float(H.abs().mean().item()),
        "H_max_abs": float(H.abs().max().item()),
        "M_mean_abs": float(M_norm.abs().mean().item()),
        "M_max_abs": float(M_norm.abs().max().item()),
        "R_mean": float(constraints["R"].mean().item()),
        "det_tg_mean_dev": float((constraints["det_tilde_gamma"] - 1.0).abs().mean().item()),
        "det_tg_max_dev": float((constraints["det_tilde_gamma"] - 1.0).abs().max().item()),
    }


# ============================================================
# 5) Example: compute constraints after some evolution steps
# ============================================================

# Choose which state is in memory
state = state_bl if "state_bl" in globals() else state_flat

cons = compute_constraints(fd, state)
cn = constraint_norms(cons)

print("=== Constraint norms (vacuum) ===")
for k, v in cn.items():
    print(f"{k:>16s}: {v:.6e}")
```

Block 7 — Diagnostics Logging, Checkpointing, and Basic 1D Probes (Torch-first)
==============================================================================

Purpose
-------
Now that we can evolve (Block 5) and monitor constraints (Block 6), we need
infrastructure to:

1) Log time series of:
   - basic field diagnostics (Block 5 diagnostics_basic)
   - constraint norms (Block 6 constraint_norms)
2) Save/restore checkpoints:
   - full BSSN state tensors
   - B^i driver field
   - current step/time
   - configuration metadata (dtype, grid, dt, etc.)
3) Extract simple probes (cheap but very useful):
   - along coordinate axes (x-axis line through center)
   - at a specific point (center or an extraction radius index)
   - optionally compute a simple "wave proxy" from φ or K at radii

Notes
-----
- This is NOT full gravitational wave extraction (Ψ4 etc.). That typically needs:
  - tetrads, Weyl tensor, and/or Cauchy-characteristic extraction
  - a dedicated module and careful outer boundary conditions
We'll do that later once the evolution is stable.

Prerequisites
-------------
- Block 1: cfg
- Block 2: BSSNState object `state`
- Block 5: `evolve_bssn` or `rk4_step_bssn`
- Block 6: `compute_constraints` and `constraint_norms`

```python

# ------------------------------------------------------------
# 1) Checkpoint I/O
# ------------------------------------------------------------

def pack_state(state) -> Dict[str, torch.Tensor]:
    """
    Convert a BSSNState-like object into a pure tensor dict for saving.
    """
    return {
        "phi": state.phi,
        "tilde_gamma": state.tilde_gamma,
        "K": state.K,
        "tilde_A": state.tilde_A,
        "tilde_Gamma": state.tilde_Gamma,
        "alpha": state.alpha,
        "beta": state.beta,
    }

def unpack_state(state, data: Dict[str, torch.Tensor]):
    """
    Load tensors into an existing state object (in-place).
    """
    state.phi = data["phi"]
    state.tilde_gamma = data["tilde_gamma"]
    state.K = data["K"]
    state.tilde_A = data["tilde_A"]
    state.tilde_Gamma = data["tilde_Gamma"]
    state.alpha = data["alpha"]
    state.beta = data["beta"]
    return state

def save_checkpoint(
    path: str,
    state,
    B: torch.Tensor,
    step: int,
    t: float,
    cfg: Any,
    extra: Optional[Dict[str, Any]] = None,
):
    """
    Save a checkpoint to disk with torch.save (portable, simple).

    Parameters
    ----------
    path : str
        File path, e.g. "checkpoints/ckpt_000100.pt"
    state : BSSNState-like
    B : torch.Tensor
        Shift driver auxiliary field B^i, shape (3,...)
    step : int
    t : float
    cfg : SimulationConfigTorch
    extra : dict
        Optional metadata (e.g. gauge params, KO params, git hash, notes)
    """
    os.makedirs(os.path.dirname(path), exist_ok=True)

    payload = {
        "step": int(step),
        "t": float(t),
        "state": pack_state(state),
        "B": B,
        "cfg_summary": cfg.summary() if hasattr(cfg, "summary") else None,
        "extra": extra if extra is not None else {},
    }
    torch.save(payload, path)

def load_checkpoint(path: str, state_template=None, map_location: Optional[str] = None):
    """
    Load a checkpoint. If you pass `state_template`, tensors are loaded into it.
    Otherwise returns the raw payload.

    Returns
    -------
    (payload, state, B) if state_template provided
    else payload
    """
    payload = torch.load(path, map_location=map_location)
    if state_template is None:
        return payload

    state = unpack_state(state_template, payload["state"])
    B = payload["B"]
    return payload, state, B


# ------------------------------------------------------------
# 2) Probes (1D lines, points)
# ------------------------------------------------------------

def center_indices(cfg) -> Tuple[int, int, int]:
    """
    Return integer indices (ix,iy,iz) for grid center.
    """
    g = cfg.grid
    ix = g.Nx // 2
    iy = g.Ny // 2
    iz = g.Nz // 2
    return ix, iy, iz

@torch.no_grad()
def extract_line_x_center(field: torch.Tensor, cfg) -> torch.Tensor:
    """
    Extract a 1D line along x through the domain center:
      field[ix, iy_center, iz_center] for all ix.

    field must be scalar (Nx,Ny,Nz).
    """
    ix0, iy0, iz0 = center_indices(cfg)
    return field[:, iy0, iz0].clone()

@torch.no_grad()
def extract_point(field: torch.Tensor, idx: Tuple[int, int, int]) -> torch.Tensor:
    """
    Extract field value at a single grid point.
    For scalar: returns 0-dim tensor
    """
    ix, iy, iz = idx
    return field[ix, iy, iz].clone()

@torch.no_grad()
def extract_radius_proxy(field: torch.Tensor, cfg, r_index: int) -> torch.Tensor:
    """
    Very simple 'radius proxy' for quick waveform-like monitoring:
    - Assume the grid is centered and symmetric.
    - Take the value on the +x axis at (ix_center + r_index, iy_center, iz_center).

    This is NOT an invariant wave extraction, but it is useful early on.
    """
    ix0, iy0, iz0 = center_indices(cfg)
    ix = min(ix0 + r_index, cfg.grid.Nx - 1)
    return field[ix, iy0, iz0].clone()


# ------------------------------------------------------------
# 3) Logger (store time series on CPU to save GPU memory)
# ------------------------------------------------------------

@dataclass
class RunLogger:
    """
    Stores time series diagnostics as Python floats on CPU.

    It is common to:
    - compute diagnostics on GPU tensors,
    - immediately convert to floats,
    - store only small scalars in RAM.

    This avoids building big history tensors.
    """
    times: List[float]
    steps: List[int]
    diag: Dict[str, List[float]]
    cons: Dict[str, List[float]]

    # Optional probe time series
    probes: Dict[str, List[float]]

    def __init__(self):
        self.times = []
        self.steps = []
        self.diag = {}
        self.cons = {}
        self.probes = {}

    def _append_dict(self, storage: Dict[str, List[float]], values: Dict[str, float]):
        for k, v in values.items():
            if k not in storage:
                storage[k] = []
            storage[k].append(float(v))

    def log_step(
        self,
        step: int,
        t: float,
        diag_vals: Dict[str, float],
        cons_vals: Dict[str, float],
        probe_vals: Optional[Dict[str, float]] = None,
    ):
        self.steps.append(int(step))
        self.times.append(float(t))
        self._append_dict(self.diag, diag_vals)
        self._append_dict(self.cons, cons_vals)
        if probe_vals is not None:
            self._append_dict(self.probes, probe_vals)

    def to_torch(self, device="cpu") -> Dict[str, Any]:
        """
        Convert stored histories to torch tensors for plotting/postprocessing.
        """
        out = {
            "steps": torch.tensor(self.steps, device=device),
            "times": torch.tensor(self.times, device=device),
            "diag": {k: torch.tensor(v, device=device) for k, v in self.diag.items()},
            "cons": {k: torch.tensor(v, device=device) for k, v in self.cons.items()},
            "probes": {k: torch.tensor(v, device=device) for k, v in self.probes.items()},
        }
        return out


# ------------------------------------------------------------
# 4) Evolution driver with logging + periodic checkpoints
# ------------------------------------------------------------

def evolve_with_logging(
    fd,
    state,
    B: torch.Tensor,
    cfg,
    gauge_params,
    Nt: int,
    dt: float,
    log_every: int = 10,
    ckpt_every: int = 100,
    ckpt_dir: str = "checkpoints",
    ko: Optional[object] = None,
    dissipation_enabled: bool = False,
    probe_r_index: int = 10,
) -> Tuple[Any, torch.Tensor, RunLogger]:
    """
    Run evolution and:
    - log diagnostics + constraints every `log_every` steps
    - save checkpoint every `ckpt_every` steps

    Returns updated (state, B, logger).
    """
    logger = RunLogger()

    t = 0.0
    for n in range(Nt):
        # One RK4 step (Block 5)
        state, B = rk4_step_bssn(
            fd=fd,
            state=state,
            B=B,
            cfg_int=IntegratorConfig(
                dt=dt,
                project_each_step=True,
                nan_check=True,
                dissipation=DissipationConfig(enabled=dissipation_enabled),
            ),
            gauge_params=gauge_params,
            ko=ko,
        )
        t += dt

        # Logging
        if (n % log_every) == 0 or (n == Nt - 1):
            d_basic = diagnostics_basic(state)  # Block 5
            cons = compute_constraints(fd, state)  # Block 6
            d_cons = constraint_norms(cons)

            # Probe: field values at a pseudo-radius along +x axis
            # Using phi and K as simple "wave proxies" (not physical extraction yet)
            p_phi = extract_radius_proxy(state.phi, cfg, r_index=probe_r_index)
            p_K = extract_radius_proxy(state.K, cfg, r_index=probe_r_index)

            logger.log_step(
                step=n,
                t=t,
                diag_vals=d_basic,
                cons_vals=d_cons,
                probe_vals={
                    f"phi_r{probe_r_index}": float(p_phi.item()),
                    f"K_r{probe_r_index}": float(p_K.item()),
                }
            )

            print(f"[log step {n:05d}] t={t:.6e} | "
                  f"Hmean={d_cons['H_mean_abs']:.3e} Hmax={d_cons['H_max_abs']:.3e} | "
                  f"Mmean={d_cons['M_mean_abs']:.3e} Mmax={d_cons['M_max_abs']:.3e} | "
                  f"phi(r)={float(p_phi.item()):+.3e} K(r)={float(p_K.item()):+.3e}")

        # Checkpoints
        if (ckpt_every > 0) and ((n % ckpt_every) == 0) and (n != 0):
            ckpt_path = os.path.join(ckpt_dir, f"ckpt_{n:06d}.pt")
            save_checkpoint(
                path=ckpt_path,
                state=state,
                B=B,
                step=n,
                t=t,
                cfg=cfg,
                extra={
                    "dt": dt,
                    "dissipation_enabled": dissipation_enabled,
                    "probe_r_index": probe_r_index,
                }
            )
            print(f"  saved checkpoint: {ckpt_path}")

    return state, B, logger


# ------------------------------------------------------------
# 5) Example: run a short logged evolution
# ------------------------------------------------------------

# Choose a state (puncture or flat)
state = state_bl if "state_bl" in globals() else state_flat

# Ensure B exists
if "B_driver" not in globals():
    B_driver = torch.zeros_like(state.beta)

dt_float = float(cfg.dt.item())
Nt_run = min(200, int(cfg.Nt))  # keep small at first

use_KO = ("ko" in globals())
state, B_driver, logger = evolve_with_logging(
    fd=fd,
    state=state,
    B=B_driver,
    cfg=cfg,
    gauge_params=GaugeParams(eta=2.0, beta_driver_coeff=0.75),
    Nt=Nt_run,
    dt=dt_float,
    log_every=10,
    ckpt_every=100,
    ckpt_dir="checkpoints",
    ko=(ko if use_KO else None),
    dissipation_enabled=False,   # set True if you see high-freq noise
    probe_r_index=10,            # adjust based on resolution
)

hist = logger.to_torch(device="cpu")
print("Logged keys (diag):", list(hist["diag"].keys()))
print("Logged keys (cons):", list(hist["cons"].keys()))
print("Logged keys (probes):", list(hist["probes"].keys()))
```

Qué desbloquea Block 7 (y por qué es importante)

Te permite ver si la simulación “es física” (constraints) y no solo si “no crashea”.

Con checkpoints puedes:

reanudar si cambias parámetros,

hacer barridos de CFL/KO,

depurar sin recomputar desde cero.

Los probes te dan una primera señal de “radiación” o dinámica, aunque aún no sea Ψ4.

Block 8 — Minimal Ψ4 Wave Extraction on Coordinate Spheres (Torch-first)
========================================================================

What we implement (incremental, practical)
------------------------------------------
We implement a *baseline* gravitational-wave extraction module that computes an
approximate Newman–Penrose Weyl scalar Ψ4 on coordinate spheres r = const.

We use the 3+1 split (vacuum) and compute the electric/magnetic parts of the Weyl tensor:
    E_ij = R_ij + K K_ij - K_i^k K_kj
    B_ij = ε_i^{kl} D_k K_lj

Then on a sphere we build an orthonormal triad (e_r, e_theta, e_phi) and define:
    m^i = (e_theta^i + i e_phi^i)/sqrt(2)

and compute the outgoing-radiation proxy:
    Ψ4 ≈ (E_ij - i B_ij) * \bar{m}^i \bar{m}^j

This is a common “first working” extraction in NR prototypes, but:
- it is sensitive to gauge and outer boundary conditions,
- it is NOT as clean as full tetrad/Weyl-from-4D metric extraction,
- accuracy improves with larger extraction radii and better BCs.

Still, it gives you a waveform-like complex signal vs time.

Dependencies (must already exist)
---------------------------------
- cfg          (Block 1)
- state        (BSSNState, Block 2)
- fd           (FDOperators, Block 3)
- ricci_BSSN   (from Block 4 first version)  OR we re-define a minimal call here
- inv_3x3_field, det_3x3_field, trace_wrt_metric helpers (Block 4/5 style)

We will:
1) Compute physical metric γ_ij and extrinsic curvature K_ij from BSSN variables.
2) Compute Ricci tensor R_ij from BSSN split.
3) Build E_ij and B_ij on the full grid (heavy but straightforward).
4) Sample E_ij, B_ij onto a coordinate sphere using torch.grid_sample (trilinear interpolation).
5) Construct tetrad directions (e_theta, e_phi) for each sphere point.
6) Evaluate Ψ4(θ,φ) as a complex tensor (Nθ,Nφ).

You can call this inside the evolution loop every N steps (e.g. at log steps in Block 7).

```python

# ============================================================
# 0) Basic helpers (3D Levi-Civita, inverse, etc.)
# ============================================================

def det_3x3_field(g: torch.Tensor) -> torch.Tensor:
    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]
    return a*(e*i - f*h) - b*(d*i - f*g2) + c*(d*h - e*g2)

def inv_3x3_field(g: torch.Tensor, eps: float = 1e-30) -> torch.Tensor:
    detg = det_3x3_field(g)
    detg_safe = detg + eps

    a = g[0,0]; b = g[0,1]; c = g[0,2]
    d = g[1,0]; e = g[1,1]; f = g[1,2]
    g2 = g[2,0]; h = g[2,1]; i = g[2,2]

    C00 = (e*i - f*h)
    C01 = -(d*i - f*g2)
    C02 = (d*h - e*g2)

    C10 = -(b*i - c*h)
    C11 = (a*i - c*g2)
    C12 = -(a*h - b*g2)

    C20 = (b*f - c*e)
    C21 = -(a*f - c*d)
    C22 = (a*e - b*d)

    adj = torch.stack([
        torch.stack([C00, C10, C20], dim=0),
        torch.stack([C01, C11, C21], dim=0),
        torch.stack([C02, C12, C22], dim=0),
    ], dim=0)
    return adj / detg_safe.unsqueeze(0).unsqueeze(0)

def trace_wrt_metric(T: torch.Tensor, g_inv: torch.Tensor) -> torch.Tensor:
    return torch.einsum("ij...,ij...->...", g_inv, T)

def levi_civita_symbol(device, dtype) -> torch.Tensor:
    """
    ε_{ijk} with ε_{012}=+1 in (x,y,z) ~ (0,1,2) indexing.
    """
    eps = torch.zeros((3,3,3), device=device, dtype=dtype)
    eps[0,1,2] = 1.0
    eps[1,2,0] = 1.0
    eps[2,0,1] = 1.0
    eps[0,2,1] = -1.0
    eps[2,1,0] = -1.0
    eps[1,0,2] = -1.0
    return eps


# ============================================================
# 1) Build physical ADM objects (γ_ij, K_ij) from BSSN variables
# ============================================================

@torch.no_grad()
def adm_from_bssn(state) -> Tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
    """
    Returns:
        gamma_ij  : (3,3,Nx,Ny,Nz)
        gamma_inv : (3,3,Nx,Ny,Nz)
        K_ij      : (3,3,Nx,Ny,Nz)
    """
    phi = state.phi
    tg = state.tilde_gamma
    tA = state.tilde_A
    K = state.K

    tg_inv = inv_3x3_field(tg)

    e4phi = torch.exp(4.0 * phi)
    em4phi = torch.exp(-4.0 * phi)

    gamma = tg * e4phi.unsqueeze(0).unsqueeze(0)
    gamma_inv = tg_inv * em4phi.unsqueeze(0).unsqueeze(0)

    # A_ij = e^{4φ} \tilde{A}_{ij}
    A = tA * e4phi.unsqueeze(0).unsqueeze(0)

    # K_ij = A_ij + (1/3) γ_ij K
    Kij = A + (1.0/3.0) * gamma * K.unsqueeze(0).unsqueeze(0)

    return gamma, gamma_inv, Kij


# ============================================================
# 2) Spatial covariant derivative of K_ij
#    D_k K_{ij} = ∂_k K_{ij} - Γ^m_{k i} K_{m j} - Γ^m_{k j} K_{i m}
# ============================================================

def christoffel_conformal(fd, tilde_gamma: torch.Tensor, tilde_gamma_inv: torch.Tensor) -> torch.Tensor:
    """
    \tilde{Γ}^i_{jk} = 1/2 \tilde{γ}^{il} (∂_j \tilde{γ}_{lk} + ∂_k \tilde{γ}_{lj} - ∂_l \tilde{γ}_{jk})
    Returns: (i,j,k,...)
    """
    d_tg = fd.grad_tensor2(tilde_gamma)  # (l,k,j,...)=∂_j tilde_gamma_{lk}
    G = torch.zeros((3,3,3) + tilde_gamma.shape[2:], device=tilde_gamma.device, dtype=tilde_gamma.dtype)
    for i in range(3):
        for j in range(3):
            for k in range(3):
                acc = 0.0
                for l in range(3):
                    acc = acc + tilde_gamma_inv[i, l] * (d_tg[l, k, j] + d_tg[l, j, k] - d_tg[j, k, l])
                G[i, j, k] = 0.5 * acc
    return G

def christoffel_physical_from_conformal(
    tilde_gamma: torch.Tensor,
    tilde_gamma_inv: torch.Tensor,
    dphi: torch.Tensor,
    Gamma_tilde: torch.Tensor
) -> torch.Tensor:
    """
    Γ^i_{jk} = \tilde{Γ}^i_{jk} + 2(δ^i_j ∂_k φ + δ^i_k ∂_j φ - \tilde{γ}_{jk} \tilde{γ}^{il} ∂_l φ)
    Returns: (i,j,k,...)
    """
    V = torch.einsum("il...,l...->i...", tilde_gamma_inv, dphi)
    Gamma = Gamma_tilde.clone()
    for i in range(3):
        for j in range(3):
            for k in range(3):
                delta_ij = 1.0 if i == j else 0.0
                delta_ik = 1.0 if i == k else 0.0
                Gamma[i, j, k] = Gamma[i, j, k] + 2.0 * (delta_ij * dphi[k] + delta_ik * dphi[j] - tilde_gamma[j, k] * V[i])
    return Gamma

@torch.no_grad()
def covariant_derivative_Kij(fd, state, Kij: torch.Tensor) -> torch.Tensor:
    """
    Compute D_k K_{ij} for all i,j,k.

    Output:
        DK : (3,3,3,Nx,Ny,Nz) where DK[k,i,j] = D_k K_{ij}
    """
    phi = state.phi
    tg = state.tilde_gamma
    tg_inv = inv_3x3_field(tg)

    dphi = fd.grad_scalar(phi)
    Gamma_tilde = christoffel_conformal(fd, tg, tg_inv)
    Gamma = christoffel_physical_from_conformal(tg, tg_inv, dphi, Gamma_tilde)  # (m,k,i,...)?? careful: (i,j,k)=Γ^i_{jk}

    DK = torch.zeros((3,3,3) + Kij.shape[2:], device=Kij.device, dtype=Kij.dtype)

    # DK[k,i,j] = ∂_k K_{ij} - Γ^m_{k i} K_{m j} - Γ^m_{k j} K_{i m}
    for k in range(3):
        for i in range(3):
            for j in range(3):
                partial = fd.d1(Kij[i, j], axis=k)
                corr1 = 0.0
                corr2 = 0.0
                for m in range(3):
                    corr1 = corr1 + Gamma[m, k, i] * Kij[m, j]  # Γ^m_{k i} K_{m j}
                    corr2 = corr2 + Gamma[m, k, j] * Kij[i, m]  # Γ^m_{k j} K_{i m}
                DK[k, i, j] = partial - corr1 - corr2

    return DK


# ============================================================
# 3) Weyl electric/magnetic parts and Ψ4 on grid
# ============================================================

@torch.no_grad()
def compute_E_B(fd, state) -> Tuple[torch.Tensor, torch.Tensor]:
    """
    Compute E_ij and B_ij on the full grid.

    E_ij = R_ij + K K_ij - K_i^k K_kj
    B_ij = ε_i^{kl} D_k K_{l j}

    Returns:
        E : (3,3,Nx,Ny,Nz)
        B : (3,3,Nx,Ny,Nz)
    """
    device, dtype = state.phi.device, state.phi.dtype
    eps_sym = levi_civita_symbol(device, dtype)

    gamma, gamma_inv, Kij = adm_from_bssn(state)
    K = state.K

    # Ricci tensor from BSSN split (reuse Block 4 function if already defined)
    tg_inv = inv_3x3_field(state.tilde_gamma)
    R_ij, _ = ricci_BSSN(fd, state.phi, state.tilde_gamma, tg_inv)

    # K_i^k = γ^{k m} K_{i m}
    Ki_up = torch.einsum("km...,im...->ik...", gamma_inv, Kij)  # (i,k,...)
    KikKkj = torch.einsum("ik...,kj...->ij...", Ki_up, Kij)     # K_i^k K_kj

    E = R_ij + K.unsqueeze(0).unsqueeze(0) * Kij - KikKkj

    # Magnetic part:
    # B_ij = ε_i^{kl} D_k K_{l j}
    DK = covariant_derivative_Kij(fd, state, Kij)  # DK[k,l,j,...]
    B = torch.zeros_like(E)
    for i in range(3):
        acc_ij = 0.0
        for k in range(3):
            for l in range(3):
                # ε_i^{kl} DK[k,l,j]
                # We treat ε_i^{kl} as the flat symbol; for a more exact version you'd use the
                # spatial Levi-Civita tensor with metric factors. This is a standard "minimal" proxy.
                acc_ij = acc_ij + eps_sym[i, k, l] * DK[k, l]  # DK[k,l] is (3,...) over j
        B[i] = acc_ij  # (j,...)
    # Here B[i] has shape (3,...) which matches B_ij

    return E, B


# ============================================================
# 4) Sphere sampling using torch.grid_sample (trilinear)
# ============================================================

@dataclass
class SphereGrid:
    """
    Coordinate sphere sampling grid:
        points in physical coordinates (x,y,z) and grid_sample coords in [-1,1].

    grid_sample expects input as (N,C,D,H,W) and grid as (N, D_out, H_out, W_out, 3)
    We treat:
        D_out = Ntheta
        H_out = Nphi
        W_out = 1
    """
    r: float
    theta: torch.Tensor      # (Ntheta,)
    phi: torch.Tensor        # (Nphi,)
    grid: torch.Tensor       # (1,Ntheta,Nphi,1,3) with order (x,y,z) in normalized coords
    x: torch.Tensor          # (Ntheta,Nphi)
    y: torch.Tensor          # (Ntheta,Nphi)
    z: torch.Tensor          # (Ntheta,Nphi)

def build_sphere_grid(cfg, r: float, Ntheta: int = 48, Nphi: int = 96) -> SphereGrid:
    """
    Build a coordinate sphere r (centered at origin) and the corresponding normalized grid
    for torch.grid_sample.

    Assumes the domain is centered (x_min=-L/2, x_max=+L/2, etc.) as in Block 1 defaults.
    """
    device = cfg.grid.device
    dtype = cfg.grid.dtype

    theta = torch.linspace(1e-6, torch.pi - 1e-6, Ntheta, device=device, dtype=dtype)
    phi = torch.linspace(0.0, 2.0*torch.pi, Nphi, device=device, dtype=dtype)

    TH, PH = torch.meshgrid(theta, phi, indexing="ij")

    x = r * torch.sin(TH) * torch.cos(PH)
    y = r * torch.sin(TH) * torch.sin(PH)
    z = r * torch.cos(TH)

    # Normalize to [-1,1] for grid_sample:
    # x_norm = 2*(x - x_min)/(x_max - x_min) - 1
    g = cfg.grid
    x_norm = 2.0 * (x - g.x_min) / (g.x_max - g.x_min) - 1.0
    y_norm = 2.0 * (y - g.y_min) / (g.y_max - g.y_min) - 1.0
    z_norm = 2.0 * (z - g.z_min) / (g.z_max - g.z_min) - 1.0

    grid = torch.stack([x_norm, y_norm, z_norm], dim=-1)  # (Ntheta,Nphi,3)
    grid = grid.unsqueeze(0).unsqueeze(3)                 # (1,Ntheta,Nphi,1,3)

    return SphereGrid(r=r, theta=theta, phi=phi, grid=grid, x=x, y=y, z=z)


@torch.no_grad()
def sample_tensor2_on_sphere(Tij: torch.Tensor, sph: SphereGrid, mode: str = "bilinear") -> torch.Tensor:
    """
    Sample a 3x3 tensor field Tij (3,3,Nx,Ny,Nz) on a sphere using grid_sample.

    Returns:
        Ts : (3,3,Ntheta,Nphi)  sampled tensor components
    """
    # grid_sample expects input shape (N,C,D,H,W). Our field is (Nx,Ny,Nz).
    # We'll treat D=Nx, H=Ny, W=Nz, but grid_sample uses (D,H,W) ordering.
    # Our tensors are (Nx,Ny,Nz) already. We just need to add batch & channel.
    # We'll sample each component separately (9 calls). This is not fastest but simple.

    device, dtype = Tij.device, Tij.dtype
    Ntheta = sph.grid.shape[1]
    Nphi = sph.grid.shape[2]

    out = torch.zeros((3,3,Ntheta,Nphi), device=device, dtype=dtype)

    for i in range(3):
        for j in range(3):
            vol = Tij[i, j].unsqueeze(0).unsqueeze(0)  # (1,1,Nx,Ny,Nz)
            # align_corners=True makes mapping linear w.r.t normalized coords
            samp = F.grid_sample(vol, sph.grid, mode=mode, padding_mode="zeros", align_corners=True)
            # samp shape (1,1,Ntheta,Nphi,1) -> squeeze
            out[i, j] = samp[0, 0, :, :, 0]
    return out


# ============================================================
# 5) Build tetrad vectors (e_theta, e_phi) and compute Ψ4
# ============================================================

@torch.no_grad()
def tetrad_mbar_on_sphere(sph: SphereGrid) -> Tuple[torch.Tensor, torch.Tensor]:
    """
    Construct mbar^i = (e_theta^i - i e_phi^i)/sqrt(2) on the sphere.

    Returns:
        mbar : complex tensor (3,Ntheta,Nphi)
        (also returns e_r optionally later if needed)
    """
    device = sph.x.device
    dtype = sph.x.dtype

    # Use spherical unit vectors in flat background:
    # e_r     = (sinθ cosφ, sinθ sinφ, cosθ)
    # e_theta = (cosθ cosφ, cosθ sinφ, -sinθ)
    # e_phi   = (-sinφ, cosφ, 0)
    TH, PH = torch.meshgrid(sph.theta, sph.phi, indexing="ij")

    er = torch.stack([
        torch.sin(TH) * torch.cos(PH),
        torch.sin(TH) * torch.sin(PH),
        torch.cos(TH)
    ], dim=0)  # (3,Ntheta,Nphi)

    et = torch.stack([
        torch.cos(TH) * torch.cos(PH),
        torch.cos(TH) * torch.sin(PH),
        -torch.sin(TH)
    ], dim=0)

    ep = torch.stack([
        -torch.sin(PH),
        torch.cos(PH),
        torch.zeros_like(PH)
    ], dim=0)

    # Build mbar = (et - i ep)/sqrt(2)
    # Use complex dtype matching float dtype
    cdtype = torch.complex128 if dtype == torch.float64 else torch.complex64
    et_c = et.to(dtype=cdtype)
    ep_c = ep.to(dtype=cdtype)
    mbar = (et_c - 1j * ep_c) / torch.sqrt(torch.tensor(2.0, device=device, dtype=cdtype))

    return mbar, er


@torch.no_grad()
def psi4_on_sphere(fd, state, sph: SphereGrid) -> torch.Tensor:
    """
    Compute Ψ4(θ,φ) on the extraction sphere.

    Steps:
    1) Compute E_ij, B_ij on grid
    2) Sample E_ij, B_ij on sphere (9+9 grid_sample calls)
    3) Compute Ψ4 = (E_ij - i B_ij) mbar^i mbar^j   (with mbar = conjugate of m)
       Here we directly use mbar, so contraction is:
           Ψ4 = (E_ij - i B_ij) * mbar^i * mbar^j
    """
    E, B = compute_E_B(fd, state)

    Es = sample_tensor2_on_sphere(E, sph, mode="bilinear")  # (3,3,Nθ,Nφ)
    Bs = sample_tensor2_on_sphere(B, sph, mode="bilinear")

    # Complex combination: C_ij = E_ij - i B_ij
    dtype = Es.dtype
    cdtype = torch.complex128 if dtype == torch.float64 else torch.complex64
    C = Es.to(cdtype) - 1j * Bs.to(cdtype)

    mbar, _ = tetrad_mbar_on_sphere(sph)  # (3,Nθ,Nφ), complex

    # Ψ4 = C_ij mbar^i mbar^j
    # Use einsum: (i,j,θ,φ) * (i,θ,φ) * (j,θ,φ) -> (θ,φ)
    psi4 = torch.einsum("ijuv,iuv,juv->uv", C, mbar, mbar)

    return psi4  # complex (Ntheta,Nphi)


# ============================================================
# 6) Example usage (after some evolution steps)
# ============================================================

# Build a sphere inside the domain.
# Choose r so that sphere is well inside boundaries (e.g., 1/4 to 1/3 of half-domain).
# Example: if x spans [-Lx/2, Lx/2], pick r ~ 0.25*Lx.
Lx = float(cfg.grid.x_max - cfg.grid.x_min)
r_ex = 0.25 * Lx

sph = build_sphere_grid(cfg, r=r_ex, Ntheta=32, Nphi=64)

state = state_bl if "state_bl" in globals() else state_flat

psi4 = psi4_on_sphere(fd, state, sph)
print("Psi4 shape:", tuple(psi4.shape), "dtype:", psi4.dtype)
print("Psi4 |mean|:", float(psi4.abs().mean().item()), "Psi4 |max|:", float(psi4.abs().max().item()))

# Simple “time series” proxy at the equator point (theta ~ pi/2, phi ~ 0 index)
it = psi4.shape[0] // 2
ip = 0
print("Psi4(theta=pi/2,phi=0):", complex(psi4[it, ip].item()))
```

Requisitos computacionales (importante)

Este Block 8 es caro porque:

calcula E y B en la malla completa,

luego hace 18 grid_sample (9 para E + 9 para B).

Para no matar rendimiento:

ejecuta Ψ4 cada 10–50 pasos, no cada paso,

usa Ntheta,Nphi modestos (p.ej. 32×64),

y usa un radio de extracción fijo.

```python
"""
Block 9 — Mode Extraction: Project Ψ4(θ,φ) onto spin-weighted harmonics  _{-2}Y_{ℓm}
===================================================================================

Goal
----
Given Ψ4 sampled on a coordinate sphere (Block 8):
    psi4(θ_i, φ_j)  (complex, shape [Nθ, Nφ])

we compute the complex mode coefficients:
    Ψ4^{ℓm}(t) = ∫ Ψ4(θ,φ,t) * conj( _{-2}Y_{ℓm}(θ,φ) ) dΩ

with dΩ = sinθ dθ dφ.

Implementation strategy (Torch-only)
-----------------------------------
We implement spin-weighted spherical harmonics using Wigner small-d matrices:

    _sY_{ℓm}(θ,φ) = (-1)^s * sqrt((2ℓ+1)/(4π)) * d^ℓ_{m,-s}(θ) * exp(i m φ)

For s = -2:
    _{-2}Y_{ℓm}(θ,φ) = (+1) * sqrt((2ℓ+1)/(4π)) * d^ℓ_{m,2}(θ) * exp(i m φ)

We compute d^ℓ_{m,m'}(θ) using the standard finite sum formula (stable enough for small ℓ):
    d^ℓ_{m,m'}(θ) = Σ_k (-1)^{k-m+m'} * sqrt((ℓ+m)!(ℓ-m)!(ℓ+m')!(ℓ-m')!)
                    /[(ℓ+m-k)!(ℓ-m'-k)!k!(k-m+m')!]
                    * (cos(θ/2))^{2ℓ+m-m'-2k} * (sin(θ/2))^{2k-m+m'}

We then:
1) Precompute _{-2}Y_{ℓm}(θ,φ) on your (θ,φ) grid once (cache).
2) Each time you have psi4(θ,φ), project to modes with a weighted sum.

This block provides:
- SpinWeightedYlmCache: precomputes harmonics and quadrature weights
- project_psi4_modes: returns dict {(ℓ,m): complex mode}

Notes
-----
- Use modest lmax first (e.g. 4 or 6). This is meant for BBH wave modes (ℓ=2 dominates).
- This uses coordinate sphere sampling; accuracy improves with larger radius and better BCs.
"""

from dataclasses import dataclass
from typing import Dict, Tuple, Optional
import math
import torch


# ============================================================
# 0) Utilities: factorial and Wigner d (Torch-friendly)
# ============================================================

def _factorial_table(nmax: int, device, dtype) -> torch.Tensor:
    """
    Return factorials as a tensor fact[n] = n! for n=0..nmax (float dtype).
    """
    fact = torch.ones(nmax + 1, device=device, dtype=dtype)
    for n in range(2, nmax + 1):
        fact[n] = fact[n - 1] * n
    return fact

def _wigner_d_l_m_mp(theta: torch.Tensor, l: int, m: int, mp: int, fact: torch.Tensor) -> torch.Tensor:
    """
    Compute d^l_{m,mp}(theta) for theta array (Nθ,) using the finite sum formula.
    Returns real-valued tensor (Nθ,) in float dtype.

    This is implemented for small l (typical l<=8) and is fine for NR mode extraction.
    """
    device = theta.device
    dtype = theta.dtype

    # precompute cos/sin half-angle
    c = torch.cos(theta * 0.5)
    s = torch.sin(theta * 0.5)

    # Prefactor sqrt((l+m)!(l-m)!(l+mp)!(l-mp)!)
    pref = torch.sqrt(fact[l + m] * fact[l - m] * fact[l + mp] * fact[l - mp])

    # k range constraints so factorial arguments non-negative:
    # (l+m-k) >= 0  -> k <= l+m
    # (l-mp-k) >= 0 -> k <= l-mp
    # (k) >= 0
    # (k - m + mp) >= 0 -> k >= m - mp
    kmin = max(0, m - mp)
    kmax = min(l + m, l - mp)

    out = torch.zeros_like(theta)

    for k in range(kmin, kmax + 1):
        denom = fact[l + m - k] * fact[l - mp - k] * fact[k] * fact[k - m + mp]

        sign = -1.0 if ((k - m + mp) % 2 == 1) else 1.0
        coeff = sign * pref / denom

        pow_c = 2 * l + m - mp - 2 * k
        pow_s = 2 * k - m + mp

        out = out + coeff * (c ** pow_c) * (s ** pow_s)

    return out


# ============================================================
# 1) Spin-weighted spherical harmonics for s=-2
# ============================================================

def sYlm_s_minus2(theta: torch.Tensor, phi: torch.Tensor, l: int, m: int,
                  fact: torch.Tensor) -> torch.Tensor:
    """
    Compute _{-2}Y_{l m}(theta, phi) on a tensor-product grid:
        theta: (Nθ,)
        phi:   (Nφ,)
    Returns complex tensor (Nθ, Nφ).
    """
    device = theta.device
    ftype = theta.dtype
    cdtype = torch.complex128 if ftype == torch.float64 else torch.complex64

    # _{-2}Y_{lm} = sqrt((2l+1)/(4π)) d^l_{m,2}(θ) e^{i m φ}
    d = _wigner_d_l_m_mp(theta, l=l, m=m, mp=2, fact=fact)  # (Nθ,) real

    norm = math.sqrt((2 * l + 1) / (4.0 * math.pi))
    phase = torch.exp(1j * torch.tensor(float(m), device=device, dtype=cdtype) * phi.to(cdtype))  # (Nφ,)
    Y = (torch.tensor(norm, device=device, dtype=ftype) * d).to(cdtype).unsqueeze(1) * phase.unsqueeze(0)
    return Y


# ============================================================
# 2) Quadrature weights on (theta,phi) grid
# ============================================================

def sphere_quadrature_weights(theta: torch.Tensor, phi: torch.Tensor) -> torch.Tensor:
    """
    Build weights W[i,j] approximating integral:
        ∫ f(θ,φ) sinθ dθ dφ  ≈  Σ_{i,j} f[i,j] * W[i,j]

    We use simple trapezoidal rule in θ and φ:
        wθ_i = trapezoid spacing
        wφ_j = trapezoid spacing
        W_ij = sinθ_i * wθ_i * wφ_j

    Returns W: (Nθ,Nφ) float tensor.
    """
    device = theta.device
    dtype = theta.dtype
    Nθ = theta.numel()
    Nφ = phi.numel()

    # θ weights
    wθ = torch.zeros((Nθ,), device=device, dtype=dtype)
    dθ = theta[1:] - theta[:-1]
    # trapezoid: interior gets (dθ_{i-1}+dθ_i)/2, endpoints get dθ0/2 and dθ_{N-2}/2
    wθ[0] = dθ[0] * 0.5
    wθ[-1] = dθ[-1] * 0.5
    for i in range(1, Nθ - 1):
        wθ[i] = 0.5 * (dθ[i - 1] + dθ[i])

    # φ weights (periodic grid often includes endpoint 2π; if you included both 0 and 2π, trapezoid will double-count)
    # Recommended: phi in [0,2π) without repeating endpoint. If your phi includes 2π, remove last sample.
    wφ = torch.zeros((Nφ,), device=device, dtype=dtype)
    dφ = phi[1:] - phi[:-1]
    if Nφ > 1:
        wφ[0] = dφ[0] * 0.5
        wφ[-1] = dφ[-1] * 0.5
        for j in range(1, Nφ - 1):
            wφ[j] = 0.5 * (dφ[j - 1] + dφ[j])
    else:
        wφ[0] = torch.tensor(2.0 * math.pi, device=device, dtype=dtype)

    W = (torch.sin(theta).unsqueeze(1) * wθ.unsqueeze(1)) * wφ.unsqueeze(0)  # (Nθ,Nφ)
    return W


# ============================================================
# 3) Cache: precompute harmonics and weights once
# ============================================================

@dataclass
class SpinWeightedYlmCache:
    """
    Cache for _{-2}Y_{lm}(θ,φ) and quadrature weights on the sphere grid.
    """
    theta: torch.Tensor          # (Nθ,)
    phi: torch.Tensor            # (Nφ,)
    lmax: int
    weights: torch.Tensor        # (Nθ,Nφ) float
    Ylm: Dict[Tuple[int,int], torch.Tensor]  # (l,m) -> complex (Nθ,Nφ)

    @staticmethod
    def build(theta: torch.Tensor, phi: torch.Tensor, lmax: int) -> "SpinWeightedYlmCache":
        device = theta.device
        ftype = theta.dtype

        # factorial table up to (lmax + 2) is enough for mp=2
        fact = _factorial_table(nmax=2 * lmax + 2, device=device, dtype=ftype)

        W = sphere_quadrature_weights(theta, phi)

        Ylm = {}
        for l in range(2, lmax + 1):
            for m in range(-l, l + 1):
                Ylm[(l, m)] = sYlm_s_minus2(theta, phi, l=l, m=m, fact=fact)

        return SpinWeightedYlmCache(theta=theta, phi=phi, lmax=lmax, weights=W, Ylm=Ylm)


# ============================================================
# 4) Projection: psi4 -> modes
# ============================================================

@torch.no_grad()
def project_psi4_modes(
    psi4: torch.Tensor,
    cache: SpinWeightedYlmCache,
    normalize: bool = True
) -> Dict[Tuple[int,int], complex]:
    """
    Compute mode coefficients:
        ψ_{lm} = ∫ ψ4(θ,φ) * conj(_{-2}Y_{lm}) dΩ

    Parameters
    ----------
    psi4 : complex tensor (Nθ,Nφ)
    cache : SpinWeightedYlmCache
    normalize : bool
        If True, assumes harmonics are orthonormal on the continuous sphere.
        With discrete quadrature, small deviations happen; this is usually OK.

    Returns
    -------
    modes : dict {(l,m): complex Python}
    """
    assert psi4.ndim == 2
    Nθ, Nφ = psi4.shape
    assert cache.weights.shape == (Nθ, Nφ)

    # Weight the integrand
    W = cache.weights

    modes: Dict[Tuple[int,int], complex] = {}
    for (l, m), Y in cache.Ylm.items():
        # integral ≈ Σ psi4 * conj(Y) * W
        val = torch.sum(psi4 * torch.conj(Y) * W.to(psi4.dtype))
        modes[(l, m)] = complex(val.item())

    return modes


# ============================================================
# 5) Convenience: compute ψ4 modes directly from state + sphere
# ============================================================

@torch.no_grad()
def psi4_modes_from_state(
    fd,
    state,
    sph,                       # SphereGrid from Block 8
    cache: SpinWeightedYlmCache,
) -> Dict[Tuple[int,int], complex]:
    """
    End-to-end:
    - psi4_on_sphere(fd,state,sph) from Block 8
    - project to modes
    """
    psi4 = psi4_on_sphere(fd, state, sph)   # complex (Nθ,Nφ)
    modes = project_psi4_modes(psi4, cache)
    return modes


# ============================================================
# 6) Example usage (single snapshot)
# ============================================================

# Use the same sphere resolution as Block 8 sph (must match theta/phi used by cache)
# IMPORTANT: if your `build_sphere_grid` includes phi endpoint 2π, you should remove last point
# to avoid duplicate endpoint in trapezoid rule. The Block 8 builder uses linspace(0,2π,Nphi)
# including endpoint by default. We'll fix it here safely.

# Rebuild sphere grid with phi excluding endpoint:
def build_sphere_grid_noendpoint(cfg, r: float, Ntheta: int = 32, Nphi: int = 64):
    device = cfg.grid.device
    dtype = cfg.grid.dtype
    theta = torch.linspace(1e-6, torch.pi - 1e-6, Ntheta, device=device, dtype=dtype)
    phi = torch.linspace(0.0, 2.0*torch.pi, Nphi + 1, device=device, dtype=dtype)[:-1]  # drop endpoint

    TH, PH = torch.meshgrid(theta, phi, indexing="ij")
    x = r * torch.sin(TH) * torch.cos(PH)
    y = r * torch.sin(TH) * torch.sin(PH)
    z = r * torch.cos(TH)

    g = cfg.grid
    x_norm = 2.0 * (x - g.x_min) / (g.x_max - g.x_min) - 1.0
    y_norm = 2.0 * (y - g.y_min) / (g.y_max - g.y_min) - 1.0
    z_norm = 2.0 * (z - g.z_min) / (g.z_max - g.z_min) - 1.0

    grid = torch.stack([x_norm, y_norm, z_norm], dim=-1).unsqueeze(0).unsqueeze(3)
    # Create a lightweight object mimicking SphereGrid
    class _Sphere:
        pass
    sph = _Sphere()
    sph.r = r
    sph.theta = theta
    sph.phi = phi
    sph.grid = grid
    sph.x, sph.y, sph.z = x, y, z
    return sph

# Build sphere and cache
Lx = float(cfg.grid.x_max - cfg.grid.x_min)
r_ex = 0.25 * Lx

sph9 = build_sphere_grid_noendpoint(cfg, r=r_ex, Ntheta=32, Nphi=64)
cache = SpinWeightedYlmCache.build(theta=sph9.theta, phi=sph9.phi, lmax=4)

state = state_bl if "state_bl" in globals() else state_flat

modes = psi4_modes_from_state(fd, state, sph9, cache)

print("Computed Ψ4 modes (l<=4):")
for (l, m) in sorted(modes.keys()):
    if l == 2 and abs(m) == 2:
        print(f"  (l={l}, m={m:+d}) : {modes[(l,m)]}")
```
